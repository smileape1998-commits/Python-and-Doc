## 🔄 บทที่ 5 — While Loop

```python
# โครงสร้าง (ต้องมีทั้ง 3 ส่วน)
count = 0           # ① ตัวแปรเริ่มต้น
while count < 5:    # ② เงื่อนไข
    print(count)
    count += 1      # ③ อัพเดท (อย่าลืม!)

# break — หยุดทันที
while True:
    ans = input("คำตอบ: ")
    if ans == "quit":
        break

# continue — ข้ามรอบนี้
while i < 10:
    i += 1
    if i % 2 == 0:
        continue   # ข้ามเลขคู่
    print(i)

# while-else (else รันเมื่อไม่มี break)
while condition:
    ...
else:
    print("จบ loop ปกติ")
```

## 🗂️ บทที่ 7 — Dictionary & Set

```python
# === DICTIONARY ===
# สร้าง
person = {"name": "Alice", "age": 30}

# เข้าถึง
person["name"]          # "Alice"  (Error ถ้าไม่มี key)
person.get("score")     # None     (ปลอดภัยกว่า)
person.get("score", 0)  # 0        (ค่า default)

# แก้ไข / เพิ่ม / ลบ
person["age"] = 31          # อัพเดท
person["email"] = "a@b.com" # เพิ่มใหม่
del person["email"]         # ลบ
val = person.pop("age")     # ลบ + คืนค่า

# วน loop
for k in person.keys():   ...
for v in person.values(): ...
for k, v in person.items(): ...

# === SET ===
s = {1, 2, 3}          # ค่าซ้ำไม่ได้
s = set([1,1,2,2,3])   # → {1,2,3}

s.add(4)               # เพิ่ม
s.discard(4)           # ลบ (ไม่ Error ถ้าไม่มี)
s.update([5, 6])       # เพิ่มหลายตัว

3 in s                 # ตรวจสอบ
set_a & set_b          # Intersection
set_a | set_b          # Union
set_a - set_b          # Difference
```

## 🛠️ บทที่ 8 — Function

```python
# สร้าง function
def greet(name, greeting="สวัสดี"):  # default argument
    return f"{greeting} {name}!"

# เรียกใช้
greet("Alice")                # positional
greet(name="Bob", greeting="Hello")  # keyword

# *args — รับ argument ไม่จำกัดจำนวน
def total(*args):
    return sum(args)
total(1, 2, 3, 4)  # 10

# **kwargs — รับ keyword argument ไม่จำกัด
def info(**kwargs):
    for k, v in kwargs.items():
        print(f"{k}: {v}")
info(name="Alice", age=30)

# return หลายค่า (คืนเป็น Tuple)
def divide(a, b):
    return a // b, a % b
q, r = divide(10, 3)

# Lambda (anonymous function)
square = lambda x: x ** 2
add    = lambda a, b: a + b

# filter / map
evens  = list(filter(lambda x: x % 2 == 0, numbers))
double = list(map(lambda x: x * 2, numbers))

# import module
import math
from math import sqrt, pi
from math import *
```

## 📁 บทที่ 11 — File I/O & Error Handling

```python
# === FILE I/O ===
# Mode: 'r'=อ่าน  'w'=เขียนทับ  'a'=ต่อท้าย  'rb'/'wb'=binary

# อ่าน
with open("file.txt", "r", encoding="utf-8") as f:
    content = f.read()          # อ่านทั้งหมด
    lines   = f.readlines()     # อ่านเป็น List ของบรรทัด
    for line in f: ...          # อ่านทีละบรรทัด

# เขียน
with open("file.txt", "w", encoding="utf-8") as f:
    f.write("Hello\n")
    f.writelines(["line1\n", "line2\n"])

# === ERROR HANDLING ===
try:
    result = 10 / 0
except ZeroDivisionError:
    print("หารด้วยศูนย์ไม่ได้")
except ValueError as e:
    print(f"ValueError: {e}")
except Exception as e:      # จับทุก error
    print(f"Error: {e}")
else:
    print("ไม่มี error")
finally:
    print("รันเสมอ ไม่ว่าจะเกิด error หรือไม่")

# Error ที่พบบ่อย
# ZeroDivisionError  — หารด้วย 0
# ValueError         — แปลงชนิดข้อมูลผิด
# TypeError          — ใช้ operation ผิดชนิด
# IndexError         — index เกินขอบ
# KeyError           — ไม่มี key ใน dict
# FileNotFoundError  — ไม่พบไฟล์
```

## 🔢 บทที่ 12 — NumPy

```python
import numpy as np

# สร้าง Array
a = np.array([1, 2, 3, 4, 5])       # 1D
b = np.array([[1,2,3],[4,5,6]])      # 2D
np.arange(0, 10, 2)                  # [0,2,4,6,8]
np.zeros((3, 3))                     # matrix ศูนย์
np.ones((2, 4))                      # matrix หนึ่ง
np.linspace(0, 1, 5)                 # [0, .25, .5, .75, 1.]

# ข้อมูล Array
a.shape    # (5,)       ขนาด
a.dtype    # int64      ชนิด
a.ndim     # 1          จำนวนมิติ

# เข้าถึง 2D
b[0, 1]    # 2          [row, col]
b[1, :]    # [4,5,6]    ทั้งแถว
b[:, 0]    # [1,4]      ทั้งคอลัมน์

# คำนวณ (Vectorized)
a + 10     a * 2     a ** 2
np.sum(a)       np.mean(a)
np.max(a)       np.min(a)
np.std(a)       np.sqrt(a)
```

## 📊 บทที่ 13 — CSV & Pandas

```python
# === CSV Module ===
import csv

# อ่าน CSV
with open("data.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)            # ['Alice', '30', '90']

# เขียน CSV
with open("out.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "age", "score"])  # header
    writer.writerow(["Alice", 30, 90])

# === Pandas ===
import pandas as pd

# สร้าง DataFrame
df = pd.DataFrame({"name": ["Alice","Bob"], "score": [90, 85]})

# อ่าน/เขียน CSV
df = pd.read_csv("data.csv")          # อ่าน
df.to_csv("output.csv", index=False)  # เขียน

# ดูข้อมูล
df.head()       # 5 แถวแรก
df.tail()       # 5 แถวสุดท้าย
df.shape        # (rows, cols)
df.info()       # ชนิดข้อมูลแต่ละคอลัมน์
df.describe()   # สถิติพื้นฐาน

# เลือกข้อมูล
df["score"]           # เลือก 1 คอลัมน์ → Series
df[["name","score"]]  # เลือกหลายคอลัมน์ → DataFrame
df.iloc[0]            # แถวที่ 0 (ด้วย index)
df.loc[0, "score"]    # แถว 0 คอลัมน์ score

# กรองข้อมูล
df[df["score"] >= 80]

# สถิติ
df["score"].mean()    df["score"].sum()
df["score"].max()     df["score"].min()
```
