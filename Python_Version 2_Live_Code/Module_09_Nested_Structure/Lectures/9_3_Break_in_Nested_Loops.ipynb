## 9.3 Breaking out of Inner Loops (การออกจากลูปชั้นในและการหลุดลูปซ้อน)

### อธิบายแนวคิด (Concept Explanation)
การควบคุมการหลุดออกจากวงวนที่ซ้อนกัน (Nested Loops) มีแนวคิดและเทคนิคสำคัญ 2 ประการดังนี้:

1. **การออกจากลูปซ้อนทุกลูปด้วยตัวแปรเสริม (Flag Variable):**
คำสั่ง `break` จะยุติการทำงานเฉพาะลูปชั้นในสุดที่คำสั่งนั้นอยู่เท่านั้น หากต้องการออกจากลูปชั้นนอกด้วย สามารถใช้ตัวแปรเสริม เช่น `found = True` แล้วสั่ง `if found: break` ในลูปชั้นนอก

2. **การแยกวงวนชั้นในไปเขียนเป็นฟังก์ชัน (Extracting Inner Loop to Function):**
นอกจากทำให้โค้ดอ่านง่ายขึ้นแล้ว การย้ายลูปย่อยไปไว้ในฟังก์ชันจะทำให้สามารถใช้คำสั่ง `return` เพื่อหยุดการทำงานและส่งค่ากลับได้ทันทีโดยไม่ต้องใช้ `break` ซ้อนกันหลายชั้น

### ข้อควรระวัง (Common Pitfalls)
1. **เข้าใจผิดว่า break ออกทุกลูป:** การสั่ง `break` ในลูปใน **ไม่หลุด** ออกจากลูปนอก ต้องใช้ Flag Variable หรือสั่ง `return` ในฟังก์ชันแทน
2. **การคืนค่าผิดตำแหน่ง:** เมื่อย้ายลูปไปอยู่ในฟังก์ชัน ควรตรวจสอบให้แน่ใจว่าตำแหน่งของคำสั่ง `return` อยู่นอกลูปหรือในลูปอย่างถูกต้อง
