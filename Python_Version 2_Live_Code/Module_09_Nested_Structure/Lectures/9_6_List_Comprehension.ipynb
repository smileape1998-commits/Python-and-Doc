## 9.6 List Comprehension (การสร้างลิสต์แบบกระชับ)

### อธิบายแนวคิด (Concept Explanation)
List Comprehension คือ วิธีการเขียนคำสั่งสร้างลิสต์ใหม่จากลิสต์เดิมอย่างสั้น กระชับ และมีประสิทธิภาพ ซึ่งเทียบเท่ากับการเขียนลูป `for` และ `append()` ในหลายบรรทัด

**รูปแบบไวยากรณ์หลัก 3 รูปแบบ:**
1. **แปลงทุกค่าในลิสต์:** `t = [2*e for e in x]`
2. **เลือกเฉพาะบางค่าในลิสต์:** `t = [e for e in x if e >= 0]`
3. **เลือกบางค่าและแปลงค่านั้น:** `t = [2*e for e in x if e >= 0]`

### ข้อควรระวัง (Common Pitfalls)
1. **ความซับซ้อนของโค้ด:** หากใส่เงื่อนไขซับซ้อนเกินไป จะทำให้อ่านยาก ควรใช้เฉพาะเมื่อโค้ดมีความตรงไปตรงมาในบรรทัดเดียว
2. **การลืมแปลงชนิดข้อมูล:** ในการรับอินพุต ต้องครอบฟังก์ชันแปลงประเภท เช่น `int(e)` หรือ `float(e)` ในส่วน expression หน้าคำสั่ง `for`
