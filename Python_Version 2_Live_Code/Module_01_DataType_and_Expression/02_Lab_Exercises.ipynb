## 01-03: ช่วงเวลา

โปรแกรมนี้แสดงข้อมูล นาที และวินาที ของเวลาเริ่มต้นและเวลาสิ้นสุด

จากนั้นแสดงข้อมูลช่วงเวลาเป็นจำนวน นาที และวินาที ระหว่างเวลาเริ่มต้นกับเวลาสิ้นสุด

โดยมีข้อจำกัดว่า เวลาเริ่มต้นจะต้องไม่เกินเวลาสิ้นสุด

เช่น ถ้าป้อนเลข 2 10 20 4 0 0 บรรทัดละจำนวน จะได้ผลลัพธ์คือ  
`1:49:40` แต่ถ้าป้อน 2 0 0 1 0 0 บรรทัดละจำนวน จะได้ผลลัพธ์คือ  
`-1:0:0` ซึ่งผิด ที่ถูกควรเป็น `23:0:0`

จงเปรียบเทียบและปรับให้ถูกต้องทั้งในกรณีที่เวลาเริ่มต้นน้อยกว่า เท่ากับ หรือมากกว่า เวลาสิ้นสุด

โดยมีข้อจำกัดอีกว่า เวลาสิ้นสุดต้องไม่เกินกว่า 23:59:59 ของวันเดียวกัน (กำหนดให้พิจารณาเฉพาะภายในหนึ่งวัน)

### ข้อแนะนำ

- ถ้าเวลาเริ่มต้น > เวลาสิ้นสุด ให้แสดงผลเป็นค่าลบ แต่สามารถแก้ไขได้โดยการคำนวณจาก h1 ถึง h2
- แนบมาที่ ๆ ที่ท่านไม่เข้าใจจะบอกว่า เวลาสิ้นสุดก่อนเวลาเริ่มต้น กรุณาใช้ตรรกะแก้ไข เช่น h1 = 1 ถึง h2 = 2 ก็พอท่านกับ h2 - h1 = 2 - 1 = 1 ชั่วโมง ซึ่งจะได้เวลานี้คือ h2 ≥ h1
- ถ้าดูจับกับ ให้ h1 = 2 และ h2 = 1 ข้างบอก 2 นาฬิกา ถึง 1 นาฬิกา ย้อมไม่มาท่านกับ 1 - 2 = -1 แต่มาท่านกับ 23 ซึ่นโมง
- ถ้าดูสี ๆ 23 = 24 + (-1) จึงพอบนี้พูดว่างบอก h1 ถึง h2 ไม่พ่าท่านกับ 24 + (h2 - h1) ก็จะได้เวลนมากกว่า h2 < h1
- ถ้าบรับเขาพูดให้เป็น (24 + (h2 - h1)) % 24 ก็จะบอกได้ว่า h2 ≥ h1 หรือ h2 < h1 (ลองดูอธิบ)
