## ข้อควรระวัง (Common Pitfalls)

1. **แปลง String ที่ไม่ใช่ตัวเลขเป็น int จะเกิด Error**
   - `int("Hello")` จะเกิด ValueError

2. **int() ตัดทศนิยมทิ้ง ไม่ปัดเศษ**
   - `int(3.9)` ได้ 3 ไม่ใช่ 4

3. **ไม่สามารถนำ String และ int มาบวกกันโดยตรง**
   - `"Age: " + 25` จะเกิด TypeError
   - ต้องแปลง: `"Age: " + str(25)`

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: แปลง String ที่ไม่ใช่ตัวเลขเป็น int
# number = int("Hello")  # ValueError

# ผิด: ไม่สามารถนำ String และ int มาบวกกันโดยตรง
# result = "Age: " + 25  # TypeError

# ถูก: แปลง int เป็น String ก่อน
result = "Age: " + str(25)
print(result)

# ถูก: แปลง String เป็น int ก่อนคำนวณ
num1 = "10"
num2 = "20"
total = int(num1) + int(num2)
print(total)  # ได้ 30 ไม่ใช่ "1020"
```

## สรุป

- `int()` แปลงเป็นจำนวนเต็ม (ตัดทศนิยมทิ้ง)
- `float()` แปลงเป็นทศนิยม
- `str()` แปลงเป็นข้อความ
- `bool()` แปลงเป็นค่าตรรกะ
- ต้องแปลงชนิดข้อมูลให้ตรงกันก่อนดำเนินการ
- int() ตัดทศนิยม ไม่ปัดเศษ
