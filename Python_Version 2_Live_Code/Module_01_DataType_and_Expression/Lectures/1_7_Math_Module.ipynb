## แนวคิด (Concept)

### Math Module

**Math Module** เป็นโมดูลสำหรับฟังก์ชันทางคณิตศาสตร์ขั้นสูง

**การใช้งาน:**
1. Import โมดูล: `import math`
2. เรียกใช้ฟังก์ชัน: `math.function_name()`

**ฟังก์ชันที่ใช้บ่อย:**
- `math.sqrt(x)`: รากที่สอง
- `math.pow(x, y)`: ยกกำลัง
- `math.floor(x)`: ปัดเศษลง
- `math.ceil(x)`: ปัดเศษขึ้น
- `math.sin(x)`, `math.cos(x)`: ฟังก์ชันตรีโกณมิติ (รับค่าเป็น radian)
- `math.radians(x)`: แปลงองศาเป็น radian
- `math.log(x, base)`: ลอการิทึม
- `math.pi`: ค่าคงที่ pi (3.14159...)
- `math.e`: ค่าคงที่ e (2.71828...)

## Built-in Functions (ฟังก์ชันพื้นฐาน)

Python มีฟังก์ชันพื้นฐานที่สามารถใช้งานได้ทันที **ไม่ต้อง import**

### ตารางฟังก์ชันที่ใช้บ่อย:

| ฟังก์ชัน | ความหมาย | ตัวอย่าง |
|---------|---------|----------|
| `abs(x)` | ค่าสัมบูรณ์ | `abs(-2)` → `2` |
| `round(x, n)` | ปัดเศษทศนิยม n ตำแหน่ง | `round(2/3, 2)` → `0.67` |
| `max(list)` | หาค่ามากสุด | `max([4,1,5,3])` → `5` |
| `min(list)` | หาค่าน้อยสุด | `min([4,1,5,3])` → `1` |
| `sum(list)` | ผลรวมของข้อมูลใน list | `sum([4,1,5,3])` → `13` |
| `len(list)` | จำนวนข้อมูลใน list | `len([4,1,5,3])` → `4` |
| `str(x)` | แปลงเป็น string | `str(1234)` → `"1234"` |
| `int(x)` | แปลงเป็น integer | `int("123")` → `123` |
| `float(x)` | แปลงเป็น float | `float("-123.4")` → `-123.4` |
| `input()` | รับข้อมูลจากผู้ใช้ | `k = input()` |
| `print()` | แสดงผลออกหน้าจอ | `print(a, b, c)` |

## ข้อควรระวัง (Common Pitfalls)

```python
# ผิด: ลืม import module
# print(math.sqrt(25))  # NameError: name 'math' is not defined

# ถูก: import ก่อนใช้งาน
import math
print(math.sqrt(25))

# ผิด: ใช้ sqrt() โดยตรงโดยไม่มี math.
# print(sqrt(25))  # NameError

# ถูก: ต้องใช้ math.sqrt()
print(math.sqrt(25))
```

## สรุป

### Math Module:
- ต้อง `import math` ก่อนใช้งาน
- เรียกใช้ฟังก์ชันด้วย `math.function_name()`
- ฟังก์ชันที่ใช้บ่อย: sqrt(), pow(), sin(), cos(), floor(), ceil()
- ค่าคงที่: math.pi, math.e

### Built-in Functions:
- ใช้ได้ทันที ไม่ต้อง import
- ฟังก์ชันที่ใช้บ่อย: abs(), round(), max(), min(), sum(), len()
- ฟังก์ชันแปลงชนิด: int(), float(), str(), bool()
- ฟังก์ชัน I/O: input(), print()
