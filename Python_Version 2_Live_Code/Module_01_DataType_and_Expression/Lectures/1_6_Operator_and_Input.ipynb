# 1.6 Operator และ input()

## แนวคิด (Concept)

### Arithmetic Operators (ตัวดำเนินการทางคณิตศาสตร์)

Python มีตัวดำเนินการทางคณิตศาสตร์หลายชนิด:

| Operator | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|----------|---------|----------|----------|
| `+` | บวก | `5 + 3` | `8` |
| `-` | ลบ | `5 - 3` | `2` |
| `*` | คูณ | `5 * 3` | `15` |
| `/` | หาร (ได้ทศนิยม) | `5 / 2` | `2.5` |
| `//` | หารแบบตัดทศนิยม | `5 // 2` | `2` |
| `%` | หารเอาเศษ (Modulo) | `5 % 2` | `1` |
| `**` | ยกกำลัง | `2 ** 3` | `8` |

### input() Function

- ใช้รับข้อมูลจากผู้ใช้ทางแป้นพิมพ์
- ค่าที่ได้จาก `input()` จะเป็น **String เสมอ**
- ต้องแปลงชนิดข้อมูลก่อนนำไปคำนวณ

## ตัวอย่างที่ 1: Arithmetic Operators พื้นฐาน

```python
# Arithmetic Operators ตัวอย่าง
a = 10
b = 3

print("บวก:", a + b)        # 13
print("ลบ:", a - b)         # 7
print("คูณ:", a * b)        # 30
print("หาร:", a / b)        # 3.333...
print("หารตัดทศนิยม:", a // b)  # 3
print("หารเอาเศษ:", a % b)     # 1
print("ยกกำลัง:", a ** b)      # 1000
```

## ตัวอย่างที่ 2: การใช้ input() รับข้อมูล

```python
# รับข้อมูล String
name = input("กรุณาใส่ชื่อ: ")
print("สวัสดี", name)

# รับข้อมูลตัวเลข (ต้องแปลงชนิด)
age_text = input("กรุณาใส่อายุ: ")
age = int(age_text)
print("อายุของคุณคือ", age, "ปี")

# แบบย่อ: แปลงชนิดพร้อมรับข้อมูล
height = float(input("กรุณาใส่ส่วนสูง (cm): "))
print("ส่วนสูง:", height, "cm")
```

## ตัวอย่างที่ 3: โปรแกรมคำนวณพื้นที่

```python
# โปรแกรมคำนวณพื้นที่สี่เหลี่ยมผืนผ้า
width = float(input("ความกว้าง: "))
height = float(input("ความสูง: "))

area = width * height
print("พื้นที่ =", area, "ตารางหน่วย")
```

## ตัวอย่างที่ 4: การแปลงและคำนวณข้อมูล

```python
# รับข้อมูลจาก input และแปลงเป็น float
a = float(input())

# เปลี่ยนจำนวนลบเป็นบวก จำนวนบวกเป็นลบ
a = -a
print("a =", a)

# ปัดเศษของ a ทิ้ง
b = int(a)
print("b =", b, "# ปัดเศษของ a ทิ้ง")

# ปัดเศษของ a แบบมีขึ้นหรือลง
c = int(a + 0.5)
print("c =", c, "# ปัดเศษของ a แบบมีขึ้นหรือลง")

# ค่าเลขจุดทศนิยมของ a
d = a - int(a)
print("d =", d, "# ค่าเลขจุดทศนิยมของ a")

# ค่าเลขจุดทศนิยมของ a (อีกวิธี)
d = a % 1
print("d =", d, "# ค่าเลขจุดทศนิยมของ a")
```

## ตัวอย่างที่ 5: การหาเลขหลักต่างๆ

```python
# สมมติ b = 1234 (จำนวนเต็ม)
b = 1234

# เลขหลักหน่วยของ b
e = b % 10
print("e =", e, "# เลขหลักหน่วยของ b")

# เลขหลักสิบของ b
f = b // 10 % 10
print("f =", f, "# เลขหลักสิบของ b")

# เลขหลักร้อยของ b
g = b // 100 % 10
print("g =", g, "# เลขหลักร้อยของ b")

# เลขหลักพันของ b
h = b // 1000 % 10
print("h =", h, "# เลขหลักพันของ b")
```

## ตัวอย่างที่ 6: การใช้ตัวแปรชื่อเดิม

```python
# h เก็บสตริง
h = input("ใส่ข้อมูล: ")
print("h เก็บสตริง:", type(h))

# h เก็บจำนวนเต็ม (ใช้ตัวแปรชื่อเดิมได้)
h = int(h)
print("h เก็บจำนวนเต็ม:", type(h))
```

## ข้อควรระวัง (Common Pitfalls)

```python
# ผิด: ลืมแปลงชนิดข้อมูล
# num1 = input("เลขที่ 1: ")  # ได้ String
# num2 = input("เลขที่ 2: ")  # ได้ String
# print(num1 + num2)  # ถ้าใส่ 5 และ 3 จะได้ "53" แทนที่จะเป็น 8

# ถูก: แปลงเป็น int หรือ float ก่อน
num1 = int(input("เลขที่ 1: "))
num2 = int(input("เลขที่ 2: "))
print("ผลบวก:", num1 + num2)
```

## สรุป

### Arithmetic Operators:
- `+`, `-`, `*` : บวก ลบ คูณ
- `/` : หาร (ได้ทศนิยม)
- `//` : หารตัดทศนิยม
- `%` : หารเอาเศษ
- `**` : ยกกำลัง

### input() Function:
- รับข้อมูลจากผู้ใช้
- ผลลัพธ์เป็น String เสมอ
- ต้องแปลงชนิดข้อมูลก่อนคำนวณ (int(), float())
