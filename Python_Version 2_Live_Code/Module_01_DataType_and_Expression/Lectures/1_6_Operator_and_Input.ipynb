## แนวคิด (Concept)

### Arithmetic Operators (ตัวดำเนินการทางคณิตศาสตร์)

Python มีตัวดำเนินการทางคณิตศาสตร์หลายชนิด:

| Operator | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|----------|---------|----------|----------|
| `+` | บวก | `5 + 3` | `8` |
| `-` | ลบ | `5 - 3` | `2` |
| `*` | คูณ | `5 * 3` | `15` |
| `/` | หาร (ได้ทศนิยม) | `5 / 2` | `2.5` |
| `//` | หารแบบตัดทศนิยม | `5 // 2` | `2` |
| `%` | หารเอาเศษ (Modulo) | `5 % 2` | `1` |
| `**` | ยกกำลัง | `2 ** 3` | `8` |

### input() Function

- ต้องแปลงชนิดข้อมูลก่อนนำไปคำนวณ

## สรุป

### Arithmetic Operators:
- `+`, `-`, `*` : บวก ลบ คูณ
- `/` : หาร (ได้ทศนิยม)
- `//` : หารตัดทศนิยม
- `%` : หารเอาเศษ
- `**` : ยกกำลัง

### input() Function:
- รับข้อมูลจากผู้ใช้
- ผลลัพธ์เป็น String เสมอ
- ต้องแปลงชนิดข้อมูลก่อนคำนวณ (int(), float())
