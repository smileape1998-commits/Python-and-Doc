- วงเล็บก่อนเสมอ
- คูณหารก่อนบวกลบ
- ถ้าลำดับเท่ากัน ทำจากซ้ายไปขวา

## ข้อควรระวัง (Common Pitfalls)

```python
# ผิด: คิดว่าบวกก่อนคูณ
wrong = 2 + 3 * 4  # คิดว่าได้ 20 แต่จริงๆ ได้ 14
print("ไม่ใช้วงเล็บ:", wrong)

# ถูก: ใช้วงเล็บเพื่อให้ชัดเจน
correct = (2 + 3) * 4  # ได้ 20 ตามต้องการ
print("ใช้วงเล็บ:", correct)

# ผิด: ยกกำลังทำจากซ้ายไปขวา
# 2 ** 3 ** 2 ≠ (2 ** 3) ** 2
result1 = 2 ** 3 ** 2  # 2⁹ = 512
result2 = (2 ** 3) ** 2  # 8² = 64
print("2 ** 3 ** 2 =", result1)
print("(2 ** 3) ** 2 =", result2)
```

## เคล็ดลับ (Tips)

1. **เมื่อสงสัย ใช้วงเล็บเสมอ** - ทำให้โค้ดอ่านง่ายและไม่เกิดข้อผิดพลาด
2. **เขียนให้ชัดเจน** - แม้ว่าจะรู้ลำดับ precedence ก็ตาม การใช้วงเล็บช่วยให้คนอื่นเข้าใจโค้ดได้ง่ายขึ้น
3. **แยกคำนวณซับซ้อน** - ถ้า expression ยาวและซับซ้อน แยกออกเป็นหลายบรรทัด

## สรุป

### ลำดับความสำคัญ (Operator Precedence):
1. `()` - วงเล็บ (ลำดับสูงสุด)
2. `**` - ยกกำลัง (ทำจากขวาไปซ้าย)
3. `*`, `/`, `//`, `%` - คูณ หาร (ทำจากซ้ายไปขวา)
4. `+`, `-` - บวก ลบ (ทำจากซ้ายไปขวา)

### หลักการจำง่ายๆ:
- **วงเล็บก่อนเสมอ**
- **คูณหารก่อนบวกลบ**
- **เมื่อสงสัย ใส่วงเล็บ**
