## แนวคิด (Concept)

โปรแกรม Python ประกอบด้วย:
- **Statement**: คำสั่งหนึ่งบรรทัด ที่ Python จะทำงานตามลำดับจากบนลงล่าง
- **Comment**: คำอธิบายที่ขึ้นต้นด้วย `#` ซึ่ง Python จะไม่ทำงาน ใช้เพื่ออธิบายโค้ด
- **Expression**: นิพจน์ที่คำนวณค่าได้ เช่น `2 + 3`

**หลักการสำคัญ:**
- Python อ่านและทำงานทีละบรรทัดจากบนลงล่าง
- การเว้นวรรค (indentation) มีความสำคัญมากใน Python

## อธิบายโครงสร้าง

จากตัวอย่างข้างต้น:

1. **Import statements**: นำเข้า library ที่ต้องการใช้งาน
   ```python
   import matplotlib.pyplot as plt
   import math
   ```

2. **Variable assignment**: กำหนดค่าให้ตัวแปร
   ```python
   n = 300
   ```

3. **Control flow (if statement)**: การควบคุมการทำงานตามเงื่อนไข
   ```python
   if n < 10:
       n = 10  # บรรทัดนี้เยื้อง 4 spaces
   ```

4. **Loop (for statement)**: การทำงานซ้ำ
   ```python
   for k in range(n):
       # โค้ดภายใน loop ต้องเยื้อง
   ```

5. **Function calls**: การเรียกใช้ฟังก์ชัน
   ```python
   plt.show()
   ```

## Comments (คอมเมนต์)

````python
# นี่คือคอมเมนต์บรรทัดเดียว (single-line comment)
# Python จะไม่ execute บรรทัดที่เริ่มด้วย #

print("Hello, World!")  # คำสั่งแสดงข้อความ

# คำสั่งหลายบรรทัด
print("บรรทัดที่ 1")
print("บรรทัดที่ 2")
print("บรรทัดที่ 3")
````

## ข้อควรระวัง (Common Pitfalls)

1. **การเยื้องไม่ถูกต้อง** - Python จะเกิด IndentationError ถ้าเยื้องผิด
2. **ผสม tabs และ spaces** - อาจทำให้เกิดปัญหา แนะนำให้ใช้ spaces อย่างเดียว (4 spaces)
3. **ลืมใส่ colon (:)** - เมื่อเริ่มต้น code block ใหม่ (เช่น if, for, def)
4. **Case sensitivity** - Python แยกตัวพิมพ์เล็กและใหญ่ (print และ Print ต่างกัน)

## สรุป

- Python อ่านโค้ดทีละบรรทัดจากบนลงล่าง
- การเยื้อง (indentation) ใช้กำหนดโครงสร้างของโค้ด
- Comment ขึ้นต้นด้วย # ใช้อธิบายโค้ด
- การเยื้องมาตรฐานคือ 4 spaces
