## แนวคิด (Concept)

Python มีชนิดข้อมูลพื้นฐาน 5 ชนิดหลัก:

1. **str** (String): ข้อความ
2. **int** (Integer): จำนวนเต็ม
3. **float** (Floating Point): จำนวนจริง (มีจุดทศนิยม)
4. **bool** (Boolean): ค่าจริงกับเท็จเท่านั้น (`True` กับ `False`)
5. **list**: ชุดข้อมูล เก็บเรียงเป็นลำดับ

**ใช้ `type()` เพื่อตรวจสอบชนิดข้อมูล**

## ข้อควรระวัง (Common Pitfalls)

1. **Boolean ต้องขึ้นต้นด้วยตัวพิมพ์ใหญ่**
   - ผิด: `true`, `false`
   - ถูก: `True`, `False`

2. **String ที่มีตัวเลขไม่ใช่ตัวเลข**
   - `"123"` เป็น str ไม่ใช่ int
   - ต้องแปลงก่อนใช้งานทางคณิตศาสตร์

3. **int และ float ต่างกัน**
   - `5` เป็น int
   - `5.0` เป็น float

## สรุป

- Python มี 5 ชนิดข้อมูลพื้นฐาน: str, int, float, bool, list
- ใช้ `type()` เพื่อตรวจสอบชนิดข้อมูล
- Boolean มีแค่ 2 ค่า: `True` และ `False` (ต้องขึ้นต้นด้วยตัวพิมพ์ใหญ่)
- List ใช้เก็บชุดข้อมูลหลายค่า เข้าถึงด้วย index เริ่มจาก 0
