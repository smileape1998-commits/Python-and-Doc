## 4.3 Bisection Search (การค้นหาแบบทวิภาค)

### อธิบายแนวคิด (Concept Explanation)
วิธี Bisection Search เป็นอัลกอริทึมค้นหาคำตอบตัวเลขทศนิยมในช่วงที่ต้องการ เช่น การหาค่ารากที่สอง (Square Root) ของ $x$ โดยการแบ่งครึ่งช่วงการค้นหา `[low, high]` ซ้ำๆ จนกว่าช่วงคำตอบจะแคบลงกว่าค่าความคลาดเคลื่อนที่ยอมรับได้ (`epsilon`)

### ตัวอย่าง 4.3: การหาค่ารากที่สองของ x ด้วย Bisection Search
