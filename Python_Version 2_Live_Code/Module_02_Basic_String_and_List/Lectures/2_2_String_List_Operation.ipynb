## ข้อควรระวัง (Common Pitfalls)

1. **ไม่สามารถต่อ String กับ Number โดยตรง**
   - `"Age: " + 25` จะเกิด TypeError
   - ต้องแปลง: `"Age: " + str(25)`

2. **การต่อ List ไม่เหมือนการ append**
   - `[1, 2] + [3]` ได้ `[1, 2, 3]`
   - `[1, 2] + 3` จะเกิด TypeError

3. **การคูณ List จะสร้าง reference เดียวกัน**
   - `[[0]] * 3` อาจทำให้เกิดปัญหาถ้าแก้ไขข้างใน

4. **Index เริ่มที่ 0**
   - วันที่ 1 ต้องใช้ `days[0]`
   - อย่าลืมลบ 1 เมื่อใช้ input

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: ต่อ String กับ Number
# message = "อายุ " + 25  # TypeError

# ถูก:
message = "อายุ " + str(25)
print(message)

# หรือใช้ comma ใน print()
print("อายุ", 25)

# ผิด: ต่อ List กับตัวเลข
# numbers = [1, 2] + 3  # TypeError

# ถูก: ใส่ในวงเล็บเหลี่ยม
numbers = [1, 2] + [3]
print(numbers)
```

## สรุป

### การต่อ (Concatenation):
- ใช้ `+` เพื่อต่อ String หรือ List
- String + String = String ใหม่
- List + List = List ใหม่

### การทำซ้ำ (Repetition):
- ใช้ `*` เพื่อทำซ้ำ
- `"Ha" * 3` ได้ `"HaHaHa"`
- `[0] * 5` ได้ `[0, 0, 0, 0, 0]`

### len() Function:
- หาความยาวของ String (จำนวนตัวอักษร)
- หาจำนวนสมาชิกของ List

### การประยุกต์ใช้:
- ใช้ List เก็บข้อมูลที่มีลำดับ
- ใช้ Index เข้าถึงข้อมูล
- รวม Index, Slice และ Concatenation เพื่อจัดการข้อมูล
