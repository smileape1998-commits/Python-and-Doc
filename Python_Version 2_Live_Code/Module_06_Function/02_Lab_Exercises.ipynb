## โจทย์แบบฝึกหัดที่ 2: 06-04 การบวกเลขฐานสอง

### อธิบายโจทย์
ภาษา Python มีฟังก์ชันมาตรฐานสำหรับแปลงและประมวลผลเลขฐาน ได้แก่:
- `bin(x)`: แปลงจำนวนเต็ม `x` เป็นสตริงเลขฐานสองขึ้นต้นด้วย `"0b"`
- `int(x, base)`: แปลงสตริง `x` ในฐาน `base` ให้เป็นจำนวนเต็มฐานสิบ

จงเขียนโปรแกรมรับจำนวนเต็มในรูปฐานสอง 2 จำนวนที่คั่นด้วยช่องว่าง เพื่อหาผลบวกและแสดงผลลัพธ์ในรูปเลขฐานสอง (โดยไม่ติดสัญลักษณ์ `"0b"`)

### ตัวอย่างการทำงาน
**ตัวอย่างที่ 1:**
- Input: `100101 11001`
- Output: `111110`

**ตัวอย่างที่ 2:**
- Input: `000000 00000`
- Output: `0`

**ตัวอย่างที่ 3:**
- Input: `1111111111111111111111111 1`
- Output: `10000000000000000000000000`

````python
# 06-04: การบวกเลขฐานสอง

# รับสตริงเลขฐานสอง 2 ชุด คั่นด้วยช่องว่าง
binary1_str, binary2_str = input().split()

# แปลงจากสตริงเลขฐานสองเป็นจำนวนเต็มฐานสิบ
num1 = int(binary1_str, 2)
num2 = int(binary2_str, 2)

# คำนวณผลบวก
sum_val = num1 + num2

# แปลงผลบวกเป็นสตริงเลขฐานสอง และตัดสัญลักษณ์ "0b" ออกด้านหน้า
result_binary = bin(sum_val)[2:]

# แสดงผลลัพธ์
print(result_binary)
````
