### 6. ข้อผิดพลาดที่พบบ่อยในการนิยามฟังก์ชัน (Common Pitfalls)
1. **การแปลงข้อมูลซ้ำในฟังก์ชันโดยไม่จำเป็น**: เมื่อพารามิเตอร์ถูกส่งเข้ามาแล้ว มีค่าพร้อมใช้งานทันที ไม่ต้องใส่ `input()` หรือรับค่าใหม่ซ้อนในฟังก์ชัน
2. **การลืมระบุคำสั่ง `return`**: หากไม่ได้ใส่คำสั่ง `return` ฟังก์ชันจะคืนค่า `None` โดยอัตโนมัติ
3. **ขอบเขตตัวแปร (Scope Errors)**: การอ้างอิงตัวแปรในฟังก์ชันอื่น หรือเข้าใจผิดว่าตัวแปร Local ในฟังก์ชันจะสามารถนำออกมาใช้ข้างนอกได้
