### 4. ข้อผิดพลาดที่พบบ่อย (Common Pitfalls)
1. **การแก้ไขข้อมูลโดยไม่เจตนา (Side Effects)**: เขียนฟังก์ชันแก้ไขสมาชิกในลิสต์โดยผู้ใช้ไม่ทราบ ส่งผลให้ข้อมูลต้นฉบับเสียหาย
2. **ความสับสนระหว่างการแก้ไขสมาชิกกับการกำหนดค่าตัวแปรใหม่**: การใช้ `data[k] = val` จะแก้ไขลิสต์เดิม แต่การใช้ `data = [1, 2, 3]` เป็นเพียงการสร้างตัวแปรใหม่ใน Local Scope ซึ่งไม่กระทบลิสต์เดิมภายนอก
