### อธิบายแนวคิด (Concept Explanation)
ในหลายปัญหาทางโปรแกรมมิ่ง เราจำเป็นต้องเปรียบเทียบหรือประมวลผลสมาชิกลิสต์คู่ที่อยู่ติดกัน เช่น เปรียบเทียบ `x[i]` กับ `x[i+1]`
