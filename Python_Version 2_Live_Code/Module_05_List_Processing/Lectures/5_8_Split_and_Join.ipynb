## 5.8 Split & Join (การแยกและรวมสตริงกับลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การแปลงข้อมูลระหว่าง สตริง (String) และ ลิสต์ของสตริง (List of Strings) เป็นทักษะสำคัญในการจัดการข้อมูลข้อความ:

1. **`str.split(sep)`:** แปลงสตริงให้กลายเป็นลิสต์ โดยใช้ตัวคั่น `sep`
   - `input().split()` : หากไม่ระบุตัวคั่น จะตัดด้วยช่องว่างทุกชนิด (Space, Tab, Newline) และจะข้ามช่องว่างซ้ำๆ ให้อัตโนมัติ
   - `s.split("/")` หรือ `s.split(",")` : ตัดด้วยอักขระตัวคั่นตามที่ระบุ
   - **Unpacking:** สามารถกระจายผลลัพธ์ใส่หลายตัวแปรได้ เช่น `d, m, y = s.split("/")`

2. **`sep.join(list)`:** รวมสมาชิกในลิสต์ของสตริงเข้าด้วยกันให้เป็นสตริงเดียว โดยคั่นด้วย `sep`
   - เช่น `", ".join(list)`, `" ".join(list)`, หรือ `"x".join(list)`

### ข้อควรระวัง (Common Pitfalls)
- **ชนิดข้อมูลในลิสต์ขณะใช้ `join()`:** เมธอด `join()` จะต้องใช้กับลิสต์ที่สมาชิกทุกตัวเป็น **สตริง (String)** เท่านั้น หากสมาชิกเป็นตัวเลข ต้องแปลงเป็นสตริงก่อนด้วย `str(k)` เช่น `results.append(str(k))` หรือ `" ".join([str(e) for e in list])`
