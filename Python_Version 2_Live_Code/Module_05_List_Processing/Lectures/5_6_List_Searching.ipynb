## 5.6 List Searching (การค้นหาข้อมูลในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การจัดโครงสร้างข้อมูลเพื่อรองรับการค้นหาข้ามชุดข้อมูล มี 3 รูปแบบหลัก:

1. **Single Flat List (ลิสต์เดี่ยวผสม):** เก็บข้อมูลคู่เรียงต่อกันในลิสต์เดียว เช่น `['KTB', 18.9, 'BBL', 176.5, ...]`
2. **Parallel Lists (ลิสต์ขนาน):** แยกข้อมูลออกเป็น 2 ลิสต์ที่มีดัชนีตำแหน่งสอดคล้องกัน เช่น `eng = ['BKK', 'CMI']` และ `th = ['กทม', 'ชม']`
3. **Nested List (ลิสต์ซ้อนลิสต์ / 2D List):** เก็บข้อมูลแต่ละคู่เป็นลิสต์ย่อยภายในลิสต์หลัก เช่น `[['BKK', 'กทม'], ['CMI', 'ชม']]`
