# 2.1 Index และ Slice

## แนวคิด (Concept)

### Index (การเข้าถึงข้อมูลด้วยตำแหน่ง)

- **Index** คือตำแหน่งของข้อมูลใน String หรือ List
- เริ่มนับจาก `0` สำหรับตำแหน่งแรก
- ใช้ `[index]` เพื่อเข้าถึงข้อมูล
- Index ติดลบเริ่มจากท้ายสุด (`-1` คือตัวสุดท้าย)

**รูปแบบ:**
- `string[index]` หรือ `list[index]`

### Slice (การตัดข้อมูล)

- **Slice** ใช้ตัดข้อมูลช่วงหนึ่งจาก String หรือ List
- รูปแบบ: `[start:stop:step]`
  - `start`: ตำแหน่งเริ่มต้น (รวม)
  - `stop`: ตำแหน่งสิ้นสุด (ไม่รวม)
  - `step`: ระยะห่าง (ค่าเริ่มต้น = 1)

**กรณีพิเศษ:**
- `[:stop]`: เริ่มจากตัวแรกถึง stop
- `[start:]`: เริ่มจาก start ถึงตัวสุดท้าย
- `[:]`: คัดลอกทั้งหมด
- `[::step]`: ทั้งหมด โดยกระโดดทุก step

### len() Function

- ใช้หาความยาว (จำนวนตัวอักษรใน String หรือจำนวนสมาชิกใน List)
- รูปแบบ: `len(string)` หรือ `len(list)`

## ตัวอย่างที่ 1: Index กับ String

```python
# ตัวอย่าง Index กับ String
#       01234567890123456
text = "Python Programming"

print("ตัวแรก:", text[0])   # P
print("ตัวที่ 2:", text[1])   # y
print("ตัวที่ 7:", text[6])   # n

# Index ติดลบ (นับจากท้าย)
print("ตัวสุดท้าย:", text[-1])  # g
print("ตัวรองสุดท้าย:", text[-2])  # n
```

## ตัวอย่างที่ 2: Index กับ List

```python
# ตัวอย่าง Index กับ List
numbers = [10, 20, 30, 40, 50]
#          0   1   2   3   4

print("ตัวแรก:", numbers[0])   # 10
print("ตัวที่ 3:", numbers[2])   # 30
print("ตัวสุดท้าย:", numbers[4])   # 50

# Index ติดลบ
print("ตัวสุดท้าย:", numbers[-1])  # 50
print("ตัวที่ 3 จากท้าย:", numbers[-3])  # 30
```

## ตัวอย่างที่ 3: Slice กับ String พื้นฐาน

```python
# ตัวอย่าง Slice กับ String
text = "Python Programming"

# [start:stop]
print(text[0:6])    # Python (ตำแหน่ง 0-5)
print(text[7:18])   # Programming (ตำแหน่ง 7-17)

# ละเว้น start หรือ stop
print(text[:6])     # Python (เริ่มตั้งแต่ต้น)
print(text[7:])     # Programming (ไปจนถึงท้าย)
print(text[:])      # Python Programming (ทั้งหมด)
```

## ตัวอย่างที่ 4: Slice กับ Step

```python
text = "Python Programming"

# ใช้ step
print(text[::2])    # Pto rgamn (เอาทุก 2 ตัว)
print(text[::3])    # Ph gmi (เอาทุก 3 ตัว)

# กลับข้อความด้วย step = -1
print(text[::-1])   # gnimmargorP nohtyP

# ใช้ทั้ง start, stop, step
print(text[0:10:2]) # Pto r (ตำแหน่ง 0-9 ทุก 2 ตัว)
```

## ตัวอย่างที่ 5: Slice กับ List

```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

# [start:stop]
print(numbers[0:3])    # [10, 20, 30]
print(numbers[5:8])    # [60, 70, 80]

# ละเว้น start หรือ stop
print(numbers[:5])     # [10, 20, 30, 40, 50]
print(numbers[5:])     # [60, 70, 80, 90]

# ใช้ step
print(numbers[::2])    # [10, 30, 50, 70, 90]
print(numbers[::-1])   # [90, 80, 70, 60, 50, 40, 30, 20, 10]
```

## ตัวอย่างที่ 6: การใช้งานร่วมกัน

```python
#       01234
s = "Hello"; t = "Python"
x = [1, 3, 5, 7]; y = [5, 6, 7, 8]
r = s + t
z = x + y
a = 10 * [0]

# length (ความยาว)
print("len(s) =", len(s), "len(r) =", len(r))
print("len(x) =", len(x), "len(z) =", len(z))

# index
print("s[1] =", s[1], "x[2] =", x[2])
print("s[-1] =", s[-1], "s[-2] =", s[-2])
print("x[-1] =", x[-1], "x[-2] =", x[-2])

# slice
print("s[0:3:1] =", s[0:3:1])  # Hel
```

## ตัวอย่างที่ 7: List เปลี่ยนค่าได้ String ไม่ได้

```python
# List สามารถเปลี่ยนค่าได้
x = [1, 3, 5, 7]
print("ก่อนเปลี่ยน:", x)

x[3] = 99
print("หลังเปลี่ยน:", x)  # [1, 3, 5, 99]

# String เปลี่ยนค่าไม่ได้
# s = "Hello"
# s[0] = "h"  # TypeError: 'str' object does not support item assignment
```

## ข้อควรระวัง (Common Pitfalls)

1. **Index เกินขอบเขต**
   - `text = "Python"` มี index 0-5
   - `text[10]` จะเกิด IndexError

2. **Slice stop ไม่รวมตำแหน่งสุดท้าย**
   - `text[0:3]` จะได้ตำแหน่ง 0, 1, 2 (ไม่รวม 3)

3. **String เปลี่ยนค่าไม่ได้**
   - `s[0] = "X"` จะเกิด TypeError
   - ต้องสร้าง String ใหม่แทน

4. **Index ติดลบ**
   - `-1` คือตัวสุดท้าย
   - `-2` คือตัวรองสุดท้าย

## สรุป

### Index:
- เริ่มนับจาก 0
- Index ติดลบนับจากท้าย (-1 = ตัวสุดท้าย)
- ใช้ `[index]` เข้าถึงข้อมูล

### Slice:
- รูปแบบ: `[start:stop:step]`
- start รวม, stop ไม่รวม
- step = -1 เพื่อกลับลำดับ

### ความแตกต่าง:
- List เปลี่ยนค่าได้ (mutable)
- String เปลี่ยนค่าไม่ได้ (immutable)
