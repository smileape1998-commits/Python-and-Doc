# 2.3 การแยก String (String Splitting)

## แนวคิด (Concept)

### split() Method

- ใช้แยก String เป็น List ตามตัวคั่น (delimiter)
- รูปแบบ: `string.split(separator)`
- ถ้าไม่ระบุ separator จะแยกด้วยช่องว่าง (whitespace)
- ผลลัพธ์เป็น List ของ String

### join() Method

- ใช้รวม List เป็น String
- รูปแบบ: `separator.join(list)`
- separator คือตัวคั่นที่จะใช้ต่อระหว่างสมาชิก

## ตัวอย่างที่ 1: split() พื้นฐาน - แยกด้วยช่องว่าง

```python
# แยกด้วยช่องว่าง (ค่าเริ่มต้น)
sentence = "Python is awesome"
words = sentence.split()
print(words)  # ["Python", "is", "awesome"]
print(type(words))  # <class 'list'>

# แยกด้วยช่องว่างหลายตัว (split() จะจัดการให้อัตโนมัติ)
text = "a   b   c"
result = text.split()
print(result)  # ["a", "b", "c"]
```

## ตัวอย่างที่ 2: split() ด้วยตัวคั่นที่กำหนด

```python
# แยกด้วยเครื่องหมายจุลภาค
data = "apple,banana,orange"
fruits = data.split(",")
print(fruits)  # ["apple", "banana", "orange"]

# แยกด้วยเครื่องหมาย -
date = "2024-08-14"
parts = date.split("-")
print(parts)  # ["2024", "08", "14"]

# แยกด้วยเครื่องหมาย :
time = "10:30:45"
time_parts = time.split(":")
print(time_parts)  # ["10", "30", "45"]
```

## ตัวอย่างที่ 3: ความแตกต่างระหว่าง split() และ split(" ")

```python
# split() - แยกด้วย whitespace ทุกชนิด และตัดช่องว่างหลายตัว
result1 = "a   b".split()
print("split():", result1)  # ["a", "b"]

# split(" ") - แยกด้วยช่องว่างเดียว
result2 = "a   b".split(" ")
print("split(' '):", result2)  # ["a", "", "", "b"]

# ตัวอย่างเพิ่มเติม
result3 = "a,,,b".split(",")
print("a,,,b split(','):", result3)  # ["a", "", "", "b"]

result4 = "a,,,b".split(",,")
print("a,,,b split(',,')", result4)  # ["a", ",b"]
```

## ตัวอย่างที่ 4: การใช้งานจริง - แยกวันที่

```python
# รับวันที่รูปแบบ d/m/y
date_input = input("ใส่วันที่ (d/m/y): ")  # เช่น 1/12/2562

# แยกด้วย /
date_parts = date_input.split("/")

# แปลงเป็น int
day = int(date_parts[0])
month = int(date_parts[1])
year = int(date_parts[2])

print("วันที่:", day)
print("เดือน:", month)
print("ปี:", year)
```

## ตัวอย่างที่ 5: การใช้งานจริง - แยกเวลา

```python
# รับเวลารูปแบบ h:m:s
time_input = input("ใส่เวลา (h:m:s): ")  # เช่น 10:30:45

# แยกด้วย :
time_parts = time_input.split(":")

# แปลงเป็น int
hours = int(time_parts[0])
minutes = int(time_parts[1])
seconds = int(time_parts[2])

print("ชั่วโมง:", hours)
print("นาที:", minutes)
print("วินาที:", seconds)
```

## ตัวอย่างที่ 6: การหาค่าเฉลี่ยจากข้อมูลที่คั่นด้วย comma

```python
# รับตัวเลขคั่นด้วย comma และช่องว่าง
# เช่น 24, 35, 43, 67
numbers_input = input("ใส่ตัวเลข 4 จำนวน (คั่นด้วย , ): ")

# แยกด้วย comma และช่องว่าง
numbers_str = numbers_input.split(", ")

# แปลงเป็น int และหาผลรวม
total = 0
for num_str in numbers_str:
    total += int(num_str)

# หาค่าเฉลี่ย
average = total / len(numbers_str)
print("ค่าเฉลี่ย:", average)
```

## ตัวอย่างที่ 7: join() - รวม List เป็น String

```python
# รวมด้วยช่องว่าง
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(sentence)  # "Python is awesome"

# รวมด้วยเครื่องหมายจุลภาค
fruits = ["apple", "banana", "orange"]
csv_data = ",".join(fruits)
print(csv_data)  # "apple,banana,orange"

# รวมด้วย -
date_parts = ["2024", "08", "14"]
date_string = "-".join(date_parts)
print(date_string)  # "2024-08-14"

# รวมโดยไม่มีตัวคั่น
letters = ["H", "e", "l", "l", "o"]
word = "".join(letters)
print(word)  # "Hello"
```

## ตัวอย่างที่ 8: split() และ join() ร่วมกัน

```python
# แยกแล้วรวมใหม่
sentence = "Python is awesome"
words = sentence.split()
new_sentence = "-".join(words)
print(new_sentence)  # "Python-is-awesome"

# แปลง CSV เป็นรูปแบบอื่น
csv = "apple,banana,orange"
items = csv.split(",")
formatted = " | ".join(items)
print(formatted)  # "apple | banana | orange"
```

## ข้อควรระวัง (Common Pitfalls)

1. **ผลลัพธ์ของ split() เป็น String เสมอ**
   - `"10,20,30".split(",")` ได้ `["10", "20", "30"]` (ไม่ใช่ตัวเลข)
   - ต้องแปลงเป็น int หรือ float ก่อนคำนวณ

2. **split() กับ split(" ") ต่างกัน**
   - `split()` ตัดช่องว่างหลายตัวออก
   - `split(" ")` แยกทุกช่องว่าง (อาจได้ String ว่าง)

3. **join() ต้องการ List ของ String**
   - `",".join([1, 2, 3])` จะเกิด TypeError
   - ต้องแปลง: `",".join([str(1), str(2), str(3)])`

4. **Index เกินขอบเขต**
   - ตรวจสอบจำนวนสมาชิกก่อนใช้ Index
   - ใช้ len() เช็คความยาว

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: ลืมแปลงชนิดข้อมูล
numbers_str = "10,20,30".split(",")
# total = sum(numbers_str)  # TypeError: unsupported operand type(s)

# ถูก: แปลงเป็น int ก่อน
numbers = [int(x) for x in numbers_str]
total = sum(numbers)
print("ผลรวม:", total)

# ผิด: join() กับ List ของตัวเลข
# result = ",".join([1, 2, 3])  # TypeError

# ถูก: แปลงเป็น String ก่อน
result = ",".join([str(1), str(2), str(3)])
print(result)  # "1,2,3"
```

## สรุป

### split() Method:
- แยก String เป็น List
- `string.split()` - แยกด้วย whitespace
- `string.split(sep)` - แยกด้วย separator ที่กำหนด
- ผลลัพธ์เป็น List ของ String

### join() Method:
- รวม List เป็น String
- `separator.join(list)` - ใช้ separator คั่นระหว่างสมาชิก
- List ต้องเป็น String ทั้งหมด

### การใช้งาน:
- แยกข้อมูลที่มาจาก input (เช่น CSV, วันที่, เวลา)
- ต้องแปลงชนิดข้อมูลหลัง split() ถ้าต้องการคำนวณ
- ใช้ join() สร้างข้อความจาก List
