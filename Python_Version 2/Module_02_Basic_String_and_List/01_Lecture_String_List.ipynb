## 2.1 Index and Slice

### Index (การเข้าถึงข้อมูลด้วยตำแหน่ง)

- **Index** คือตำแหน่งของข้อมูลใน String หรือ List
- เริ่มนับจาก `0` สำหรับตำแหน่งแรก
- ใช้ `[index]` เพื่อเข้าถึงข้อมูล
- Index ติดลบเริ่มจากท้ายสุด (`-1` คือตัวสุดท้าย)

**รูปแบบ:**
- `string[index]` หรือ `list[index]`

```python
# ตัวอย่าง Index กับ String
text = "Python"

print(text[0])   # P (ตัวแรก)
print(text[1])   # y (ตัวที่ 2)
print(text[5])   # n (ตัวสุดท้าย)

# Index ติดลบ (นับจากท้าย)
print(text[-1])  # n (ตัวสุดท้าย)
print(text[-2])  # o (ตัวรองสุดท้าย)
```

```python
# ตัวอย่าง Index กับ List
numbers = [10, 20, 30, 40, 50]

print(numbers[0])   # 10 (ตัวแรก)
print(numbers[2])   # 30 (ตัวที่ 3)
print(numbers[4])   # 50 (ตัวสุดท้าย)

# Index ติดลบ
print(numbers[-1])  # 50 (ตัวสุดท้าย)
print(numbers[-3])  # 30 (ตัวที่ 3 จากท้าย)
```

### Slice (การตัดข้อมูล)

- **Slice** ใช้ตัดข้อมูลช่วงหนึ่งจาก String หรือ List
- รูปแบบ: `[start:stop:step]`
  - `start`: ตำแหน่งเริ่มต้น (รวม)
  - `stop`: ตำแหน่งสิ้นสุด (ไม่รวม)
  - `step`: ระยะห่าง (ค่าเริ่มต้น = 1)

**กรณีพิเศษ:**
- `[:stop]`: เริ่มจากตัวแรกถึง stop
- `[start:]`: เริ่มจาก start ถึงตัวสุดท้าย
- `[:]`: คัดลอกทั้งหมด
- `[::step]`: ทั้งหมด โดยกระโดดทุก step

```python
# ตัวอย่าง Slice กับ String
text = "Python Programming"

# [start:stop]
print(text[0:6])    # Python
print(text[7:18])   # Programming

# ละเว้น start หรือ stop
print(text[:6])     # Python (เริ่มตั้งแต่ต้น)
print(text[7:])     # Programming (ไปจนถึงท้าย)

# ใช้ step
print(text[::2])    # Pto rgamn (เอาทุก 2 ตัว)
print(text[::-1])   # gnimmargorP nohtyP (กลับข้อความ)
```

## 2.2 Basic String & List Operation

### String & List Concatenation (การต่อ String และ List)

- ใช้ `+` เพื่อต่อ String หรือ List
- สามารถใช้ Index และ Slice ร่วมกับการต่อได้

```python
s = "ABCDEF"
x = [1, 3, 5, 7, 9]

# index & slice
t = s[0] + s[-1]
y = x[:2] + x[-2:]

print("t =", t)  # AF
print("y =", y)  # [1, 3, 7, 9]
```

## 2.3 String Splitting

### split() Method

- ใช้แยก String เป็น List ตามตัวคั่น (delimiter)
- รูปแบบ: `string.split(separator)`
- ถ้าไม่ระบุ separator จะแยกด้วยช่องว่าง

```python
# split() ตัวอย่าง

# แยกด้วยช่องว่าง (ค่าเริ่มต้น)
sentence = "Python is awesome"
words = sentence.split()
print(words)  # ["Python", "is", "awesome"]

# แยกด้วยเครื่องหมายจุลภาค
data = "apple,banana,orange"
fruits = data.split(",")
print(fruits)  # ["apple", "banana", "orange"]
```

### join() Method

- ใช้รวม List เป็น String
- รูปแบบ: `separator.join(list)`

```python
# join() ตัวอย่าง

# รวมด้วยช่องว่าง
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(sentence)  # "Python is awesome"

# รวมด้วยเครื่องหมายจุลภาค
fruits = ["apple", "banana", "orange"]
data = ",".join(fruits)
print(data)  # "apple,banana,orange"
```

### อธิบาย split() แบบละเอียด

#### x = s.split()
- แยกสตริง s ออกเป็นส่วนๆ ได้เป็นลิสต์ (แยกด้วยช่องว่าง)
- `s = "11    2 33"`
- `s.split()` ได้ `["11", "2", "33"]`

#### x = s.split(sep)
- แยกสตริง s ออกเป็นส่วนๆ ได้เป็นลิสต์ ใช้ sep เป็นตัวแยก
- `s = "11:2B 33"`
- `s.split(":B")` ได้ `["11", "2", " 33"]`

### ตัวอย่างที่น่าสนใจ

```python
# ตัวอย่างการใช้ split() แบบต่างๆ

# 1. แยกด้วย comma
result1 = "a,,,b".split(",")
print(result1)  # ["a", "", "", "b"]

# 2. แยกด้วยช่องว่าง
result2 = "a   b".split(" ")
print(result2)  # ["a", "", "", "b"]

# 3. แยกด้วย split() ไม่มี argument (แยกด้วยช่องว่างอัตโนมัติ)
result3 = "a   b".split()
print(result3)  # ["a", "b"]

# 4. แยกด้วย comma แล้วเจอ comma ซ้อน
result4 = "a,,,b".split(",,")
print(result4)  # ["a", ",b"]
```
