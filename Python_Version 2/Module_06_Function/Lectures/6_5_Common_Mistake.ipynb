## 6.5 Common Mistake (ข้อผิดพลาดที่พบบ่อยในการใช้งานฟังก์ชัน)

### 1. ข้อควรระวัง: อย่ากำหนดค่าตัวแปรพารามิเตอร์ใหม่โดยไม่แก้ไขสมาชิก
เมื่อส่งตัวแปรเข้ามาเป็นพารามิเตอร์ `x` การกำหนดค่าใหม่ด้วย `x = 0` หรือ `x = [0] * len(x)` จะเป็นการเปลี่ยนจุดอ้างอิงของตัวแปรภายในฟังก์ชันเท่านั้น **ไม่ส่งผลเปลี่ยนแปลงตัวแปรต้นฉบับภายนอก**

หากต้องการให้ตัวแปรต้นฉบับภายนอกเปลี่ยนค่าด้วย ต้องทำการแก้ไขสมาชิกผ่านดรรชนี เช่น `x[i] = 0` (In-place Modification)

```python
# เปรียบเทียบการกำหนดค่าพารามิเตอร์แบบต่าง ๆ
def f1(x):
    # ตัวแปร x เปลี่ยนค่าภายในฟังก์ชัน แต่ตัวแปรต้นฉบับที่ส่งมาไม่เปลี่ยน
    x = 0

def f2(x):
    # ตัวแปร x ถูกกำหนดเป็นลิสต์ใหม่ แต่ตัวแปรต้นฉบับที่ส่งมาไม่เปลี่ยน
    x = [0] * len(x)

def f3(x):
    # แก้ไขสมาชิกผ่านดรรชนี ตัวแปรต้นฉบับที่ส่งมาจะเปลี่ยนค่าด้วย
    for i in range(len(x)):
        x[i] = 0

# ทดสอบผลลัพธ์
a = 9
f1(a)
print("ค่า a หลังเรียก f1(a):", a)  # a ยังคงเป็น 9 เหมือนเดิม

b = [1, 2, 3, 4]
f2(b)
print("ค่า b หลังเรียก f2(b):", b)  # b ยังคงเป็น [1, 2, 3, 4] เหมือนเดิม

f3(b)
print("ค่า b หลังเรียก f3(b):", b)  # b เปลี่ยนเป็น [0, 0, 0, 0]
```

### 2. สรุปความผิดพลาดที่พบบ่อยเกี่ยวกับฟังก์ชัน
1. **ไม่มีผลคืนมา แต่มีตัวรับผล**: ฟังก์ชันไม่ได้ระบุ `return` (หรือคืน `None`) แต่นำตัวแปรมารับค่า ทำให้ตัวแปรนั้นมีค่าเป็น `None` เช่น `x = hello("Bell")`
2. **คืนผล แต่ไม่มีตัวรับผล**: ฟังก์ชันมีคำสั่ง `return` คืนค่า แต่ผู้เรียกไม่ได้นำตัวแปรมารับค่า ทำให้ผลลัพธ์สูญเปล่า เช่น `double([1, 2, 3])`
3. **มีคำสั่งอยู่ตามหลัง `return`**: คำสั่งที่อยู่หลัง `return` ภายในบล็อกเดียวกันจะไม่ถูกประมวลผล (Unreachable Code)
4. **บางกรณีคืนค่า บางกรณีไม่คืนค่า (Inconsistent Return)**: เช่น เขียนคำสั่ง `if` ที่มี `return True` แต่ไม่ได้ระบุ `else` หรือกรณีอื่น ส่งผลให้เมื่อไม่ตรงเงื่อนไข ฟังก์ชันจะคืนค่า `None` โดยไม่ได้เจตนา

```python
# ตัวอย่างข้อผิดพลาดเกี่ยวกับ Return Value
def hello(name):
    print("Hello", name)

def double(u):
    return 2 * u

# 1. ไม่มีผลคืนมา แต่มีตัวรับผล (x จะได้ค่า None)
x = hello("Bell")
print("ค่าของตัวแปร x คือ:", x)

# 2. คืนผลแต่ไม่มีตัวรับผล (ค่า 2*[1,2,3] จะไม่ถูกบันทึกไว้)
double([1, 2, 3])

# 3. คำสั่งตามหลัง return จะไม่ทำงาน
def clip_wrong(x):
    if x < 0:
        return 0
        x += 2  # คำสั่งนี้ไม่มีทางถูกเรียกใช้งาน
    return x

# 4. บางกรณีคืนค่า บางกรณีไม่คืนค่า (Inconsistent Return)
def is_odd_wrong(x):
    if x % 2 == 1:
        return True
    # หาก x เป็นเลขคู่ จะคืนค่า None โดยไม่เจตนา

print("is_odd_wrong(3):", is_odd_wrong(3))
print("is_odd_wrong(4):", is_odd_wrong(4))  # ได้ค่า None
```

### 3. แบบฝึกหัดประจำบทเรียน 6.5

#### แบบฝึกหัด 6-5 ข้อที่ 1: การเรียกใช้ฟังก์ชัน `mean_sd(x)` เพื่อแสดงค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน
**โจทย์:** ฟังก์ชัน `mean_sd` ข้างล่างนี้หาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน (ทำงานถูกต้องแล้ว ไม่ต้องแก้ไข) จงเขียนคำสั่งเพิ่มเติมเพื่อหาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานของข้อมูลในลิสต์ `x` แสดงบรรทัดละค่า (แสดงค่าเฉลี่ยก่อน)

```python
# ฟังก์ชัน mean_sd สำหรับหาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน
def mean_sd(x):
    mean = sum(x) / len(x)
    s2 = 0
    for e in x:
        s2 += (e - mean) ** 2
    sd = (s2 / len(x)) ** 0.5
    return mean, sd

# ข้อมูลทดสอบ
x = [1, 2, 2, 3, 2, 3, 2, 3, 2, 3, 2, 3, 2, 2, 3, 2]

# คำสั่งเพิ่มเติมเพื่อรับและแสดงผลลัพธ์
m, s = mean_sd(x)
print(m)
print(s)
```
