## 6.1 Function Call (การเรียกใช้งานฟังก์ชัน)

### 1. บทเกริ่นนำและการใช้งาน Built-in Function
ที่ผ่านมาเราเคยเรียกใช้งาน Built-in Function ในภาษา Python มาแล้วหลายฟังก์ชัน เช่น `print("Hello")`, `input()`, `len(x)`, `round(2/3, 2)`, และ `abs(-2)`

การเขียนฟังก์ชันทำขึ้นเพื่อ **แบ่งหน้าที่การทำงานออกเป็นส่วน ๆ** ให้เป็นระเบียบและนำกลับมาใช้ซ้ำได้ง่าย การจะใช้งานฟังก์ชันใด ผู้เขียนโปรแกรมต้องรู้สิ่งสำคัญ 3 ประการ ได้แก่:
1. **ชื่อฟังก์ชัน** (Function Name)
2. **ข้อมูลที่ต้องส่งให้ฟังก์ชัน** (Arguments)
3. **ผลลัพธ์ที่ได้จากฟังก์ชัน** (Return Value)

**เปรียบเทียบฟังก์ชันที่มีและไม่มีค่าคืนกลับ (Return Value):**
- **ตัวอย่าง `print(x)`**: รับข้อมูลแบบใดก็ได้เพื่อแสดงผลบนจอภาพ แต่ **ไม่มีผลลัพธ์คืนกลับมา** จึงไม่นิยมเขียนในลักษณะ `a = print(3/5)` (เพราะ `a` จะได้ค่า `None`)
- **ตัวอย่าง `round(x, d)`**: รับ `x` (จำนวน) และ `d` (จำนวนหลักหลังจุดทศนิยม) โดย **คืนค่าเป็นจำนวนจริง** ที่ปัดเศษแล้ว เช่น `round(2/3, 2)` ได้ `0.67` จึงมักเขียนนำค่าไปใช้งานต่อ เช่น `a = round(2/3, 2)`

````python
# ตัวอย่างการใช้งาน Built-in Function แบบไม่มีผลลัพธ์คืนกลับ (print) และแบบมีผลลัพธ์คืนกลับ (round, abs, len)
print("--- ตัวอย่าง Built-in Functions ---")

# print แสดงผล แต่คืนค่า None
result_print = print("กำลังทดสอบแสดงผล")
print("ค่าที่ได้จาก print คือ:", result_print)

# round และ abs คืนค่ากลับมาให้นำไปคำนวณต่อได้
value = 2 / 3
rounded_value = round(value, 2)
abs_value = abs(-10)

print("ค่าที่ปัดเศษแล้ว:", rounded_value)
print("ค่าสัมบูรณ์:", abs_value)
````

### 2. ฟังก์ชันแรก: `hello(name)`
เราสามารถนิยามฟังก์ชันขึ้นมาเองโดยใช้คำสั่ง `def` เมื่อมีการเรียกใช้งานฟังก์ชัน ค่าของอาร์กิวเมนต์ที่ส่งไปจะถูกคัดลอกให้ตัวแปรพารามิเตอร์ของฟังก์ชัน จากนั้นการทำงานจะย้ายไปยังคำสั่งภายในฟังก์ชัน เมื่อทำงานเสร็จแล้วจึงจะกลับมาทำงานคำสั่งถัดไป

````python
# การนิยามฟังก์ชัน hello รับพารามิเตอร์ name
def hello(name):
    print("Hello", name)

# เรียกใช้งานฟังก์ชัน hello ผ่านการวนลูปลิสต์
BLACKPINK = ["Jisoo", "Jennie", "Rosé", "Lisa"]
for e in BLACKPINK:
    hello(e)
````

### 3. ฟังก์ชันหนึ่งเรียกอีกฟังก์ชันหนึ่ง (Nested Function Calls)
ฟังก์ชันสามารถเรียกใช้งานฟังก์ชันอื่น ๆ ได้ เพื่อช่วยย่อยงานที่มีขนาดใหญ่ให้อยู่ในรูปแบบซับซ้อนน้อยลง

````python
# ฟังก์ชัน hello ทักทายรายคน
def hello(name):
    print("Hello", name)

# ฟังก์ชัน hello_all เรียกใช้ฟังก์ชัน hello วนลูปตามรายชื่อในลิสต์
def hello_all(names):
    for e in names:
        hello(e)

# ทดสอบเรียกใช้ hello_all
BLACKPINK = ["Jisoo", "Jennie", "Rosé", "Lisa"]
hello_all(BLACKPINK)
hello_all(["Joe", "John", "Kong"])
````

### 4. ตัวอย่างประยุกต์: เพลง Happy Birthday

#### ตัวอย่าง 6.1.1: การพิมพ์เนื้อเพลงโดยใช้ฟังก์ชัน `sing(msg, times)`
**โจทย์:** เขียนคำสั่งแสดงเนื้อเพลง Happy Birthday to You โดยการเรียกใช้ฟังก์ชัน `sing` ให้เป็นประโยชน์

**ผลลัพธ์ที่ต้องการ:**
```
Happy Birthday to You
Happy Birthday to You
Happy Birthday
Happy Birthday
Happy Birthday to You
```

````python
# ฟังก์ชัน sing พิมพ์ข้อความ msg เป็นจำนวน times รอบ
def sing(msg, times):
    for k in range(times):
        print(msg)

# การเรียกใช้งานฟังก์ชัน sing เพื่อร้องเพลง Happy Birthday
sing("Happy Birthday to You", 2)
sing("Happy Birthday", 2)
sing("Happy Birthday to You", 1)
````

#### ตัวอย่าง 6.1.2: ฟังก์ชัน `HBD(name)` เรียกใช้ฟังก์ชัน `sing` ซ้อนอีกที
**โจทย์:** เขียนคำสั่งในฟังก์ชัน `HBD(name)` ให้แสดงคำร้องอวยพรวันเกิดระบุชื่อคนด้วย โดยเรียกใช้ฟังก์ชัน `sing` ให้เป็นประโยชน์

**ผลลัพธ์ตัวอย่าง (เมื่อป้อนชื่อ JJ):**
```
Happy Birthday to You
Happy Birthday to You
Happy Birthday Dear JJ
Happy Birthday to You
```

````python
# ฟังก์ชัน sing สำหรับร้องข้อความซ้ำ
def sing(msg, times):
    for k in range(times):
        print(msg)

# ฟังก์ชัน HBD ระบุชื่อคนและเรียกใช้ sing
def HBD(name):
    sing("Happy Birthday to You", 2)
    sing("Happy Birthday Dear " + name, 1)
    sing("Happy Birthday to You", 1)

# รับชื่อและเรียกใช้ฟังก์ชัน HBD
person_name = input()
HBD(person_name)
````

#### 5.2 การปรับปรุงโครงสร้างโปรแกรมด้วยการสร้างและเรียกใช้งานฟังก์ชัน (Refactored with Function Calls)
เมื่อสร้างฟังก์ชันช่วยประมวลผล เช่น `get_zodiac(d, m)` และ `get_days_in_feb(y)` เราจะสามารถลดความซ้ำซ้อนโดยใช้การเรียกใช้งานฟังก์ชัน (Function Call) ทำให้โค้ดอ่านง่าย สั้นลง และลดโอกาสเกิดข้อผิดพลาด

````python
# โค้ดที่ได้รับการปรับปรุงโครงสร้างด้วยการเรียกใช้งานฟังก์ชัน
mname = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
         "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

def read_date(mname_list):
    # ฟังก์ชันสำหรับอ่านและแปลงข้อมูลวันที่
    date_data = input().split()
    d = int(date_data[0])
    m = mname_list.index(date_data[1][:3]) + 1
    y = int(date_data[2])
    return d, m, y

def get_zodiac(d, m):
    # ฟังก์ชันคำนวณราศีจากวันและเดือน
    if (d >= 22 and m == 3) or (d <= 21 and m == 4): return "Aries"
    elif (d >= 22 and m == 4) or (d <= 21 and m == 5): return "Taurus"
    elif (d >= 22 and m == 5) or (d <= 21 and m == 6): return "Gemini"
    elif (d >= 22 and m == 6) or (d <= 21 and m == 7): return "Cancer"
    elif (d >= 22 and m == 7) or (d <= 21 and m == 8): return "Leo"
    elif (d >= 22 and m == 8) or (d <= 21 and m == 9): return "Virgo"
    elif (d >= 22 and m == 9) or (d <= 21 and m == 10): return "Libra"
    elif (d >= 22 and m == 10) or (d <= 21 and m == 11): return "Scorpio"
    elif (d >= 22 and m == 11) or (d <= 21 and m == 12): return "Sagittarius"
    elif (d >= 22 and m == 12) or (d <= 20 and m == 1): return "Capricorn"
    elif (d >= 21 and m == 1) or (d <= 20 and m == 2): return "Aquarius"
    elif (d >= 21 and m == 2) or (d <= 21 and m == 3): return "Pisces"

def get_days_in_feb(y):
    # ฟังก์ชันตรวจสอบจำนวนวันเดือนกุมภาพันธ์ในปี ค.ศ. y
    if y % 400 == 0 or (y % 100 != 0 and y % 4 == 0):
        return 29
    return 28

# เรียกใช้งานฟังก์ชัน (Function Calls)
d1, m1, y1 = read_date(mname)
z1 = get_zodiac(d1, m1)
days_feb1 = get_days_in_feb(y1)

d2, m2, y2 = read_date(mname)
z2 = get_zodiac(d2, m2)
days_feb2 = get_days_in_feb(y2)

print("ราศีวันที่ 1 และ 2:", z1, z2)
print("จำนวนวัน ก.พ. วันที่ 1 และ 2:", days_feb1, days_feb2)
````

### 6. ข้อผิดพลาดที่พบบ่อยในการเรียกใช้งานฟังก์ชัน (Common Pitfalls)
1. **การลืมใส่เครื่องหมายวงเล็บ `()`**: หากพิมพ์เพียงชื่อฟังก์ชันโดยไม่ใส่ `()` โปรแกรมจะอ้างอิงถึงตัววัตถุฟังก์ชัน (Function Object) แทนที่จะเป็นการเรียกประมวลผลฟังก์ชัน
2. **จำนวนหรือชนิดของอาร์กิวเมนต์ไม่ถูกต้อง**: ส่ง Argument น้อยหรือมากกว่าที่ฟังก์ชันต้องการ หรือส่งชนิดข้อมูลที่ไม่รองรับ ทำให้เกิด `TypeError`
3. **การคัดลอกโค้ดโดยไม่ใช้ฟังก์ชัน**: ทำให้ลืมเปลี่ยนชื่อตัวแปร เช่น คัดลอกโค้ดวันที่ 1 มาใช้วันที่ 2 แต่ลืมเปลี่ยน `d1` เป็น `d2`
