## 6.1 Function Call (การเรียกใช้งานฟังก์ชัน)

### 1. บทเกริ่นนำและการใช้งาน Built-in Function
ที่ผ่านมาเราเคยเรียกใช้งาน Built-in Function ในภาษา Python มาแล้วหลายฟังก์ชัน เช่น `print("Hello")`, `input()`, `len(x)`, `round(2/3, 2)`, และ `abs(-2)`

การเขียนฟังก์ชันทำขึ้นเพื่อ **แบ่งหน้าที่การทำงานออกเป็นส่วน ๆ** ให้เป็นระเบียบและนำกลับมาใช้ซ้ำได้ง่าย การจะใช้งานฟังก์ชันใด ผู้เขียนโปรแกรมต้องรู้สิ่งสำคัญ 3 ประการ ได้แก่:
1. **ชื่อฟังก์ชัน** (Function Name)
2. **ข้อมูลที่ต้องส่งให้ฟังก์ชัน** (Arguments)
3. **ผลลัพธ์ที่ได้จากฟังก์ชัน** (Return Value)

**เปรียบเทียบฟังก์ชันที่มีและไม่มีค่าคืนกลับ (Return Value):**
- **ตัวอย่าง `print(x)`**: รับข้อมูลแบบใดก็ได้เพื่อแสดงผลบนจอภาพ แต่ **ไม่มีผลลัพธ์คืนกลับมา** จึงไม่นิยมเขียนในลักษณะ `a = print(3/5)` (เพราะ `a` จะได้ค่า `None`)
- **ตัวอย่าง `round(x, d)`**: รับ `x` (จำนวน) และ `d` (จำนวนหลักหลังจุดทศนิยม) โดย **คืนค่าเป็นจำนวนจริง** ที่ปัดเศษแล้ว เช่น `round(2/3, 2)` ได้ `0.67` จึงมักเขียนนำค่าไปใช้งานต่อ เช่น `a = round(2/3, 2)`

````python
# ตัวอย่างการใช้งาน Built-in Function แบบไม่มีผลลัพธ์คืนกลับ (print) และแบบมีผลลัพธ์คืนกลับ (round, abs, len)
print("--- ตัวอย่าง Built-in Functions ---")

# print แสดงผล แต่คืนค่า None
result_print = print("กำลังทดสอบแสดงผล")
print("ค่าที่ได้จาก print คือ:", result_print)

# round และ abs คืนค่ากลับมาให้นำไปคำนวณต่อได้
value = 2 / 3
rounded_value = round(value, 2)
abs_value = abs(-10)

print("ค่าที่ปัดเศษแล้ว:", rounded_value)
print("ค่าสัมบูรณ์:", abs_value)
````

### 2. ฟังก์ชันแรก: `hello(name)`
เราสามารถนิยามฟังก์ชันขึ้นมาเองโดยใช้คำสั่ง `def` เมื่อมีการเรียกใช้งานฟังก์ชัน ค่าของอาร์กิวเมนต์ที่ส่งไปจะถูกคัดลอกให้ตัวแปรพารามิเตอร์ของฟังก์ชัน จากนั้นการทำงานจะย้ายไปยังคำสั่งภายในฟังก์ชัน เมื่อทำงานเสร็จแล้วจึงจะกลับมาทำงานคำสั่งถัดไป

````python
# การนิยามฟังก์ชัน hello รับพารามิเตอร์ name
def hello(name):
    print("Hello", name)

# เรียกใช้งานฟังก์ชัน hello ผ่านการวนลูปลิสต์
BLACKPINK = ["Jisoo", "Jennie", "Rosé", "Lisa"]
for e in BLACKPINK:
    hello(e)
````

### 3. ฟังก์ชันหนึ่งเรียกอีกฟังก์ชันหนึ่ง (Nested Function Calls)
ฟังก์ชันสามารถเรียกใช้งานฟังก์ชันอื่น ๆ ได้ เพื่อช่วยย่อยงานที่มีขนาดใหญ่ให้อยู่ในรูปแบบซับซ้อนน้อยลง

````python
# ฟังก์ชัน hello ทักทายรายคน
def hello(name):
    print("Hello", name)

# ฟังก์ชัน hello_all เรียกใช้ฟังก์ชัน hello วนลูปตามรายชื่อในลิสต์
def hello_all(names):
    for e in names:
        hello(e)

# ทดสอบเรียกใช้ hello_all
BLACKPINK = ["Jisoo", "Jennie", "Rosé", "Lisa"]
hello_all(BLACKPINK)
hello_all(["Joe", "John", "Kong"])
````

### 4. ตัวอย่างประยุกต์: เพลง Happy Birthday

#### ตัวอย่าง 6.1.1: การพิมพ์เนื้อเพลงโดยใช้ฟังก์ชัน `sing(msg, times)`
**โจทย์:** เขียนคำสั่งแสดงเนื้อเพลง Happy Birthday to You โดยการเรียกใช้ฟังก์ชัน `sing` ให้เป็นประโยชน์

**ผลลัพธ์ที่ต้องการ:**
```
Happy Birthday to You
Happy Birthday to You
Happy Birthday
Happy Birthday
Happy Birthday to You
```

````python
# ฟังก์ชัน sing พิมพ์ข้อความ msg เป็นจำนวน times รอบ
def sing(msg, times):
    for k in range(times):
        print(msg)

# การเรียกใช้งานฟังก์ชัน sing เพื่อร้องเพลง Happy Birthday
sing("Happy Birthday to You", 2)
sing("Happy Birthday", 2)
sing("Happy Birthday to You", 1)
````

#### ตัวอย่าง 6.1.2: ฟังก์ชัน `HBD(name)` เรียกใช้ฟังก์ชัน `sing` ซ้อนอีกที
**โจทย์:** เขียนคำสั่งในฟังก์ชัน `HBD(name)` ให้แสดงคำร้องอวยพรวันเกิดระบุชื่อคนด้วย โดยเรียกใช้ฟังก์ชัน `sing` ให้เป็นประโยชน์

**ผลลัพธ์ตัวอย่าง (เมื่อป้อนชื่อ JJ):**
```
Happy Birthday to You
Happy Birthday to You
Happy Birthday Dear JJ
Happy Birthday to You
```

````python
# ฟังก์ชัน sing สำหรับร้องข้อความซ้ำ
def sing(msg, times):
    for k in range(times):
        print(msg)

# ฟังก์ชัน HBD ระบุชื่อคนและเรียกใช้ sing
def HBD(name):
    sing("Happy Birthday to You", 2)
    sing("Happy Birthday Dear " + name, 1)
    sing("Happy Birthday to You", 1)

# รับชื่อและเรียกใช้ฟังก์ชัน HBD
person_name = input()
HBD(person_name)
````

#### 5.2 การปรับปรุงโครงสร้างโปรแกรมด้วยการสร้างและเรียกใช้งานฟังก์ชัน (Refactored with Function Calls)
เมื่อสร้างฟังก์ชันช่วยประมวลผล เช่น `get_zodiac(d, m)` และ `get_days_in_feb(y)` เราจะสามารถลดความซ้ำซ้อนโดยใช้การเรียกใช้งานฟังก์ชัน (Function Call) ทำให้โค้ดอ่านง่าย สั้นลง และลดโอกาสเกิดข้อผิดพลาด

````python
# โค้ดที่ได้รับการปรับปรุงโครงสร้างด้วยการเรียกใช้งานฟังก์ชัน
mname = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
         "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

def read_date(mname_list):
    # ฟังก์ชันสำหรับอ่านและแปลงข้อมูลวันที่
    date_data = input().split()
    d = int(date_data[0])
    m = mname_list.index(date_data[1][:3]) + 1
    y = int(date_data[2])
    return d, m, y

def get_zodiac(d, m):
    # ฟังก์ชันคำนวณราศีจากวันและเดือน
    if (d >= 22 and m == 3) or (d <= 21 and m == 4): return "Aries"
    elif (d >= 22 and m == 4) or (d <= 21 and m == 5): return "Taurus"
    elif (d >= 22 and m == 5) or (d <= 21 and m == 6): return "Gemini"
    elif (d >= 22 and m == 6) or (d <= 21 and m == 7): return "Cancer"
    elif (d >= 22 and m == 7) or (d <= 21 and m == 8): return "Leo"
    elif (d >= 22 and m == 8) or (d <= 21 and m == 9): return "Virgo"
    elif (d >= 22 and m == 9) or (d <= 21 and m == 10): return "Libra"
    elif (d >= 22 and m == 10) or (d <= 21 and m == 11): return "Scorpio"
    elif (d >= 22 and m == 11) or (d <= 21 and m == 12): return "Sagittarius"
    elif (d >= 22 and m == 12) or (d <= 20 and m == 1): return "Capricorn"
    elif (d >= 21 and m == 1) or (d <= 20 and m == 2): return "Aquarius"
    elif (d >= 21 and m == 2) or (d <= 21 and m == 3): return "Pisces"

def get_days_in_feb(y):
    # ฟังก์ชันตรวจสอบจำนวนวันเดือนกุมภาพันธ์ในปี ค.ศ. y
    if y % 400 == 0 or (y % 100 != 0 and y % 4 == 0):
        return 29
    return 28

# เรียกใช้งานฟังก์ชัน (Function Calls)
d1, m1, y1 = read_date(mname)
z1 = get_zodiac(d1, m1)
days_feb1 = get_days_in_feb(y1)

d2, m2, y2 = read_date(mname)
z2 = get_zodiac(d2, m2)
days_feb2 = get_days_in_feb(y2)

print("ราศีวันที่ 1 และ 2:", z1, z2)
print("จำนวนวัน ก.พ. วันที่ 1 และ 2:", days_feb1, days_feb2)
````

### 6. ข้อผิดพลาดที่พบบ่อยในการเรียกใช้งานฟังก์ชัน (Common Pitfalls)
1. **การลืมใส่เครื่องหมายวงเล็บ `()`**: หากพิมพ์เพียงชื่อฟังก์ชันโดยไม่ใส่ `()` โปรแกรมจะอ้างอิงถึงตัววัตถุฟังก์ชัน (Function Object) แทนที่จะเป็นการเรียกประมวลผลฟังก์ชัน
2. **จำนวนหรือชนิดของอาร์กิวเมนต์ไม่ถูกต้อง**: ส่ง Argument น้อยหรือมากกว่าที่ฟังก์ชันต้องการ หรือส่งชนิดข้อมูลที่ไม่รองรับ ทำให้เกิด `TypeError`
3. **การคัดลอกโค้ดโดยไม่ใช้ฟังก์ชัน**: ทำให้ลืมเปลี่ยนชื่อตัวแปร เช่น คัดลอกโค้ดวันที่ 1 มาใช้วันที่ 2 แต่ลืมเปลี่ยน `d1` เป็น `d2`

## 6.2 Function Definition (การสร้างและนิยามฟังก์ชัน)

### 1. องค์ประกอบของฟังก์ชัน (Function Components)
ในการสร้างและนิยามฟังก์ชันด้วยคำสั่ง `def` จะมีองค์ประกอบสำคัญได้แก่ **ชื่อฟังก์ชัน** (Function Name), **รายการของพารามิเตอร์** (Parameters), และคำสั่ง **`return`** เพื่อคืนผลลัพธ์กลับไปยังจุดที่เรียกใช้งานฟังก์ชัน

**ข้อกำหนดสำคัญเรื่อง Indentation:**
- ต้องเยื้องคำสั่งทั้งหลายภายในบล็อกของฟังก์ชันไปทางขวาให้ตรงกันเสมอ
- คำสั่ง `return` จะทำหน้าที่จบการทำงานของฟังก์ชัน กลับไปทำงานต่อที่จุดที่เรียกใช้ และคืนผลลัพธ์ให้ผู้เรียกได้ด้วย

````python
# ตัวอย่าง 6.2.1: การสร้างฟังก์ชัน dot คำนวณ Dot Product ของลิสต์ 2 ชุด
def dot(u, v):
    p = 0
    for i in range(len(u)):
        p += u[i] * v[i]
    return p

# การเรียกใช้งานฟังก์ชัน dot ในรูปแบบต่าง ๆ
a = [1, 2, 3]
b = [1, 3, 6]
d1 = dot(a, b)
d2 = dot(a, [1, 3, 2])
print("d1:", d1)
print("d2:", d2)
print("2 * dot([1,2,3], b) =", 2 * dot([1, 2, 3], b))
````

### 2. ตัวแปรภายในฟังก์ชันใดเป็นของฟังก์ชันนั้น (Function Scope & Local Variables)
พารามิเตอร์และตัวแปรที่สร้างขึ้นภายในฟังก์ชัน ถือเป็น **ตัวแปรเฉพาะถิ่น (Local Variable)** ของฟังก์ชันนั้น มีคุณสมบัติดังนี้:
1. **ชื่อซ้ำกับตัวแปรของฟังก์ชันอื่นได้** โดยไม่ขัดแย้งกัน
2. **ไม่สามารถเรียกใช้ตัวแปรที่อยู่ในฟังก์ชันอื่นได้** จากภายนอก

````python
# ตัวอย่าง 6.2.2: การใช้ตัวแปรเฉพาะถิ่น d ภายในฟังก์ชันต่าง ๆ โดยไม่ปนกัน
def read_data():
    d = []
    for i in range(int(input())):
        d.append(int(input()))
    return d

def get_mean(d):
    return sum(d) / len(d)

def get_median(d):
    d.sort()
    n = len(d)
    return (d[(n - 1) // 2] + d[n // 2]) / 2
````

### 3. พารามิเตอร์รับข้อมูลจากผู้เรียกฟังก์ชัน (Parameter Passing)
1. คำสั่งภายในฟังก์ชันจะ **เริ่มทำงานเมื่อมีคำสั่งอื่นเรียกใช้** (ถ้าไม่เรียกใช้จะยังไม่ทำงาน)
2. คำสั่งที่เรียก **ต้องส่งข้อมูลอาร์กิวเมนต์ให้จับคู่ตรงกับพารามิเตอร์** ของฟังก์ชันตามลำดับ
3. เมื่อคำสั่งในฟังก์ชันเริ่มทำงาน ถือได้เลยว่าพารามิเตอร์มีค่าแล้ว ไม่ต้องรับค่าจาก `input()` ซ้ำอีกครั้งภายในฟังก์ชัน

````python
# ตัวอย่าง 6.2.3: ฟังก์ชัน distance คำนวณระยะทางระหว่างจุด 2 จุด (x1, y1) และ (x2, y2)
def distance(x1, y1, x2, y2):
    dx = x1 - x2
    dy = y1 - y2
    d = (dx**2 + dy**2) ** 0.5
    return d

# การส่งค่าอาร์กิวเมนต์ตรงเข้าพารามิเตอร์
d1 = distance(10.5, 12.0, 30.0, 50.5)
x = 10.5
y = 11.2
d2 = distance(x, y, 30.0, 50.5)

print("ระยะทาง d1:", round(d1, 2))
print("ระยะทาง d2:", round(d2, 2))
````

### 4. ตัวอย่างการใช้ฟังก์ชันในการคำนวณวันและเดือน
ฟังก์ชันสามารถเรียกใช้วันและเดือนได้อย่างเป็นระบบเพื่อจัดการจำนวนวันในเดือนและปี

````python
# ตัวอย่าง 6.2.4: ฟังก์ชันจัดการจำนวนวันในเดือนและปี
def days_in_feb(y):
    if y % 400 == 0 or (y % 4 == 0 and y % 100 != 0):
        return 29
    else:
        return 28

def days_in_month(m, y):
    if m == 2:
        return days_in_feb(y)
    if m in [4, 6, 9, 11]:
        return 30
    else:
        return 31

def days_in_year(y):
    return 365 - 28 + days_in_feb(y)

print("จำนวนวัน ก.พ. ปี 2024:", days_in_feb(2024))
print("จำนวนวัน เดือน 4 ปี 2024:", days_in_month(4, 2024))
print("จำนวนวันรวมในปี 2024:", days_in_year(2024))
````

### 5. แบบฝึกหัดประจำบทเรียน 6.2

#### แบบฝึกหัด 6-2 ข้อที่ 1: ฟังก์ชันแปลงอุณหภูมิ `to_celsius(f)`
**โจทย์:** จงเขียนฟังก์ชัน `to_celsius(f)` ที่รับอุณหภูมิในหน่วยองศาฟาเรนไฮต์ เพื่อแปลงและคืนผลลัพธ์เป็นอุณหภูมิในหน่วยองศาเซลเซียส

````python
# เฉลยแบบฝึกหัด 6-2 ข้อที่ 1
def to_celsius(f):
    return (f - 32) * 5 / 9

# ทดสอบฟังก์ชัน
print("32°F =", to_celsius(32), "°C")
print("212°F =", to_celsius(212), "°C")
````

#### แบบฝึกหัด 6-2 ข้อที่ 2: ฟังก์ชันคำนวณพื้นที่สามเหลี่ยม `triangle_area(a, b, c)`
**โจทย์:** จงเขียนฟังก์ชัน `triangle_area(a, b, c)` รับพารามิเตอร์สามตัวแทนความยาวด้านทั้งสามของสามเหลี่ยม ฟังก์ชันนี้คำนวณและคืนพื้นที่ของสามเหลี่ยมเป็นผลลัพธ์ หมายเหตุ: สามเหลี่ยมที่มีความยาวด้านทั้งสามเป็น $a, b$ และ $c$ จะมีพื้นที่เท่ากับ $\sqrt{s(s-a)(s-b)(s-c)}$ โดยที่ $s = \frac{a+b+c}{2}$

````python
# เฉลยแบบฝึกหัด 6-2 ข้อที่ 2
def triangle_area(a, b, c):
    s = (a + b + c) / 2
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    return area

# ทดสอบฟังก์ชัน
print("พื้นที่สามเหลี่ยมด้าน 3, 4, 5 คือ:", triangle_area(3, 4, 5))
````

#### แบบฝึกหัด 6-2 ข้อที่ 3: ฟังก์ชันตรวจสอบจำนวนเฉพาะ `is_prime(n)`
**โจทย์:** จงเขียนฟังก์ชัน `is_prime` รับจำนวนเต็มบวกเป็นพารามิเตอร์หนึ่งตัว ฟังก์ชันนี้ทดสอบและคืนผลว่าจำนวนเต็มที่รับมาเป็นจำนวนเฉพาะหรือไม่ ถ้าเป็น ก็คืนจริง (`True`) ถ้าไม่เป็นจำนวนเฉพาะ ก็คืนเท็จ (`False`)

````python
# เฉลยแบบฝึกหัด 6-2 ข้อที่ 3
def is_prime(n):
    if n == 1:
        return False
    for k in range(2, n):
        if n % k == 0:
            return False
    return True

# ทดสอบฟังก์ชัน
print("is_prime(1):", is_prime(1))
print("is_prime(7):", is_prime(7))
print("is_prime(10):", is_prime(10))
````

## 6.3 List Processing Function (ฟังก์ชันสำหรับประมวลผลลิสต์)

### 1. อธิบายแนวคิด (Concept Explanation)
เราสามารถส่งลิสต์ (List) เข้าไปเป็นพารามิเตอร์ของฟังก์ชัน เพื่อทำการประมวลผลชุดข้อมูลได้อย่างเป็นระบบ เช่น การนับจำนวนสมาชิกตามเงื่อนไข การค้นหาฐานนิยม (Mode) การคำนวณผลคูณเชิงสเกลาร์ (Dot Product) หรือการหาผลบวกของเวกเตอร์สองชุด

````python
# ตัวอย่าง 6.3.1: ฟังก์ชัน count นับจำนวนสมาชิก และฟังก์ชัน mode หาค่าฐานนิยมในลิสต์
def count(x, a):
    c = 0
    for e in x:
        if e == a:
            c += 1
    return c

def mode(x):
    max_c = count(x, x[0])
    max_e = x[0]
    for e in x[1:]:
        c = count(x, e)
        if c > max_c:
            max_c = c
            max_e = e
    return max_e

# ทดสอบฟังก์ชัน count และ mode
d = [1, 2, 3, 4, 4, 34, 3, 2, 4, 22]
print("จำนวนเลข 2 ในลิสต์ d:", count(d, 2))
print("ค่าฐานนิยม (mode) ของลิสต์ d:", mode(d))
````

### 2. แบบฝึกหัดประจำบทเรียน 6.3

#### แบบฝึกหัด 6-3 ข้อที่ 1: ฟังก์ชันคำนวณ Dot Product ของ 2 เวกเตอร์ `dot(u, v)`
**โจทย์:** จงเขียนฟังก์ชัน `dot` ที่รับลิสต์ของจำนวนจริง 2 ลิสต์ แต่ละลิสต์แทนเวกเตอร์มีขนาดเท่ากัน ฟังก์ชันนี้คำนวณและคืน dot product ของสองเวกเตอร์ที่ได้รับ

````python
# เฉลยแบบฝึกหัด 6-3 ข้อที่ 1
def dot(u, v):
    d = 0
    for i in range(len(u)):
        d += u[i] * v[i]
    return d

# ทดสอบฟังก์ชัน dot
vector1 = [1, 2, 3]
vector2 = [4, 5, 6]
print("Dot Product:", dot(vector1, vector2))
````

#### แบบฝึกหัด 6-3 ข้อที่ 2: ฟังก์ชันบวกเวกเตอร์ 2 เวกเตอร์ `add(u, v)`
**โจทย์:** จงเขียนฟังก์ชัน `add` ที่รับลิสต์ของจำนวนจริง 2 ลิสต์ แต่ละลิสต์แทนเวกเตอร์มีขนาดเท่ากัน ฟังก์ชันนี้คำนวณและคืนผลบวกของสองเวกเตอร์ที่ได้รับ

````python
# เฉลยแบบฝึกหัด 6-3 ข้อที่ 2
def add(u, v):
    w = []
    for i in range(len(u)):
        w.append(u[i] + v[i])
    return w

# ทดสอบฟังก์ชัน add
v1 = [10, 20, 30]
v2 = [1, 2, 3]
print("Vector Sum:", add(v1, v2))
````

## 6.4 Modifying List Parameter (การแก้ไขค่าของลิสต์ที่ส่งผ่านพารามิเตอร์)

### 1. การคืนผลหลายค่า และรูปแบบคำสั่ง `return`
คำสั่ง `return` ในภาษา Python สามารถใช้ส่งคืนค่าข้อมูลได้หลายรูปแบบ ดังนี้:
1. **คืนผลหลายค่า (Return Multiple Values / List / Tuple)**: เช่น `return [r1, r2]` โดยผู้เรียกสามารถกระจายค่าใส่หลายตัวแปรได้ เช่น `x1, x2 = roots(6, -6, -36)`
2. **การใช้ `return` โดด ๆ**: เพื่อสั่งให้จบการทำงานของฟังก์ชันทันที โดยไม่ส่งค่าข้อมูลกลับไป (หรือคืนค่า `None`)

````python
# ตัวอย่าง 6.4.1: ฟังก์ชัน roots คำนวณคำตอบของสมการกำลังสอง ax^2 + bx + c = 0 คืนค่าเป็นลิสต์ [r1, r2]
def roots(a, b, c):
    t = (b**2 - 4 * a * c) ** 0.5
    r1 = (-b + t) / (2 * a)
    r2 = (-b - t) / (2 * a)
    return [r1, r2]

# การรับค่าคืนกลับหลายค่า
x1, x2 = roots(6, -6, -36)
print("คำตอบของสมการ x1, x2:", x1, x2)

# ตัวอย่าง 6.4.2: การใช้คำสั่ง return โดด ๆ เพื่อออกจากฟังก์ชันเมื่อตรงตามเงื่อนไข
def hello(name):
    if name == "":
        return  # ออกจากฟังก์ชันทันทีหากชื่อเป็นสตริงว่าง
    print("Hello", name)

hello("")       # ไม่แสดงผลอะไร
hello("Alice")  # แสดงผล Hello Alice
````

### 2. เปรียบเทียบการสร้างลิสต์ใหม่กับการแก้ไขลิสต์ต้นฉบับ (In-place Modification)
ลิสต์เป็นข้อมูลประเภทปรับเปลี่ยนค่าได้ (Mutable Object) เมื่อส่งลิสต์เข้าไปในฟังก์ชัน เราสามารถเลือกวิธีประมวลผลได้ 2 แบบ:
- **สร้างลิสต์ใหม่คืนกลับมา**: ไม่แก้ไขลิสต์เดิมที่ส่งเข้าไป ต้องนำตัวแปรมารับค่าคืนกลับไปใช้งาน
- **แก้ไขสมาชิกในลิสต์ต้นฉบับโดยตรง (In-place Modification)**: แก้ไขสมาชิกผ่านดรรชนี `data[k]` ซึ่งจะส่งผลให้ลิสต์ต้นฉบับภายนอกฟังก์ชันเปลี่ยนแปลงไปด้วย โดยฟังก์ชันไม่จำเป็นต้องคืนค่าใด ๆ

````python
# เปรียบเทียบ 1: ฟังก์ชันสร้างลิสต์ใหม่ (ไม่เปลี่ยนลิสต์เดิม)
def abs_all_new(data):
    d = []
    for e in data:
        d.append(abs(e))
    return d

x1 = [1, -2, 3, -4, -5]
x1 = abs_all_new(x1)
print("ผลลัพธ์จาก abs_all_new:", x1)

# เปรียบเทียบ 2: ฟังก์ชันแก้ไขลิสต์ต้นฉบับโดยตรง (In-place modification)
def abs_all_inplace(data):
    for k in range(len(data)):
        data[k] = abs(data[k])

x2 = [1, -2, 3, -4, -5]
abs_all_inplace(x2)  # ข้อมูลในลิสต์ x2 ถูกเปลี่ยนโดยตรง
print("ผลลัพธ์จาก abs_all_inplace:", x2)
````

### 3. แบบฝึกหัดประจำบทเรียน 6.4

#### แบบฝึกหัด 6-4 ข้อที่ 1: ฟังก์ชันเติมค่าในลิสต์ `fill(x, e)`
**โจทย์:** จงเขียนฟังก์ชัน `fill(x, e)` ที่รับ `x` เป็นลิสต์ และ `e` เป็นข้อมูลประเภทใดก็ได้ `fill` จะเปลี่ยนทุกช่องในลิสต์ `x` ให้เก็บค่า `e` (`fill` ไม่คืนผลใด ๆ)

````python
# เฉลยแบบฝึกหัด 6-4 ข้อที่ 1
def fill(x, e):
    for i in range(len(x)):
        x[i] = e

# ทดสอบฟังก์ชัน fill
data = [1, 2, 3, 4, 5]
fill(data, 0)
print("ผลลัพธ์การ fill:", data)
````

#### แบบฝึกหัด 6-4 ข้อที่ 2: ฟังก์ชันจำกัดช่วงข้อมูล `clip(x)`
**โจทย์:** จงเขียนฟังก์ชัน `clip(x)` ที่รับ `x` เป็นลิสต์ของจำนวนเต็ม ฟังก์ชันนี้จะเปลี่ยนทุกข้อมูลใน `x` ที่ติดลบให้เป็น 0 และเปลี่ยนทุกตัวในลิสต์ที่เกิน 255 ให้เป็น 255 (ฟังก์ชันนี้เปลี่ยนค่าในลิสต์ `x` ที่รับมา ฟังก์ชันไม่คืนผลใด ๆ)

````python
# เฉลยแบบฝึกหัด 6-4 ข้อที่ 2
def clip(x):
    for i in range(len(x)):
        if x[i] < 0:
            x[i] = 0
        elif x[i] > 255:
            x[i] = 255

# ทดสอบฟังก์ชัน clip
scores = [-10, 50, 100, 300, 255]
clip(scores)
print("ผลลัพธ์การ clip:", scores)
````

#### แบบฝึกหัด 6-4 ข้อที่ 3: ฟังก์ชันแทนที่ข้อมูลตามจำนวน `replace(x, old, new, count)`
**โจทย์:** จงเขียนฟังก์ชัน `replace(x, old, new, count)` ที่รับ `x` เป็นลิสต์, `old` กับ `new` เป็นข้อมูลประเภทใดก็ได้ และ `count` เป็นจำนวนเต็ม ฟังก์ชันนี้จะเปลี่ยนข้อมูลใน `x` ที่มีค่า `old` ให้เป็น `new` เป็นจำนวนอย่างมาก `count` ตัว (นับจากซ้ายไปขวา) เช่น `x = [1, 2, 3, 3, 3, 2, 3]`, `replace(x, 3, 9, 2)` จะทำให้ `x` เปลี่ยนเป็น `[1, 2, 9, 9, 3, 2, 3]`

````python
# เฉลยแบบฝึกหัด 6-4 ข้อที่ 3
def replace(x, old, new, count):
    c = 0
    i = 0
    while i < len(x) and c < count:
        if x[i] == old:
            x[i] = new
            c += 1
        i += 1

# ทดสอบฟังก์ชัน replace
numbers = [1, 2, 3, 3, 3, 2, 3]
replace(numbers, 3, 9, 2)
print("ผลลัพธ์การ replace:", numbers)
````

1. **การแก้ไขข้อมูลโดยไม่เจตนา (Side Effects)**: เขียนฟังก์ชันแก้ไขสมาชิกในลิสต์โดยผู้ใช้ไม่ทราบ ส่งผลให้ข้อมูลต้นฉบับเสียหาย

## 6.5 Common Mistake (ข้อผิดพลาดที่พบบ่อยในการใช้งานฟังก์ชัน)

### 1. ข้อควรระวัง: อย่ากำหนดค่าตัวแปรพารามิเตอร์ใหม่โดยไม่แก้ไขสมาชิก
เมื่อส่งตัวแปรเข้ามาเป็นพารามิเตอร์ `x` การกำหนดค่าใหม่ด้วย `x = 0` หรือ `x = [0] * len(x)` จะเป็นการเปลี่ยนจุดอ้างอิงของตัวแปรภายในฟังก์ชันเท่านั้น **ไม่ส่งผลเปลี่ยนแปลงตัวแปรต้นฉบับภายนอก**

หากต้องการให้ตัวแปรต้นฉบับภายนอกเปลี่ยนค่าด้วย ต้องทำการแก้ไขสมาชิกผ่านดรรชนี เช่น `x[i] = 0` (In-place Modification)

````python
# เปรียบเทียบการกำหนดค่าพารามิเตอร์แบบต่าง ๆ
def f1(x):
    # ตัวแปร x เปลี่ยนค่าภายในฟังก์ชัน แต่ตัวแปรต้นฉบับที่ส่งมาไม่เปลี่ยน
    x = 0

def f2(x):
    # ตัวแปร x ถูกกำหนดเป็นลิสต์ใหม่ แต่ตัวแปรต้นฉบับที่ส่งมาไม่เปลี่ยน
    x = [0] * len(x)

def f3(x):
    # แก้ไขสมาชิกผ่านดรรชนี ตัวแปรต้นฉบับที่ส่งมาจะเปลี่ยนค่าด้วย
    for i in range(len(x)):
        x[i] = 0

# ทดสอบผลลัพธ์
a = 9
f1(a)
print("ค่า a หลังเรียก f1(a):", a)  # a ยังคงเป็น 9 เหมือนเดิม

b = [1, 2, 3, 4]
f2(b)
print("ค่า b หลังเรียก f2(b):", b)  # b ยังคงเป็น [1, 2, 3, 4] เหมือนเดิม

f3(b)
print("ค่า b หลังเรียก f3(b):", b)  # b เปลี่ยนเป็น [0, 0, 0, 0]
````

### 2. สรุปความผิดพลาดที่พบบ่อยเกี่ยวกับฟังก์ชัน
1. **ไม่มีผลคืนมา แต่มีตัวรับผล**: ฟังก์ชันไม่ได้ระบุ `return` (หรือคืน `None`) แต่นำตัวแปรมารับค่า ทำให้ตัวแปรนั้นมีค่าเป็น `None` เช่น `x = hello("Bell")`
2. **คืนผล แต่ไม่มีตัวรับผล**: ฟังก์ชันมีคำสั่ง `return` คืนค่า แต่ผู้เรียกไม่ได้นำตัวแปรมารับค่า ทำให้ผลลัพธ์สูญเปล่า เช่น `double([1, 2, 3])`
3. **มีคำสั่งอยู่ตามหลัง `return`**: คำสั่งที่อยู่หลัง `return` ภายในบล็อกเดียวกันจะไม่ถูกประมวลผล (Unreachable Code)
4. **บางกรณีคืนค่า บางกรณีไม่คืนค่า (Inconsistent Return)**: เช่น เขียนคำสั่ง `if` ที่มี `return True` แต่ไม่ได้ระบุ `else` หรือกรณีอื่น ส่งผลให้เมื่อไม่ตรงเงื่อนไข ฟังก์ชันจะคืนค่า `None` โดยไม่ได้เจตนา

````python
# ตัวอย่างข้อผิดพลาดเกี่ยวกับ Return Value
def hello(name):
    print("Hello", name)

def double(u):
    return 2 * u

# 1. ไม่มีผลคืนมา แต่มีตัวรับผล (x จะได้ค่า None)
x = hello("Bell")
print("ค่าของตัวแปร x คือ:", x)

# 2. คืนผลแต่ไม่มีตัวรับผล (ค่า 2*[1,2,3] จะไม่ถูกบันทึกไว้)
double([1, 2, 3])

# 3. คำสั่งตามหลัง return จะไม่ทำงาน
def clip_wrong(x):
    if x < 0:
        return 0
        x += 2  # คำสั่งนี้ไม่มีทางถูกเรียกใช้งาน
    return x

# 4. บางกรณีคืนค่า บางกรณีไม่คืนค่า (Inconsistent Return)
def is_odd_wrong(x):
    if x % 2 == 1:
        return True
    # หาก x เป็นเลขคู่ จะคืนค่า None โดยไม่เจตนา

print("is_odd_wrong(3):", is_odd_wrong(3))
print("is_odd_wrong(4):", is_odd_wrong(4))  # ได้ค่า None
````

### 3. แบบฝึกหัดประจำบทเรียน 6.5

#### แบบฝึกหัด 6-5 ข้อที่ 1: การเรียกใช้ฟังก์ชัน `mean_sd(x)` เพื่อแสดงค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน
**โจทย์:** ฟังก์ชัน `mean_sd` ข้างล่างนี้หาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน (ทำงานถูกต้องแล้ว ไม่ต้องแก้ไข) จงเขียนคำสั่งเพิ่มเติมเพื่อหาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานของข้อมูลในลิสต์ `x` แสดงบรรทัดละค่า (แสดงค่าเฉลี่ยก่อน)

````python
# ฟังก์ชัน mean_sd สำหรับหาค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน
def mean_sd(x):
    mean = sum(x) / len(x)
    s2 = 0
    for e in x:
        s2 += (e - mean) ** 2
    sd = (s2 / len(x)) ** 0.5
    return mean, sd

# ข้อมูลทดสอบ
x = [1, 2, 2, 3, 2, 3, 2, 3, 2, 3, 2, 3, 2, 2, 3, 2]

# คำสั่งเพิ่มเติมเพื่อรับและแสดงผลลัพธ์
m, s = mean_sd(x)
print(m)
print(s)
````
