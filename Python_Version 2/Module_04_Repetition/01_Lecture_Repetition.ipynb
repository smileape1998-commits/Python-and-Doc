# Module 4: Repetition (การทำงานซ้ำวนลูป)

ในบทนี้เราจะเรียนรู้เกี่ยวกับการควบคุมทิศทางการทำงานของโปรแกรมให้ทำงานคำสั่งเดิมซ้ำๆ (Looping หรือ Iteration) ตามเงื่อนไขที่กำหนด ช่วยลดความซ้ำซ้อนของโค้ดและเพิ่มประสิทธิภาพในการประมวลผลข้อมูลชุดใหญ่

## 4.1 Flowchart (ผังงานการทำงานซ้ำ)

### อธิบายแนวคิด (Concept Explanation)
ผังงานแสดงการทำงานซ้ำ (Loop Flowchart) ใช้สัญลักษณ์รูปเพชร (Decision Symbol) ในการตรวจสอบเงื่อนไข:
- หากเงื่อนไขเป็น **จริง (True)** โปรแกรมจะเข้าไปทำงานในกลุ่มคำสั่งภายในลูป (Loop Body) แล้ววนกลับมาตรวจสอบเงื่อนไขเดิมอีกครั้ง
- หากเงื่อนไขเป็น **เท็จ (False)** โปรแกรมจะหลุดออกจากลูปแล้วข้ามไปทำงานในคำสั่งถัดไปที่อยู่หลังลูป

### องค์ประกอบสำคัญ 3 ประการของการวนลูป
1. **การกำหนดค่าเริ่มต้น (Initialization):** การตั้งค่าตัวแปรควบคุมลูปก่อนเข้าลูป
2. **เงื่อนไขทดสอบ (Condition):** นิพจน์บูลีนที่ใช้ตัดสินใจว่าจะวนซ้ำต่อไปหรือไม่
3. **การปรับปรุงค่า (Update):** การเปลี่ยนค่าตัวแปรควบคุมลูปในแต่ละรอบ เพื่อให้เข้าใกล้เงื่อนไขสิ้นสุดลูป

### ตัวอย่างผังงาน 1: การนับจำนวนครั้งที่ n ถูกหารด้วย 2

![ผังงานนับการหารด้วย 2](images/ex04_01_01.png)

````python
# ตัวอย่าง 4.1.1: นับจำนวนครั้งที่ n ถูกหารด้วย 2 จนกว่า n <= 0
n = 16
c = 0
while n > 0:
    n //= 2
    c += 1
print(c)
````

### ตัวอย่างผังงาน 2: การแปลงเลขฐานสองเป็นเลขฐานสิบ

![ผังงานแปลงเลขฐานสองเป็นฐานสิบ](images/ex04_01_02.png)

````python
# ตัวอย่าง 4.1.2: แปลงสตริงเลขฐานสอง b เป็นเลขฐานสิบ d
b = "1011"
n = len(b)
k = 0
d = 0
while k < n:
    d += 2**k * int(b[n - k - 1])
    k += 1
print(d)
````

### ตัวอย่างผังงาน 3: การใช้เงื่อนไขซ้อนภายในลูป

![ผังงานเงื่อนไขซ้อนในลูป](images/ex04_01_03.png)

````python
# ตัวอย่าง 4.1.3: การทำงานซ้ำร่วมกับเงื่อนไข if ซ้อนภายในลูป
n = 10
c = 0
while n > 0:
    n = n - 1
    if n % 3 == 0:
        n = n - c
        c += 1
print(c)
````

## 4.2 while (การวนซ้ำด้วยคำสั่ง while)

### อธิบายแนวคิด (Concept Explanation)
คำสั่ง `while` จะตรวจสอบเงื่อนไขก่อนเริ่มทำงานเสมอ หากเงื่อนไขเป็นจริง จะทำงานตามคำสั่งภายในบล็อกย่อหน้า เหมาะสำหรับกรณีที่ไม่ทราบจำนวนรอบการทำงานซ้ำที่แน่นอนล่วงหน้า ขึ้นอยู่กับเงื่อนไขในขณะประมวลผล

### รูปแบบไวยากรณ์
```python
while เงื่อนไข:
    คำสั่งที่ต้องการทำซ้ำ
```

### ตัวอย่าง 4.2.1: การหาค่าน้อยที่สุดจากข้อมูล n จำนวน

````python
# ตัวอย่าง 4.2.1: หาค่าน้อยที่สุด min_x จากข้อมูล n จำนวน
n = int(input())
min_x = int(input())
k = 0
while k < n - 1:
    x = int(input())
    if x < min_x:
        min_x = x
    k += 1
print(min_x)
````

### ตัวอย่าง 4.2.2: การนับจำนวนเลข 0 จากข้อมูลที่แยกด้วย space

````python
# ตัวอย่าง 4.2.2: นับจำนวนเลข 0 จากข้อมูลในบรรทัดเดียวกัน
# ตัวอย่างข้อมูลอินพุต: 10 20 3 0 3 0 43 0 3 20 0 2
x = input().split()
i = 0
c = 0
while i < len(x):
    if int(x[i]) == 0:
        c += 1
    i += 1
print(c)
````

### ตัวอย่าง 4.2.3: นับสระในข้อความ t โดยใช้ while loop (พัฒนาต่อจากแบบฝึกหัดบทก่อน)

````python
# ตัวอย่าง 4.2.3: อ่านข้อความ t เพื่อนับจำนวนสระ (ทั้งตัวเล็กและตัวใหญ่)
t = input()
c = 0
vowels = "aeiouAEIOU"
k = 0
while k < len(t):
    if t[k] in vowels:
        c += 1
    k += 1
print(c)
````

### แบบฝึกหัด 4-2

````python
# แบบฝึกหัด 4-2 ข้อที่ 1:
# รับจำนวนเต็ม n จาก input เพื่อแสดงจำนวนเต็มคี่ตั้งแต่ 1 ถึงจำนวนคี่ที่ไม่เกิน n (โดยใช้ขณะวน while)
n = int(input())
i = 1
while i <= n:
    print(i)
    i += 2
````

````python
# แบบฝึกหัด 4-2 ข้อที่ 2:
# บรรทัดแรกเป็นจำนวนเต็ม n ตามด้วยอีก n บรรทัดเป็นจำนวนเต็ม d0, d1, ..., dn-1 จงหาผลรวมของจำนวนทั้งหมด
n = int(input())
k = 0
s = 0
while k < n:
    s += int(input())
    k += 1
print(s)
````

````python
# แบบฝึกหัด 4-2 ข้อที่ 3:
# รับจำนวนเต็มบรรทัดละจำนวน เพื่อหาผลรวมของจำนวนที่อ่านเข้ามา จนพบจำนวนลบ (ไม่รวมจำนวนลบ)
s = 0
x = int(input())
while x >= 0:
    s += x
    x = int(input())
print(s)
````

````python
# แบบฝึกหัด 4-2 ข้อที่ 4:
# รับจำนวนเต็มบรรทัดละจำนวน จนถึงบรรทัดสุดท้ายจะเป็นคำว่า "end" เพื่อหาผลรวมของจำนวนที่อ่านเข้ามา
s = 0
x = input()
while x != "end":
    s += int(x)
    x = input()
print(s)
````

## 4.3 Bisection (วิธีการแบ่งครึ่งช่วงเพื่อหาคำตอบ)

### อธิบายแนวคิด (Concept Explanation)
Bisection Method เป็นอัลกอริทึมค้นหาหาคำตอบโดยใช้การวนซ้ำ `while` เพื่อแบ่งครึ่งช่วงคำตอบออกเป็นสองส่วนในแต่ละรอบ ทำให้ขอบเขตคำตอบแคบลงเรื่อยๆ จนกระทั่งได้ความถูกต้องตามค่าความคลาดเคลื่อน (Tolerance) ที่ยอมรับได้

````python
# ตัวอย่าง 4.3: การหาค่ารากที่สองของ x โดยใช้วิธี Bisection Search
x = 25.0
epsilon = 0.00001 # ความคลาดเคลื่อนที่ยอมรับได้
low = 0.0
high = max(1.0, x)
ans = (high + low) / 2.0

while abs(ans**2 - x) >= epsilon:
    if ans**2 < x:
        low = ans
    else:
        high = ans
    ans = (high + low) / 2.0

print("Square root of", x, "is approximately", ans)
````

## 4.4 for i in range (การวนซ้ำตามช่วงตัวเลข)

### อธิบายแนวคิด (Concept Explanation)
คำสั่ง `for` ร่วมกับฟังก์ชัน `range()` ใช้เมื่อทราบจำนวนรอบการวนซ้ำที่แน่นอนล่วงหน้า

````python
# ตัวอย่าง 4.4: แสดงแม่สูตรคูณแม่ 5 โดยใช้ range(start, stop)
for i in range(1, 13):
    print(f"5 x {i} = {5 * i}")
````

## 4.5 for c in string (การวนลูปทีละตัวอักษรในสตริง)

### อธิบายแนวคิด (Concept Explanation)
การใช้คำสั่ง `for` ท่องไปในข้อความ ตัวแปรลูปจะดึงตัวอักษรทีละอักขระออกมารายตัวจากซ้ายไปขวาจนครบ

````python
# ตัวอย่าง 4.5: นับจำนวนสระภาษาอังกฤษในข้อความ
text = "Python Programming"
vowel_count = 0

for c in text:
    if c in "aeiouAEIOU":
        vowel_count += 1

print("จำนวนสระที่พบ:", vowel_count)
````

## 4.6 for e in list (การวนลูปทีละสมาชิกในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การใช้คำสั่ง `for` ท่องไปในลิสต์ ตัวแปรลูปจะเข้าถึงสมาชิกแต่ละตัวในลิสต์ทีละธาตุเรียงตามลำดับ Index

````python
# ตัวอย่าง 4.6: คัดกรองเฉพาะจำนวนคู่จากลิสต์ของตัวเลข
numbers = [12, 7, 9, 20, 15, 30]
even_numbers = []

for e in numbers:
    if e % 2 == 0:
        even_numbers.append(e)

print("จำนวนคู่ในลิสต์:", even_numbers)
````

## 4.7 break (คำสั่งออกจากลูปทันที)

### อธิบายแนวคิด (Concept Explanation)
คำสั่ง `break` ใช้สำหรับการยกเลิกและออกจากลูปการทำงานทันทีโดยไม่ต้องรอให้เงื่อนไขลูปเป็นเท็จ

````python
# ตัวอย่าง 4.7: การค้นหาตัวเลขแรกที่หารด้วย 7 ลงตัวในช่วง 1 ถึง 50
for i in range(1, 51):
    if i % 7 == 0:
        print("พบตัวเลขแรกที่หารด้วย 7 ลงตัวคือ:", i)
        break
````
