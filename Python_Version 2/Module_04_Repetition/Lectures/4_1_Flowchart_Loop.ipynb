## 4.1 Flowchart (ผังงานการทำงานซ้ำ)

### อธิบายแนวคิด (Concept Explanation)
ผังงานแสดงการทำงานซ้ำ (Loop Flowchart) ใช้สัญลักษณ์รูปเพชร (Decision Symbol) ในการตรวจสอบเงื่อนไข:
- หากเงื่อนไขเป็น **จริง (True)** โปรแกรมจะเข้าไปทำงานในกลุ่มคำสั่งภายในลูป (Loop Body) แล้ววนกลับมาตรวจสอบเงื่อนไขเดิมอีกครั้ง
- หากเงื่อนไขเป็น **เท็จ (False)** โปรแกรมจะหลุดออกจากลูปแล้วข้ามไปทำงานในคำสั่งถัดไปที่อยู่หลังลูป

### องค์ประกอบสำคัญ 3 ประการของการวนลูป
1. **การกำหนดค่าเริ่มต้น (Initialization):** การตั้งค่าตัวแปรควบคุมลูปก่อนเข้าลูป
2. **เงื่อนไขทดสอบ (Condition):** นิพจน์บูลีนที่ใช้ตัดสินใจว่าจะวนซ้ำต่อไปหรือไม่
3. **การปรับปรุงค่า (Update):** การเปลี่ยนค่าตัวแปรควบคุมลูปในแต่ละรอบ เพื่อให้เข้าใกล้เงื่อนไขสิ้นสุดลูป

### ตัวอย่าง 4.1.1: การนับจำนวนครั้งที่ n ถูกหารด้วย 2

![ผังงานนับการหารด้วย 2](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_01_01.png)

**คำอธิบายโจทย์:**
โปรแกรมนับว่าจำนวนเต็ม `n` สามารถนำมาหารปัดเศษด้วย 2 (`n //= 2`) ได้กี่ครั้ง จนกระทั่ง `n <= 0`

* **📥 ข้อมูลนำเข้า (Input):** กำหนดค่าตัวแปร `n` เช่น `16`
* **📤 ข้อมูลส่งออก (Output):** จำนวนรอบ `c` ที่ทำการหารสำเร็จ (สำหรับ `16` จะได้ `5` ครั้ง คือ $16 \rightarrow 8 \rightarrow 4 \rightarrow 2 \rightarrow 1 \rightarrow 0$)

```python
# ตัวอย่าง 4.1.1: นับจำนวนครั้งที่ n ถูกหารด้วย 2 จนกว่า n <= 0
n = 16
c = 0
while n > 0:
    n //= 2
    c += 1
print(c)
```

### ตัวอย่าง 4.1.2: การแปลงเลขฐานสองเป็นเลขฐานสิบ

![ผังงานแปลงเลขฐานสองเป็นฐานสิบ](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_01_02.png)

**คำอธิบายโจทย์:**
แปลงสตริงเลขฐานสอง `b` ให้เป็นจำนวนเต็มฐานสิบ `d` ตามผังงาน

* **📥 ข้อมูลนำเข้า (Input):** กำหนดสตริงเลขฐานสอง `b` เช่น `"1011"`
* **📤 ข้อมูลส่งออก (Output):** เลขฐานสิบ `d` (สำหรับ `"1011"` จะได้ `11` คือ $1(2^3) + 0(2^2) + 1(2^1) + 1(2^0)$)

```python
# ตัวอย่าง 4.1.2: แปลงสตริงเลขฐานสอง b เป็นเลขฐานสิบ d
b = "1011"
n = len(b)
k = 0
d = 0
while k < n:
    d += 2**k * int(b[n - k - 1])
    k += 1
print(d)
```

### ตัวอย่าง 4.1.3: การใช้เงื่อนไขซ้อนภายในลูป

![ผังงานเงื่อนไขซ้อนในลูป](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_01_03.png)

**คำอธิบายโจทย์:**
ทำงานซ้ำตามผังงาน โดยในแต่ละรอบจะลดค่า `n` ลง 1 แล้วตรวจสอบว่า `n % 3 == 0` หรือไม่ ถ้าใช่จะนำค่า `c` มาลบออกจาก `n` แล้วเพิ่ม `c += 1`

* **📥 ข้อมูลนำเข้า (Input):** กำหนดค่าตัวแปร `n` เช่น `10`
* **📤 ข้อมูลส่งออก (Output):** ค่าของตัวนับ `c` หลังสิ้นสุดลูป

```python
# ตัวอย่าง 4.1.3: การทำงานซ้ำร่วมกับเงื่อนไข if ซ้อนภายในลูป
n = 10
c = 0
while n > 0:
    n = n - 1
    if n % 3 == 0:
        n = n - c
        c += 1
print(c)
```
