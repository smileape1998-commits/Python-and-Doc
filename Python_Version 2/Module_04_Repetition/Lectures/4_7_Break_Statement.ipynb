## 4.7 break (การหยุดทำงานลูปก่อนกำหนด)

### อธิบายแนวคิด (Concept Explanation)
คำสั่ง `break` ใช้เพื่อบังคับให้โปรแกรม **ออกจากลูปทันที** โดยไม่สนใจว่าเงื่อนไขของลูปจะเป็นจริงหรือมีสมาชิกลูปเหลืออยู่อีกหรือไม่ เมื่อเจอคำสั่ง `break` โปรแกรมจะกระโดดข้ามไปทำงานคำสั่งถัดไปที่อยู่นอกลูปทันที

### ตัวอย่าง: การค้นหาตัวเลขแรกที่หารด้วย 7 ลงตัวในช่วง 1 ถึง 50

**คำอธิบายโจทย์:**
วนลูปหาตัวเลขตัวแรกที่หารด้วย 7 ลงตัว เมื่อพบแล้วให้แสดงผลและสั่ง `break` ออกจากลูปทันที

* **ข้อมูลส่งออก (Output):** ตัวเลข `7`

```python
# ตัวอย่าง 4.7: การค้นหาตัวเลขแรกที่หารด้วย 7 ลงตัวในช่วง 1 ถึง 50
for i in range(1, 51):
    if i % 7 == 0:
        print(f"พบตัวเลขแรกที่หารด้วย 7 ลงตัวคือ: {i}")
        break
```

### ตัวอย่าง 4.7.1: ผลรวมข้อมูลสูงสุด 10 จำนวน หยุดเมื่อเจอ -1

**คำอธิบายโจทย์:**
รับจำนวนเต็มเข้ามาบวกสะสมได้สูงสุดไม่เกิน 10 จำนวน (`range(10)`) แต่หากพบค่า `-1` ให้หยุดรับข้อมูลทันที (`break`)

* **ข้อมูลที่ต้องป้อน (Input):** ตัวเลขทีละบรรทัด (ไม่เกิน 10 ตัว หรือใส่ `-1` เพื่อหยุดก่อน) เช่น `10`, `20`, `30`, `-1`
* **ข้อมูลส่งออก (Output):** ผลรวมของตัวเลขทั้งหมดก่อนหน้า `-1` (เช่น `60`)

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `10`<br>`20`<br>`30`<br>`-1` | `60` |

```python
# ตัวอย่าง 4.7.1: ผลรวมข้อมูลสูงสุด 10 จำนวน หยุดเมื่อเจอ -1
s = 0
k = 0
for i in range(10):
    x = int(input())
    if x == -1:
        break
    s += x
    k += 1
print("Sum =", s)
print("Count =", k)
```

### ตัวอย่าง 4.7.2: การตรวจสอบจำนวนเฉพาะ (Prime Number Test)

**คำอธิบายโจทย์:**
รับจำนวนเต็ม `n` แล้ววนลูปตรวจสอบว่ามีตัวเลขใดตั้งแต่ `2` ถึง `n - 1` ที่หาร `n` ลงตัวหรือไม่ หากพบให้แสดง `"Composite"` แล้ว `break` ทันที แต่ถ้าวนจนถึง `n` แล้วไม่พบตัวหารลงตัว แสดงว่า `n` เป็น `"Prime"`

* **ข้อมูลที่ต้องป้อน (Input):** จำนวนเต็ม `n` เช่น `17` หรือ `21`
* **ข้อมูลส่งออก (Output):** `"Prime"` หรือ `"Composite"`

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `17` | `Prime` |
| `21` | `Composite` |

```python
# ตัวอย่าง 4.7.2: ตรวจสอบจำนวนเฉพาะ Prime/Composite
n = int(input())
for k in range(2, n + 1):
    if n % k == 0:
        break
if k == n and n > 1:
    print("Prime")
else:
    print("Composite")
```

### ตัวอย่าง 4.7.3: เกมทายตัวเลข (Number Guessing Game)

**คำอธิบายโจทย์:**
โปรแกรมสุ่มตัวเลขลับ `0` ถึง `99` ให้ผู้ใช้ทายตัวเลขซ้ำๆ จนกว่าจะทายถูกต้อง เมื่อทายถูกให้แสดงจำนวนครั้งที่ทายและ `break` ออกจากลูป

* **ข้อมูลที่ต้องป้อน (Input):** ตัวเลขที่ทายในแต่ละรอบ (เช่น `50`, `75`, `63`)
* **ข้อมูลส่งออก (Output):** คำใบ้ว่า `"มากไป"` หรือ `"น้อยไป"` และแสดง `"ถูกต้อง!"` เมื่อทายถูก

```python
# ตัวอย่าง 4.7.3: เกมทายตัวเลข
import random

my_number = random.randint(0, 99)
print("ยินดีต้อนรับสู่เกมทายตัวเลข 0 ถึง 99!")
c = 0
while True:
    your_guess = int(input("ทายตัวเลข: "))
    c += 1
    if your_guess == my_number:
        print(f"ถูกต้อง! คุณทายทั้งหมด {c} ครั้ง")
        break
    elif your_guess > my_number:
        print("มากไป")
    else:
        print("น้อยไป")
```

### ตัวอย่าง 4.7.4: การแปลงผังงานที่มี break (while loop)

**คำอธิบายโจทย์:**
รับจำนวนเต็ม `n` จากนั้นวนลูปรับตัวเลข `x` เข้ามาอีก `n` ครั้ง หากพบ `x < 0` ให้หยุดลูปทันที (`break`)

* **ข้อมูลที่ต้องป้อน (Input):** บรรทัดแรกคือ `n` (เช่น `5`) ตามด้วยตัวเลข เช่น `10`, `20`, `-1`
* **ข้อมูลส่งออก (Output):** ผลรวม `s` ก่อนเจอค่าลบ (เช่น `30`)

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `5`<br>`10`<br>`20`<br>`-1` | `30` |

```python
# ตัวอย่าง 4.7.4: while loop ร่วมกับ break เมื่อพบค่าลบ
n = int(input())
s = 0
k = 0
while k < n:
    x = int(input())
    if x < 0:
        break
    s += x
    k += 1
print(s)
```

### ตัวอย่าง 4.7.5: การแปลงผังงานที่มี break (for loop)

**คำอธิบายโจทย์:**
ทำงานเหมือนตัวอย่าง 4.7.4 แต่ใช้ลูป `for _ in range(n):` แทน `while`

* **ข้อมูลที่ต้องป้อน (Input):** บรรทัดแรกคือ `n` (เช่น `4`) ตามด้วยตัวเลข เช่น `5`, `8`, `-2`
* **ข้อมูลส่งออก (Output):** ผลรวม `s` (เช่น `13`)

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `4`<br>`5`<br>`8`<br>`-2` | `13` |

```python
# ตัวอย่าง 4.7.5: for loop ร่วมกับ break เมื่อพบค่าลบ
n = int(input())
s = 0
for _ in range(n):
    x = int(input())
    if x < 0:
        break
    s += x
print(s)
```
