## 4.4 for i in range (การวนซ้ำตามช่วงตัวเลข)

### อธิบายแนวคิด (Concept Explanation)
คำสั่ง `for` ร่วมกับฟังก์ชัน `range()` ใช้เมื่อทราบจำนวนรอบการวนซ้ำที่แน่นอนล่วงหน้า โดยฟังก์ชัน `range()` จะสร้างลำดับของจำนวนเต็มตามพารามิเตอร์ที่กำหนด

### รูปแบบการใช้งาน range()
1. `range(stop)` : เริ่มจาก `0` ถึง `stop - 1` (เพิ่มทีละ 1)
2. `range(start, stop)` : เริ่มจาก `start` ถึง `stop - 1` (เพิ่มทีละ 1)
3. `range(start, stop, step)` : เริ่มจาก `start` ถึงก่อนถึง `stop` เพิ่ม/ลด ครั้งละ `step`

![ตัวอย่างการทำงานของ range](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_04_11.png)

**ตัวอย่างช่วงค่าตัวแปรจาก range ในแบบต่างๆ:**
- `range(4, 100, 2)` $\rightarrow$ `k = 4, 6, 8, ..., 98`
- `range(100, 0, -1)` $\rightarrow$ `k = 100, 99, 98, ..., 1`
- `range(5, 100)` $\rightarrow$ `k = 5, 6, 7, ..., 99`
- `range(100)` $\rightarrow$ `k = 0, 1, 2, ..., 99`

### ตัวอย่าง: การพิมพ์ค่า 0 ถึง n-1

**คำอธิบายโจทย์:**
รับจำนวนเต็ม `n` แล้วพิมพ์ตัวเลขตั้งแต่ `0` ถึง `n - 1` บรรทัดละจำนวน

* **ข้อมูลที่ต้องป้อน (Input):** จำนวนเต็ม `n` เช่น `5`
* **ข้อมูลส่งออก (Output):** ตัวเลข `0, 1, 2, 3, 4` ทีละบรรทัด

```python
# ตัวอย่างการพิมพ์ค่า 0 ถึง n-1 โดยใช้ range(n)
n = int(input())
for i in range(n):
    print(i)
```

### ตัวอย่าง: แสดงแม่สูตรคูณแม่ 5

**คำอธิบายโจทย์:**
ใช้ `range(1, 13)` เพื่อแสดงแม่สูตรคูณแม่ 5 ตั้งแต่ $5 \times 1$ ถึง $5 \times 12$

* **ข้อมูลส่งออก (Output):** ตารางสูตรคูณแม่ 5

```python
# ตัวอย่าง: แสดงแม่สูตรคูณแม่ 5 โดยใช้ range(start, stop)
for i in range(1, 13):
    print(f"5 x {i} = {5 * i}")
```

### ตัวอย่าง 4.4.1: การหาผลรวม 0 ถึง n-1

![ผังงาน 4-4 ข้อ 1](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_04_01.png)

**คำอธิบายโจทย์:**
รับค่า `n` จากผู้ใช้ จากนั้นวนลูป `i` ตั้งแต่ `0` ถึง `n - 1` แล้วบวกสะสมค่า `i` เข้าในตัวแปร `s`

* **ข้อมูลที่ต้องป้อน (Input):** จำนวนเต็ม `n` เช่น `5`
* **ข้อมูลส่งออก (Output):** ผลรวม `s` ($0 + 1 + 2 + 3 + 4 = 10$)

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `5` | `10` |
| `10` | `45` |

```python
# ตัวอย่าง 4.4.1: คำนวณผลรวม i = 0 ถึง n-1
n = int(input())
s = 0
for i in range(n):
    s += i
print(s)
```

### ตัวอย่าง 4.4.2: การคำนวณผลรวมเงื่อนไขเลขคู่ (วนจาก 1 ถึง n)

![ผังงาน 4-4 ข้อ 2](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_04_02.png)

**คำอธิบายโจทย์:**
รับค่า `n` จากผู้ใช้ วนลูป `i` ตั้งแต่ `1` ถึง `n` (`range(1, n+1)`)
- ตรวจสอบว่า `i` เป็นเลขคู่หรือไม่ (`i % 2 == 0`)
- หากเป็นเลขคู่ ให้บวก `s += i` เพิ่มอีกรอบหนึ่ง
- ทุกๆ รอบจะมีการบวก `s += i` เสมอ

* **ข้อมูลที่ต้องป้อน (Input):** จำนวนเต็ม `n` เช่น `4`
* **ข้อมูลส่งออก (Output):** ผลรวม `s` ตามเงื่อนไขผังงาน (เช่น `16`)

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `4` | `16` |
| `3` | `8` |

```python
# ตัวอย่าง 4.4.2: คำนวณผลรวมเงื่อนไขเลขคู่ (1 ถึง n)
n = int(input())
s = 0
for i in range(1, n + 1):
    if i % 2 == 0:
        s += i
    s += i
print(s)
```

### ตัวอย่าง 4.4.3: การวนลูปถอยหลังพร้อมปรับเปลี่ยนตัวแปร k

![ผังงาน 4-4 ข้อ 3](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex04_04_03.png)

**คำอธิบายโจทย์:**
รับค่า `n` ตั้งค่า `s = 0` และ `k = 1` วนลูปถอยหลัง `i` ตั้งแต่ `n` ถึง `1` (`range(n, 0, -1)`)
- ถ้า `i` เป็นเลขคู่ ให้ทำ `s += i * k`
- ถ้า `i` เป็นเลขคี่ ให้ทำ `s -= k`
- ในทุกรอบการวนซ้ำจะบวก `k += 2` และบวก `s += i * k`

* **ข้อมูลที่ต้องป้อน (Input):** จำนวนเต็ม `n` เช่น `3`
* **ข้อมูลส่งออก (Output):** ค่าของ `s` เมื่อจบลูป

| ตัวอย่าง Input | ตัวอย่าง Output |
|---|---|
| `3` | `26` |

```python
# ตัวอย่าง 4.4.3: วนลูปถอยหลังพร้อมปรับเปลี่ยนตัวแปร k
n = int(input())
s = 0
k = 1
for i in range(n, 0, -1):
    if i % 2 == 0:
        s += i * k
    else:
        s -= k
    k += 2
    s += i * k
print(s)
```
