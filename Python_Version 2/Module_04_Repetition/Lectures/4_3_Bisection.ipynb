## 4.3 Bisection (วิธีการแบ่งครึ่งช่วงเพื่อหาคำตอบ)

### อธิบายแนวคิด (Concept Explanation)
Bisection Method เป็นอัลกอริทึมค้นหาหาคำตอบโดยใช้การวนซ้ำ `while` เพื่อแบ่งครึ่งช่วงคำตอบออกเป็นสองส่วนในแต่ละรอบ ทำให้ขอบเขตคำตอบแคบลงเรื่อยๆ จนกระทั่งได้ความถูกต้องตามค่าความคลาดเคลื่อน (Tolerance) ที่ยอมรับได้

```python
# ตัวอย่าง 4.3: การหาค่ารากที่สองของ x โดยใช้วิธี Bisection Search
x = 25.0
epsilon = 0.00001 # ความคลาดเคลื่อนที่ยอมรับได้
low = 0.0
high = max(1.0, x)
ans = (high + low) / 2.0

# TODO

print("Square root of", x, "is approximately", ans)
```
