## 4.6 for e in list (การวนลูปทีละสมาชิกในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การใช้คำสั่ง `for` ท่องไปในลิสต์ ตัวแปรลูปจะเข้าถึงสมาชิกแต่ละตัวในลิสต์ทีละธาตุเรียงตามลำดับ Index จากซ้ายไปขวา


```python
# ตัวอย่าง 4.6: คัดกรองเฉพาะจำนวนคู่จากลิสต์ของตัวเลข
numbers = [12, 7, 9, 20, 15, 30]
even_numbers = []

for e in numbers:
    if e % 2 == 0:
        even_numbers.append(e)

print("จำนวนคู่ในลิสต์:", even_numbers)
```

### ตัวอย่างเพิ่มเติมสำหรับการวนลูปใน List

#### ตัวอย่าง 4.6.1: การวนลูปใน Sub-list (Slicing)

**คำอธิบาย:** ตัดเฉพาะช่วงข้อมูลของลิสต์ด้วย Slicing เช่น `x[2:-2]` แล้วใช้ `for e in ...` เพื่อท่องไปในแต่ละสมาชิกของช่วงที่ตัดออกมา


```python
# ตัวอย่าง 4.6.1: วนลูปในช่วงสมาชิกที่ต้องการโดยใช้ Slicing
x = [1, 3, 4, 5, 7, 8]
for e in x[2:-2]:
    print(e)
```

#### ตัวอย่าง 4.6.2: อ่านและแสดงรหัสสินค้าจาก input

**คำอธิบาย:** รับรายการรหัสสินค้าแยกด้วยช่องว่าง เช่น `"A1023 P223 A392"` นำมาแยกด้วย `.split()` เป็นลิสต์ แล้วแสดงรหัสสินค้าทีละบรรทัดจากซ้ายไปขวา


```python
# ตัวอย่าง 4.6.2: อ่านรายการรหัสสินค้าและแสดงทีละบรรทัด
prod_ids = input().split()
for pid in prod_ids:
    print(pid)
```

#### ตัวอย่าง 4.6.3: แสดงรายการสินค้าจากขวาไปซ้าย (Reverse List Slicing)

**คำอธิบาย:** รับข้อมูลรูปแบบ `"Office Product: A1023 P223 A392"` แยกด้วย `":"` เพื่อเอารายการรหัสสินค้ามาแยกด้วย `.split()` แล้ววนลูปกลับทิศทางจากขวามาซ้ายด้วย `prod_ids[::-1]`


```python
# ตัวอย่าง 4.6.3: แสดงเฉพาะรหัสสินค้าย้อนกลับจากขวาไปซ้าย
x = input().split(":")
prod_ids = x[1].split()
for pid in prod_ids[::-1]:
    print(pid)
```

#### ตัวอย่าง 4.6.4: นับจำนวนสมาชิกที่ตรงกับค่า a ในลิสต์ x

**คำอธิบาย:** รับบรรทัดแรกแยกด้วย `.split()` เก็บในลิสต์ `x` อ่านบรรทัดที่สองเก็บใน `a` จากนั้นวนลูปนับว่ามีค่าใน `a` ซ้ำอยู่ในลิสต์ `x` ทั้งหมดกี่ตัว


```python
# ตัวอย่าง 4.6.4: นับจำนวนครั้งที่พบค่า a ในลิสต์ x
x = input().split()
a = input()
c = 0
for e in x:
    if e == a:
        c += 1
print(c)
```
