## 04-01: การหาค่าเฉลี่ยของจำนวนเต็ม

จงเขียนโปรแกรมรับจำนวนเต็มทีละจำนวนจากผู้ใช้ไปเรื่อยๆ จนกว่าผู้ใช้จะป้อนค่า -1 จากนั้นคำนวณและแสดงค่าเฉลี่ยของจำนวนทั้งหมดที่ป้อนเข้ามา (ไม่นับ -1)

### ข้อมูลนำเข้า
จำนวนเต็มหลายบรรทัด บรรทัดละ 1 จำนวน โดยจบด้วย -1

### ข้อมูลส่งออก
ค่าเฉลี่ยเป็นจำนวนจริง ทศนิยม 2 ตำแหน่ง (หากไม่มีข้อมูลป้อนเข้าเลยนอกจาก -1 ให้แสดง No data)

```python
# 04-01: การหาค่าเฉลี่ยของจำนวนเต็ม
total = 0
count = 0

while True:
    n = int(input())
    if n == -1:
        break
    total += n
    count += 1

if count > 0:
    print(f"{total / count:.2f}")
else:
    print("No data")
```

## 04-02: การนับและรวมเฉพาะจำนวนคู่ในช่วง range

จงเขียนโปรแกรมรับจำนวนเต็มสองจำนวน `a` และ `b` จากนั้นคำนวณหาผลรวมของจำนวนคู่ทั้งหมดในช่วงตั้งแต่ `a` ถึง `b` (รวมทั้ง `a` และ `b` หากเป็นจำนวนคู่)

### ข้อมูลนำเข้า
จำนวนเต็ม 2 จำนวน คั่นด้วยช่องว่าง

### ข้อมูลส่งออก
ผลรวมของจำนวนคู่

```python
# 04-02: ผลรวมจำนวนคู่ในช่วง
a, b = [int(x) for x in input().split()]
sum_even = 0

for i in range(a, b + 1):
    if i % 2 == 0:
        sum_even += i

print(sum_even)
```
