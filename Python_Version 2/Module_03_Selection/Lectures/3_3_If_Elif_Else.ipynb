# 3.3 คำสั่ง if และ if-elif-else

## แนวคิด (Concept)

### if Statement (คำสั่ง if)

- ใช้สำหรับเลือกทำคำสั่งตามเงื่อนไข
- ถ้าเงื่อนไขเป็นจริง จะทำคำสั่งภายใต้ if
- ถ้าเงื่อนไขเป็นเท็จ จะข้ามไป

### if-elif-else Statement

- ใช้เมื่อมีหลายทางเลือก
- `elif` (else if) ใช้เช็คเงื่อนไขเพิ่มเติม
- `else` ทำเมื่อไม่ตรงเงื่อนไขใดเลย
- Python จะตรวจสอบเงื่อนไขตามลำดับจากบนลงล่าง
- เมื่อเจอเงื่อนไขที่เป็นจริง จะทำคำสั่งนั้นแล้วออกจาก if-elif-else ทันที

## ตัวอย่างที่ 1: คำนวณจำนวนวันในรอบปี (Leap Year)

```python
# ตัวอย่าง: คำนวณจำนวนวันในรอบปี (ปีอธิกสุรทิน)
# เดือนกุมภาพันธ์มี 29 วัน (ปีนั้นมี 366 วัน) ก็ต่อเมื่อ:
# - y หารด้วย 400 ลงตัว หรือ
# - y หารด้วย 4 ลงตัว แต่ต้องหารด้วย 100 ไม่ลงตัว

y = int(input("ใส่ปี ค.ศ.: "))
days_in_year = 365

if (y % 400 == 0) or \
   (y % 4 == 0 and y % 100 != 0):
    days_in_year = 366

print("จำนวนวันในปี", y, "=", days_in_year, "วัน")
```

## ตัวอย่างที่ 2: ซื้อสินค้า 4 ชิ้น (ฟรีชิ้นที่ถูกที่สุด)

```python
# รับราคาสินค้า 4 ชิ้น หาชิ้นที่ราคาถูกที่สุด
# แล้วคำนวณราคารวมหลังหักส่วนลด

p1 = int(input("ราคาชิ้นที่ 1: "))
min_p = p1

p2 = int(input("ราคาชิ้นที่ 2: "))
if p2 < min_p:
    min_p = p2

p3 = int(input("ราคาชิ้นที่ 3: "))
if p3 < min_p:
    min_p = p3

p4 = int(input("ราคาชิ้นที่ 4: "))
if p4 < min_p:
    min_p = p4

print("ราคาถูกที่สุด:", min_p)
print("ราคารวม:", p1 + p2 + p3 + p4 - min_p)
```

## ตัวอย่างที่ 3: นับจำนวนสระใน String

```python
# มีตัวแปร t เก็บสตริงอยู่แล้ว โดยที่ t มีตัวอักษรภายใน 5 ตัว
# จงแสดงจำนวนตัวใน t ที่เป็นสระ (นับทั้งตัวเล็กและตัวใหญ่)

t = "Annie"

c = 0
if t[0] in "aeiouAEIOU":
    c += 1
if t[1] in "aeiouAEIOU":
    c += 1
if t[2] in "aeiouAEIOU":
    c += 1
if t[3] in "aeiouAEIOU":
    c += 1
if t[4] in "aeiouAEIOU":
    c += 1

print("จำนวนสระ:", c)
```

## ตัวอย่างที่ 4: if-elif-else - แปลงรหัสวัน

```python
# เปรียบเทียบรหัสวัน w (0=เสาร์, 1=อาทิตย์, ..., 5=พฤหัสบดี, อื่นๆ=ศุกร์)
w = int(input("ใส่รหัสวัน (0-6): "))

if w == 0:
    print("เสาร์")
elif w == 1:
    print("อาทิตย์")
elif w == 2:
    print("จันทร์")
elif w == 3:
    print("อังคาร")
elif w == 4:
    print("พุธ")
elif w == 5:
    print("พฤหัสบดี")
else:
    print("ศุกร์")
```

## ตัวอย่างที่ 6: จำแนกเกณฑ์ BMI

```python
# การจำแนกเกณฑ์ดัชนีมวลกาย (BMI Category)
bmi = float(input("ใส่ค่า BMI: "))

if bmi <= 15:
    msg = "Very severely underweight"
elif bmi <= 16:
    msg = "Severely underweight"
elif bmi <= 18.5:
    msg = "Underweight"
elif bmi <= 25:
    msg = "Normal"
elif bmi <= 30:
    msg = "Overweight"
else:
    msg = "Obese"

print("ระดับ BMI:", msg)
```

## ตัวอย่างที่ 8: เปรียบเทียบ if-else ซ้อน กับ if-elif-else

```python
# แบบที่ 1: if-else ซ้อนกัน (ยาว อ่านยาก)
score = 85

if score >= 80:
    grade = "A"
else:
    if score >= 70:
        grade = "B"
    else:
        if score >= 60:
            grade = "C"
        else:
            if score >= 50:
                grade = "D"
            else:
                grade = "F"

print("แบบซ้อน - เกรด:", grade)
```

```python
# แบบที่ 2: if-elif-else (สั้น อ่านง่าย)
score = 85

if score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
elif score >= 50:
    grade = "D"
else:
    grade = "F"

print("แบบ elif - เกรด:", grade)
```

## ข้อควรระวัง (Common Pitfalls)

1. **ลำดับของเงื่อนไข**
   - Python ตรวจสอบเงื่อนไขตามลำดับจากบนลงล่าง
   - ควรเรียงเงื่อนไขจากเฉพาะเจาะจงไปทั่วไป

2. **การใช้ elif กับ if แยกกัน**
   - `elif`: ตรวจสอบต่อเนื่องกัน ถ้าเจอเงื่อนไขจริงจะหยุด
   - `if` แยก: ตรวจสอบทุกเงื่อนไข แม้เจอเงื่อนไขจริงแล้ว

3. **ลืมใส่ else**
   - ถ้าไม่มี else และไม่มีเงื่อนไขใดเป็นจริง จะไม่ทำอะไรเลย
   - ควรมี else เพื่อจัดการกรณีที่ไม่คาดคิด

4. **เงื่อนไขที่ทับซ้อนกัน**
   - ตัวอย่าง: `if score >= 80:` แล้วตามด้วย `elif score >= 70:`
   - ถ้าเรียงผิด อาจไม่ได้ผลลัพธ์ที่ต้องการ

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: ลำดับเงื่อนไขผิด
score = 85

# แบบนี้ไม่ถูกต้อง
# if score >= 50:
#     grade = "D"  # จะได้ D แทนที่จะเป็น A
# elif score >= 80:
#     grade = "A"

# ถูก: เรียงจากมากไปน้อย
if score >= 80:
    grade = "A"
elif score >= 50:
    grade = "D"
else:
    grade = "F"

print("เกรด:", grade)
```

```python
# ผิด: ใช้ if แทน elif (ตรวจสอบทุกเงื่อนไข)
x = 15

# แบบนี้จะทำทุกเงื่อนไขที่เป็นจริง
if x > 10:
    print("x มากกว่า 10")
if x > 5:  # ยังตรวจสอบต่อแม้เงื่อนไขแรกเป็นจริง
    print("x มากกว่า 5")

print()

# ถูก: ใช้ elif (หยุดเมื่อเจอเงื่อนไขจริง)
if x > 10:
    print("x มากกว่า 10")
elif x > 5:  # ไม่ทำเพราะเงื่อนไขแรกเป็นจริงแล้ว
    print("x มากกว่า 5")
```

## สรุป

### if Statement:
- ใช้เลือกทำคำสั่งตามเงื่อนไข
- สามารถใช้หลาย if แยกกันได้
- แต่ละ if จะตรวจสอบเงื่อนไขอิสระกัน

### if-elif-else:
- ใช้เมื่อมีหลายทางเลือก
- ตรวจสอบเงื่อนไขตามลำดับจากบนลงล่าง
- เมื่อเจอเงื่อนไขจริง ทำคำสั่งแล้วออกทันที
- `else` จับกรณีที่ไม่ตรงเงื่อนไขใดเลย

### ข้อแตกต่าง:
- **หลาย if**: ตรวจสอบทุกเงื่อนไข
- **if-elif-else**: หยุดเมื่อเจอเงื่อนไขจริงตัวแรก

### เมื่อไหนใช้อะไร:
- ใช้ **หลาย if** เมื่อต้องการตรวจสอบหลายเงื่อนไขอิสระกัน
- ใช้ **if-elif-else** เมื่อมีทางเลือกหลายทาง แต่เลือกได้แค่ทางเดียว
