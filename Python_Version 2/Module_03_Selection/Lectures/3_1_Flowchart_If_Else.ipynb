## 3.1 Flowchart & if-else (ผังงานและการตัดสินใจ)

### ผังงาน : บรรยายขั้นตอนการทำงาน

![ผังงานคำนวณ BMI](images/bmi_flowchart.png)

**ขั้นตอนการทำงานของผังงานคำนวณ BMI:**
1. **`start`** : เริ่มต้นการทำงานโปรแกรม
2. **`weight`** : รับค่าข้อมูลน้ำหนัก (หน่วยเป็นกิโลกรัม) จากแป้นพิมพ์
3. **`height`** : รับค่าข้อมูลส่วนสูง (หน่วยเป็นเซนติเมตร) จากแป้นพิมพ์
4. **`คำนวณ`** :
   - `hm = height / 100` (แปลงส่วนสูงจากเซนติเมตรเป็นเมตร)
   - `bmi = weight / (hm ** 2)` (คำนวณดัชนีมวลกาย BMI)
5. **`bmi`** : แสดงผลลัพธ์ค่า BMI ออกทางจอภาพ
6. **`stop`** : สิ้นสุดการทำงาน

จากผังงานข้างต้น โปรแกรมจะรับข้อมูลน้ำหนัก (weight) และส่วนสูง (height) จากแป้นพิมพ์ จากนั้นคำนวณหาดัชนีมวลกาย (BMI) แล้วแสดงผลลัพธ์ทางจอภาพ

```python
# ตัวอย่าง 3.1.1: โปรแกรมคำนวณ BMI ตามผังงาน
weight = float(input())
height = float(input())
hm = height / 100
bmi = weight / (hm ** 2)
print(bmi)
```

### ตัวอย่าง: การใช้ผังงานเปรียบเทียบค่าเพื่อหา Min และ Max

![ผังงานหาค่า Min และ Max](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_01.png)

**ขั้นตอนการทำงาน:**
1. รับค่าจำนวนเต็ม `a` และ `b` จากแป้นพิมพ์
2. ตรวจสอบเงื่อนไข `a < b`:
   - ถ้าเป็น **True (จริง)** : ให้ `min = a` และ `max = b`
   - ถ้าเป็น **False (เท็จ)** : ให้ `min = b` และ `max = a`
3. แสดงผลลัพธ์ `min, max` ออกทางจอภาพ

```python
# ตัวอย่าง 3.1.2: โปรแกรมหาค่า min และ max ตามผังงาน
a = int(input())
b = int(input())

if a < b:
    min_val = a
    max_val = b
else:
    min_val = b
    max_val = a

print(min_val, max_val)
```

### แบบฝึกหัด 3-1 ข้อที่ 2

![ผังงานสลับค่าหรือเพิ่มค่า](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_02.png)

เขียนโปรแกรมตามผังงานข้างต้น (มีตัวแปร a และ b ให้แล้ว ไม่ต้องตั้งค่าหรือรับอินพุตใดๆ)

```python
# แบบฝึกหัด 3-1 ข้อที่ 2
a = 5
b = 2

if abs(a - b) > 2:
    a, b = b, a
else:
    a = b ** 2 + a

print(a, b)
```

### ตัวอย่างผังงานแบบเงื่อนไขซ้อน (Nested Flowchart 1)

![ผังงานเงื่อนไขซ้อน 1](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_03.png)

เขียนโปรแกรมตามผังงานข้างต้น (มีตัวแปร a ให้แล้ว ไม่ต้องตั้งค่าหรือรับอินพุตใดๆ)

```python
# ตัวอย่างผังงานแบบเงื่อนไขซ้อน 1
a = 12

if a % 2 == 0:
    if a > 10:
        a //= 2
    else:
        a = 0
else:
    a *= 2

print(a)
```

### ตัวอย่างผังงานแบบเงื่อนไขซ้อน (Nested Flowchart 2)

![ผังงานเงื่อนไขซ้อน 2](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_04.png)

เขียนโปรแกรมตามผังงานข้างต้น (มีตัวแปร a ให้แล้ว ไม่ต้องตั้งค่าหรือรับอินพุตใดๆ)

```python
# ตัวอย่างผังงานแบบเงื่อนไขซ้อน 2
a = 12

if a % 2 == 0:
    b = 2 * a
    if a > 10:
        a //= 2
    else:
        b += a
    a += b
else:
    a *= 2

print(a, b)
```
