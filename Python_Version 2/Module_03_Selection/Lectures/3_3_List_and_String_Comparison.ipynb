## 3.3 List and String Comparison (การเปรียบเทียบ List และ String)

### การเปรียบเทียบสตริงกับสตริง (String Comparison)

- **อักษรภาษาอังกฤษตัวเล็กมีค่ามากกว่าตัวใหญ่:**
  - `"A" < "Z"` เป็นจริง (`True`)
  - `"A" < "B" < "Z" < "a" < "b" < "z"` เป็นจริง (`True`)
- **สตริงตัวเลขมีค่าน้อยไปมากตามลำดับ 0 ถึง 9:**
  - `"0" < "1" < "2" < "3" < "4" < "9"` เป็นจริง (`True`)
- **เปรียบเทียบสตริงจะเปรียบเทียบทีละตัวอักษรจากซ้ายไปขวา (Lexicographical Order):**
  - `"ABC" < "aA"` เป็นจริง (`True`) — เพราะอักขระตัวแรก `'A' < 'a'`
  - `"ABC" < "ACAA"` เป็นจริง (`True`) — เพราะตัวแรกเท่ากัน แต่ตัวที่สอง `'B' < 'C'`
  - `"ABC" < "ABCC"` เป็นจริง (`True`) — เพราะ 3 ตัวแรกเท่ากัน แต่สตริงทางซ้ายยาวสั้นกว่า
  - `"100" < "19"` เป็นจริง (`True`) — เพราะตัวแรกเท่ากัน แต่ตัวที่สอง `'0' < '9'` (เทียบตามรหัสอักขระ ไม่ใช่เทียบค่าทางคณิตศาสตร์)

```python
# ตัวอย่างการเปรียบเทียบสตริง
print('"A" < "Z":', "A" < "Z")
print('"ABC" < "aA":', "ABC" < "aA")
print('"100" < "19":', "100" < "19")

# ตัวอย่างการใช้นิพจน์เปรียบเทียบตัวอักษร ตัวเลข และวันที่
x = "K"
birth_date = [2542, 5, 15] # [y, m, d]

# ถ้า x เป็นตัวอักษรอังกฤษ
if "a" <= x <= "z" or "A" <= x <= "Z":
    print("x เป็นตัวอักษรอังกฤษ")

# ถ้า x เป็นตัวเลข
if "0" <= x <= "9":
    print("x เป็นตัวเลข")

# ถ้าเกิดหลัง 31 ก.ย. 2540 ( birth_date = [y, m, d] )
if birth_date > [2540, 9, 31]:
    print("เกิดหลัง 31 ก.ย. 2540")
```

### แบบฝึกหัด 3-3

แบบฝึกหัดการใช้นิพจน์เปรียบเทียบ สตริง และ ลิสต์

```python
# แบบฝึกหัด 3-3 ข้อที่ 1:
# แสดงผลของการเปรียบเทียบว่า s เป็นสตริงที่ขึ้นต้นด้วยตัวอักษรอังกฤษตัวใหญ่และลงท้ายด้วยตัวเลข
s = "Python3"
print("A" <= s[0] <= "Z" and "0" <= s[-1] <= "9")
```

```python
# แบบฝึกหัด 3-3 ข้อที่ 2:
# มีตัวแปร d1, m1, y1 กับ d2, m2, y2 ให้แล้ว เก็บวันเดือนปีเกิดของ Tom กับ Jerry ตามลำดับ
# จงแสดงว่า Tom หรือ Jerry ใครเกิดก่อน (ให้ถือว่าไม่มีกรณีเกิดวันเดือนปีเดียวกัน)
d1, m1, y1 = 15, 4, 2541  # วันเกิด Tom
d2, m2, y2 = 20, 8, 2540  # วันเกิด Jerry

if [y1, m1, d1] < [y2, m2, d2]:
    print("Tom")
else:
    print("Jerry")
```
