## 3.4 Membership Operator (in) (ตัวดำเนินการตรวจสอบความเป็นสมาชิก)

### รูปแบบ `if a in b` : a อยู่ใน b ?

1. **ถ้า a กับ b เป็นสตริง:**
   - `a in b` หาว่า a เป็นสตริงย่อยของ b หรือไม่
   - `"python" in "I love python"` เป็นจริง (`True`)
   - `"python" in "I love Python"` เป็นเท็จ (`False`)

2. **ถ้า b เป็นลิสต์:**
   - `a in b` หาว่า a เป็นสมาชิกในลิสต์ b หรือไม่
   - `6 in [2,4,6,8,10]` เป็นจริง (`True`)
   - `"Bones" in ["Kirk", "McCoy", "Spock"]` เป็นเท็จ (`False`)

3. **การปฏิเสธเงื่อนไข (`not in`):**
   - `if not (a in b):` # ถ้า a ไม่อยู่ใน b
   - `if a not in b:` # ถ้า a ไม่อยู่ใน b

```python
# ตัวอย่าง 3.4.1: การใช้ in และ not in ตรวจสอบสมาชิกใน สตริง และ ลิสต์
print('"python" in "I love python":', "python" in "I love python")
print('"python" in "I love Python":', "python" in "I love Python")
print('6 in [2,4,6,8,10]:', 6 in [2, 4, 6, 8, 10])
print('"Bones" in ["Kirk", "McCoy", "Spock"]:', "Bones" in ["Kirk", "McCoy", "Spock"])
```
