## 3.2 Relational operators (ตัวดำเนินการเปรียบเทียบ)

### นิพจน์บูลีน: กำหนดเงื่อนไข

- ได้ผลลัพธ์เป็น `True` หรือ `False`
- ใช้เครื่องหมาย `<` `>` `<=` `>=` `!=` `==` เพื่อเปรียบเทียบว่า `<` `>` `≤` `≥` `≠` `เท่ากัน`
- เชื่อมการเปรียบเทียบได้ด้วย `and`, `or`
- กลับผลการเปรียบเทียบได้ด้วย `not`
- ถ้ามีทั้ง `and`, `or`, `not` จะทำ `not` ก่อน แล้วทำ `and` แล้วทำ `or`

```python
# ตัวอย่างการใช้ and, or, not กำหนดเงื่อนไขเกี่ยวกับเดือน
m = 4

# เช็คเดือนที่ลงท้ายด้วย "ยน" (เมษายน=4, มิถุนาย=6, กันยายน=9, พฤศจิกายน=11)
if m == 4 or m == 6 or m == 9 or m == 11:
    print('ลงท้ายด้วย "ยน"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ not ร่วมกับ or
if not (m == 2 or m == 4 or m == 6 or m == 9 or m == 11):
    print('ลงท้ายด้วย "คม"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ and ร่วมกับ !=
if m != 2 and m != 4 and m != 6 and m != 9 and m != 11:
    print('ลงท้ายด้วย "คม"')
```

### ตัวอย่างการใช้นิพจน์เงื่อนไขในรูปแบบต่างๆ



```python
# ตัวอย่างการใช้นิพจน์เงื่อนไข
x = 10
m = 5
y = 2024
student_id = "6431234521"
tel_no = "021234567"

if x % 2 == 0:     # ถ้า x เป็นจำนวนคู่
    print("x เป็นจำนวนคู่")

if 1 <= m <= 12:    # ถ้า m มีค่าตั้งแต่ 1 ถึง 12
    print("m มีค่าตั้งแต่ 1 ถึง 12")

if y % 400 == 0:    # ถ้า y หารด้วย 400 ลงตัว
    print("y หารด้วย 400 ลงตัว")

if student_id[-2:] == "21": # ถ้าเป็นนิสิตวิศวฯ
    print("เป็นนิสิตวิศวฯ")

if tel_no[:2] == "02":       # ถ้าเป็นโทรศัพท์ใน กทม
    print("เป็นโทรศัพท์ใน กทม")
```
