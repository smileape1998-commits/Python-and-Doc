# 3.2 ตัวดำเนินการเปรียบเทียบ (Comparison Operators)

## แนวคิด (Concept)

### Relational Operators (ตัวดำเนินการเปรียบเทียบความสัมพันธ์)

- ใช้เปรียบเทียบค่า 2 ค่า ได้ผลลัพธ์เป็น `True` หรือ `False`
- เครื่องหมายเปรียบเทียบ:
  - `<` น้อยกว่า
  - `>` มากกว่า
  - `<=` น้อยกว่าหรือเท่ากับ
  - `>=` มากกว่าหรือเท่ากับ
  - `==` เท่ากับ
  - `!=` ไม่เท่ากับ

### Logical Operators (ตัวดำเนินการตรรกะ)

- `and`: เป็นจริงทั้งสองฝั่ง
- `or`: เป็นจริงอย่างน้อยหนึ่งฝั่ง
- `not`: กลับค่าความจริง

### Membership Operator

- `in`: ตรวจสอบว่าเป็นสมาชิกหรือไม่
- `not in`: ตรวจสอบว่าไม่เป็นสมาชิก

## ตัวอย่างที่ 1: Relational Operators พื้นฐาน

```python
# ตัวอย่างการเปรียบเทียบตัวเลข
a = 10
b = 5

print("a > b:", a > b)    # True
print("a < b:", a < b)    # False
print("a == b:", a == b)  # False
print("a != b:", a != b)  # True
print("a >= 10:", a >= 10) # True
print("b <= 5:", b <= 5)  # True
```

## ตัวอย่างที่ 2: Logical Operators (and, or, not)

```python
# ตัวอย่างการใช้ and, or, not
m = 4  # เดือนเมษายน

# เช็คเดือนที่ลงท้ายด้วย "ยน" (เมษายน=4, มิถุนายน=6, กันยายน=9, พฤศจิกายน=11)
if m == 4 or m == 6 or m == 9 or m == 11:
    print('ลงท้ายด้วย "ยน"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ not
if not (m == 2 or m == 4 or m == 6 or m == 9 or m == 11):
    print('ลงท้ายด้วย "คม"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ and
if m != 2 and m != 4 and m != 6 and m != 9 and m != 11:
    print('ลงท้ายด้วย "คม"')
```

## ตัวอย่างที่ 3: การเปรียบเทียบช่วงค่า

```python
# ตัวอย่างการเปรียบเทียบช่วงค่า
x = 10
m = 5
y = 2024

# ถ้า x เป็นจำนวนคู่
if x % 2 == 0:
    print("x เป็นจำนวนคู่")

# ถ้า m มีค่าตั้งแต่ 1 ถึง 12
if 1 <= m <= 12:
    print("m มีค่าตั้งแต่ 1 ถึง 12")

# ถ้า y หารด้วย 400 ลงตัว
if y % 400 == 0:
    print("y หารด้วย 400 ลงตัว")
```

## ตัวอย่างที่ 4: การเปรียบเทียบ String

```python
# การเปรียบเทียบ String แบบ Lexicographical Order
print('"A" < "Z":', "A" < "Z")      # True
print('"ABC" < "aA":', "ABC" < "aA") # True (เพราะ 'A' < 'a')
print('"100" < "19":', "100" < "19") # True (เพราะ '0' < '9')

# ตัวอักษรภาษาอังกฤษ
# A-Z < a-z
print('"a" > "Z":', "a" > "Z")      # True
```

## ตัวอย่างที่ 5: การเปรียบเทียบ List

```python
# การเปรียบเทียบ List เทียบทีละตัว
birth_date = [2542, 5, 15]  # [ปี, เดือน, วัน]

# ถ้าเกิดหลัง 31 ก.ย. 2540
if birth_date > [2540, 9, 31]:
    print("เกิดหลัง 31 ก.ย. 2540")

# เปรียบเทียบวันเกิด 2 คน
d1, m1, y1 = 15, 4, 2541  # Tom
d2, m2, y2 = 20, 8, 2540  # Jerry

if [y1, m1, d1] < [y2, m2, d2]:
    print("Tom เกิดก่อน")
else:
    print("Jerry เกิดก่อน")
```

## ตัวอย่างที่ 6: Membership Operator (in)

```python
# การใช้ in กับ String
print('"python" in "I love python":', "python" in "I love python")  # True
print('"python" in "I love Python":', "python" in "I love Python")  # False (case-sensitive)

# การใช้ in กับ List
print('6 in [2,4,6,8,10]:', 6 in [2, 4, 6, 8, 10])  # True
print('7 in [2,4,6,8,10]:', 7 in [2, 4, 6, 8, 10])  # False
```

## ตัวอย่างที่ 7: ตรวจสอบสระภาษาอังกฤษ

```python
# ตัวอย่างการใช้ in ตรวจสอบสระ
x = "e"

# วิธีที่ 1: ใช้ or (ยาว)
if x == "a" or x == "e" or x == "i" or \
   x == "o" or x == "u" or x == "A" or \
   x == "E" or x == "I" or x == "O" or x == "U":
    print("x เป็นสระ (ใช้ or)")

# วิธีที่ 2: ใช้ in กับ String (สั้นกว่า)
if x in "aeiouAEIOU":
    print("x เป็นสระ (ใช้ in)")

# วิธีที่ 3: ใช้ in กับ List
if x in ["a", "e", "i", "o", "u", "A", "E", "I", "O", "U"]:
    print("x เป็นสระ (ใช้ List)")
```

## ตัวอย่างที่ 8: การใช้ not in

```python
# การใช้ not in
text = "I love Python programming"

# ตรวจสอบว่าไม่มีคำว่า "Java"
if "Java" not in text:
    print("ไม่มีคำว่า Java ในข้อความ")

# ตรวจสอบว่าเดือนไม่ใช่เดือนที่มี 31 วัน
month = 2
months_31_days = [1, 3, 5, 7, 8, 10, 12]

if month not in months_31_days:
    print("เดือนนี้ไม่มี 31 วัน")
```

## ตัวอย่างที่ 9: นิพจน์เงื่อนไขเชิงซ้อน

```python
# ตัวอย่างนิพจน์เงื่อนไขที่ซับซ้อน
student_id = "6431234521"
tel_no = "021234567"
s = "Python3"

# ถ้าเป็นนิสิตวิศวะ (รหัสลงท้ายด้วย 21)
if student_id[-2:] == "21":
    print("เป็นนิสิตวิศวะ")

# ถ้าเป็นโทรศัพท์ใน กทม (ขึ้นต้นด้วย 02)
if tel_no[:2] == "02":
    print("เป็นโทรศัพท์ใน กทม")

# ถ้า s ขึ้นต้นด้วยตัวใหญ่และลงท้ายด้วยตัวเลข
if "A" <= s[0] <= "Z" and "0" <= s[-1] <= "9":
    print("s ขึ้นต้นด้วยตัวใหญ่และลงท้ายด้วยตัวเลข")
```

## ข้อควรระวัง (Common Pitfalls)

1. **สับสนระหว่าง == และ =**
   - `==` ใช้เปรียบเทียบ
   - `=` ใช้กำหนดค่า

2. **การเปรียบเทียบ String เป็น case-sensitive**
   - `"Python" == "python"` เป็น False
   - ใช้ `.lower()` หรือ `.upper()` ถ้าต้องการเปรียบเทียบไม่สนใจตัวพิมพ์

3. **การเปรียบเทียบ String เป็น lexicographical**
   - `"100" < "19"` เป็น True (เปรียบเทียบตัวอักษร ไม่ใช่ค่าตัวเลข)
   - ต้องแปลงเป็น int ก่อนถ้าต้องการเปรียบเทียบค่าตัวเลข

4. **ลำดับความสำคัญของ operator**
   - `not` > `and` > `or`
   - ใช้วงเล็บเพื่อความชัดเจน

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: ใช้ = แทน ==
# if x = 10:  # SyntaxError

# ถูก:
x = 10
if x == 10:
    print("x เท่ากับ 10")

# ผิด: เปรียบเทียบ String เป็นตัวเลข
# if "100" > "19":  # True (ไม่ใช่ผลลัพธ์ที่คาดหวัง)

# ถูก: แปลงเป็น int ก่อน
if int("100") > int("19"):
    print("100 > 19")

# ผิด: ลำดับ operator
# if x > 5 and y < 10 or z == 0  # อาจเข้าใจผิด

# ถูก: ใช้วงเล็บ
x, y, z = 10, 5, 0
if (x > 5 and y < 10) or z == 0:
    print("เงื่อนไขเป็นจริง")
```

## สรุป

### Relational Operators:
- `<`, `>`, `<=`, `>=`, `==`, `!=`
- ได้ผลลัพธ์เป็น True หรือ False

### Logical Operators:
- `and`: ทั้งสองเป็นจริง
- `or`: อย่างน้อยหนึ่งเป็นจริง
- `not`: กลับค่า
- ลำดับ: not > and > or

### String Comparison:
- เปรียบเทียบแบบ lexicographical order
- A-Z < a-z
- Case-sensitive

### List Comparison:
- เปรียบเทียบทีละสมาชิกจากซ้ายไปขวา

### Membership Operator:
- `in`: ตรวจสอบความเป็นสมาชิก
- `not in`: ตรวจสอบว่าไม่เป็นสมาชิก
- ใช้กับ String และ List ได้
