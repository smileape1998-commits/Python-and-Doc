## 3.1 Flowchart & if-else (ผังงานและการตัดสินใจ)

### ผังงาน : บรรยายขั้นตอนการทำงาน

![ผังงานคำนวณ BMI](images/bmi_flowchart.png)

**ขั้นตอนการทำงานของผังงานคำนวณ BMI:**
1. **`start`** : เริ่มต้นการทำงานโปรแกรม
2. **`weight`** : รับค่าข้อมูลน้ำหนัก (หน่วยเป็นกิโลกรัม) จากแป้นพิมพ์
3. **`height`** : รับค่าข้อมูลส่วนสูง (หน่วยเป็นเซนติเมตร) จากแป้นพิมพ์
4. **`คำนวณ`** :
   - `hm = height / 100` (แปลงส่วนสูงจากเซนติเมตรเป็นเมตร)
   - `bmi = weight / (hm ** 2)` (คำนวณดัชนีมวลกาย BMI)
5. **`bmi`** : แสดงผลลัพธ์ค่า BMI ออกทางจอภาพ
6. **`stop`** : สิ้นสุดการทำงาน

จากผังงานข้างต้น โปรแกรมจะรับข้อมูลน้ำหนัก (weight) และส่วนสูง (height) จากแป้นพิมพ์ จากนั้นคำนวณหาดัชนีมวลกาย (BMI) แล้วแสดงผลลัพธ์ทางจอภาพ

```python
# ตัวอย่าง 3.1.1: โปรแกรมคำนวณ BMI ตามผังงาน
weight = float(input())
height = float(input())
hm = height / 100
bmi = weight / (hm ** 2)
print(bmi)
```

### ตัวอย่าง: การใช้ผังงานเปรียบเทียบค่าเพื่อหา Min และ Max

![ผังงานหาค่า Min และ Max](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_01.png)

**ขั้นตอนการทำงาน:**
1. รับค่าจำนวนเต็ม `a` และ `b` จากแป้นพิมพ์
2. ตรวจสอบเงื่อนไข `a < b`:
   - ถ้าเป็น **True (จริง)** : ให้ `min = a` และ `max = b`
   - ถ้าเป็น **False (เท็จ)** : ให้ `min = b` และ `max = a`
3. แสดงผลลัพธ์ `min, max` ออกทางจอภาพ

```python
# ตัวอย่าง 3.1.2: โปรแกรมหาค่า min และ max ตามผังงาน
a = int(input())
b = int(input())

if a < b:
    min_val = a
    max_val = b
else:
    min_val = b
    max_val = a

print(min_val, max_val)
```

### ตัวอย่างผังงานแบบเงื่อนไขซ้อน (Nested Flowchart 1)

![ผังงานเงื่อนไขซ้อน 1](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_03.png)

เขียนโปรแกรมตามผังงานข้างต้น (มีตัวแปร a ให้แล้ว ไม่ต้องตั้งค่าหรือรับอินพุตใดๆ)

```python
# ตัวอย่างผังงานแบบเงื่อนไขซ้อน 1
a = 12

if a % 2 == 0:
    if a > 10:
        a //= 2
    else:
        a = 0
else:
    a *= 2

print(a)
```

### ตัวอย่างผังงานแบบเงื่อนไขซ้อน (Nested Flowchart 2)

![ผังงานเงื่อนไขซ้อน 2](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_01_04.png)

เขียนโปรแกรมตามผังงานข้างต้น (มีตัวแปร a ให้แล้ว ไม่ต้องตั้งค่าหรือรับอินพุตใดๆ)

```python
# ตัวอย่างผังงานแบบเงื่อนไขซ้อน 2
a = 12

if a % 2 == 0:
    b = 2 * a
    if a > 10:
        a //= 2
    else:
        b += a
    a += b
else:
    a *= 2

print(a, b)
```

## 3.2 Relational operators (ตัวดำเนินการเปรียบเทียบ)

### นิพจน์บูลีน: กำหนดเงื่อนไข

![นิพจน์บูลีน กำหนดเงื่อนไข](images/boolean_expressions.png)

- ได้ผลลัพธ์เป็น `True` หรือ `False`
- ใช้เครื่องหมาย `<` `>` `<=` `>=` `!=` `==` เพื่อเปรียบเทียบว่า `<` `>` `≤` `≥` `≠` `เท่ากัน`
- เชื่อมการเปรียบเทียบได้ด้วย `and`, `or`
- กลับผลการเปรียบเทียบได้ด้วย `not`
- ถ้ามีทั้ง `and`, `or`, `not` จะทำ `not` ก่อน แล้วทำ `and` แล้วทำ `or`

```python
# ตัวอย่างการใช้ and, or, not กำหนดเงื่อนไขเกี่ยวกับเดือน
m = 4

# เช็คเดือนที่ลงท้ายด้วย "ยน" (เมษายน=4, มิถุนาย=6, กันยายน=9, พฤศจิกายน=11)
if m == 4 or m == 6 or m == 9 or m == 11:
    print('ลงท้ายด้วย "ยน"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ not ร่วมกับ or
if not (m == 2 or m == 4 or m == 6 or m == 9 or m == 11):
    print('ลงท้ายด้วย "คม"')

# เช็คเดือนที่ลงท้ายด้วย "คม" โดยใช้ and ร่วมกับ !=
if m != 2 and m != 4 and m != 6 and m != 9 and m != 11:
    print('ลงท้ายด้วย "คม"')
```

## 3.3 List and String Comparison (การเปรียบเทียบ List และ String)

### การเปรียบเทียบสตริงกับสตริง (String Comparison)

- **อักษรภาษาอังกฤษตัวเล็กมีค่ามากกว่าตัวใหญ่:**
  - `"A" < "Z"` เป็นจริง (`True`)
  - `"A" < "B" < "Z" < "a" < "b" < "z"` เป็นจริง (`True`)
- **สตริงตัวเลขมีค่าน้อยไปมากตามลำดับ 0 ถึง 9:**
  - `"0" < "1" < "2" < "3" < "4" < "9"` เป็นจริง (`True`)
- **เปรียบเทียบสตริงจะเปรียบเทียบทีละตัวอักษรจากซ้ายไปขวา (Lexicographical Order):**
  - `"ABC" < "aA"` เป็นจริง (`True`) — เพราะอักขระตัวแรก `'A' < 'a'`
  - `"ABC" < "ACAA"` เป็นจริง (`True`) — เพราะตัวแรกเท่ากัน แต่ตัวที่สอง `'B' < 'C'`
  - `"ABC" < "ABCC"` เป็นจริง (`True`) — เพราะ 3 ตัวแรกเท่ากัน แต่สตริงทางซ้ายยาวสั้นกว่า
  - `"100" < "19"` เป็นจริง (`True`) — เพราะตัวแรกเท่ากัน แต่ตัวที่สอง `'0' < '9'` (เทียบตามรหัสอักขระ ไม่ใช่เทียบค่าทางคณิตศาสตร์)

```python
# ตัวอย่างการเปรียบเทียบสตริง
print('"A" < "Z":', "A" < "Z")
print('"ABC" < "aA":', "ABC" < "aA")
print('"100" < "19":', "100" < "19")

# ตัวอย่างการใช้นิพจน์เปรียบเทียบตัวอักษร ตัวเลข และวันที่
x = "K"
birth_date = [2542, 5, 15] # [y, m, d]

# ถ้า x เป็นตัวอักษรอังกฤษ
if "a" <= x <= "z" or "A" <= x <= "Z":
    print("x เป็นตัวอักษรอังกฤษ")

# ถ้า x เป็นตัวเลข
if "0" <= x <= "9":
    print("x เป็นตัวเลข")

# ถ้าเกิดหลัง 31 ก.ย. 2540 ( birth_date = [y, m, d] )
if birth_date > [2540, 9, 31]:
    print("เกิดหลัง 31 ก.ย. 2540")
```

## 3.4 Membership Operator (in) (ตัวดำเนินการตรวจสอบความเป็นสมาชิก)

### รูปแบบ `if a in b` : a อยู่ใน b ?

1. **ถ้า a กับ b เป็นสตริง:**
   - `a in b` หาว่า a เป็นสตริงย่อยของ b หรือไม่
   - `"python" in "I love python"` เป็นจริง (`True`)
   - `"python" in "I love Python"` เป็นเท็จ (`False`)

2. **ถ้า b เป็นลิสต์:**
   - `a in b` หาว่า a เป็นสมาชิกในลิสต์ b หรือไม่
   - `6 in [2,4,6,8,10]` เป็นจริง (`True`)
   - `"Bones" in ["Kirk", "McCoy", "Spock"]` เป็นเท็จ (`False`)

3. **การปฏิเสธเงื่อนไข (`not in`):**
   - `if not (a in b):` # ถ้า a ไม่อยู่ใน b
   - `if a not in b:` # ถ้า a ไม่อยู่ใน b

```python
# ตัวอย่าง 3.4.1: การใช้ in และ not in ตรวจสอบสมาชิกใน สตริง และ ลิสต์
print('"python" in "I love python":', "python" in "I love python")
print('"python" in "I love Python":', "python" in "I love Python")
print('6 in [2,4,6,8,10]:', 6 in [2, 4, 6, 8, 10])
print('"Bones" in ["Kirk", "McCoy", "Spock"]:', "Bones" in ["Kirk", "McCoy", "Spock"])
```

## 3.5 if Statement (คำสั่ง if)

### ตัวอย่าง: คำนวณจำนวนวันในรอบปี (ปีอธิกสุรทิน / Leap Year)

- ให้ `y` เก็บเลขปี ค.ศ.
- เดือนกุมภาพันธ์มี 29 วัน (ปีนั้นมี 366 วัน) ก็ต่อเมื่อ:
  - `y` หารด้วย 400 ลงตัว หรือ
  - `y` หารด้วย 4 ลงตัว แต่ต้องหารด้วย 100 ไม่ลงตัว
  - เช่น ปี 2004, 2000 (ปี 2100 ไม่เข้าเงื่อนไข)

```python
# ตัวอย่าง 3.5.1: คำนวณจำนวนวันในรอบปี (Leap Year)
y = int(input())
days_in_year = 365

if (y % 400 == 0) or \
   (y % 4 == 0 and y % 100 != 0):
    days_in_year = 366

print(days_in_year)
```

### ตัวอย่าง: โปรแกรมซื้อสินค้า 4 ชิ้น (ชิ้นที่ถูกที่สุดได้ฟรี)

รับราคาสินค้า 4 ชิ้น `p1, p2, p3, p4` จากผู้ใช้ หาชิ้นที่ราคาถูกที่สุด (`min_p`) แล้วคำนวณราคารวมทั้งหมดหลังหักส่วนลดสินค้าถูกที่สุดออก

```python
# ตัวอย่าง 3.5.2: ซื้อสินค้า 4 ชิ้น ฟรีชิ้นที่ถูกที่สุด
p1 = int(input())
min_p = p1
p2 = int(input())
if p2 < min_p:
    min_p = p2
p3 = int(input())
if p3 < min_p:
    min_p = p3
p4 = int(input())
if p4 < min_p:
    min_p = p4

print("min =", min_p)
print("Total =", p1 + p2 + p3 + p4 - min_p)
```

## 3.6 if-elif-elif-else (คำสั่งเงื่อนไขหลายทางเลือก)

### เปรียบเทียบโครงสร้าง `if - else - if - else` กับ `if - elif - else`

การเขียน `if-else` ซ้อนกันหลายๆ ชั้น จะทำให้โค้ดยาวและอ่านยาก ในภาษา Python จึงมีคำสั่ง `elif` (ย่อมาจาก else if) เพื่อช่วยให้โครงสร้างโค้ดสั้น กระชับ และอ่านง่ายยิ่งขึ้น

```python
# ตัวอย่าง 3.6.1: เปรียบเทียบรหัสวัน w (0=เสาร์, 1=อาทิตย์, ..., 5=พฤหัสบดี, อื่นๆ=ศุกร์)
w = 2

# การใช้ if-elif-else เพื่อหาชื่อวัน
if w == 0:
    print("เสาร์")
elif w == 1:
    print("อาทิตย์")
elif w == 2:
    print("จันทร์")
elif w == 3:
    print("อังคาร")
elif w == 4:
    print("พุธ")
elif w == 5:
    print("พฤหัสบดี")
else:
    print("ศุกร์")
```

### ตัวอย่าง: แปลงรหัสสกุลเงิน (Currency Code)

```python
# ตัวอย่าง 3.6.2: แปลงรหัสสกุลเงิน
cur = input()

if cur == "AUD":
    fn = "Australia Dollar"
elif cur == "GBP":
    fn = "Great Britain Pound"
elif cur == "EUR":
    fn = "Euro"
else:
    fn = "???"

print(fn)
```

### ตัวอย่าง: การจำแนกเกณฑ์ดัชนีมวลกาย (BMI Category)
### แบบฝึกหัด 3-6


![แบบฝึกหัด 3-6 ข้อที่ 2](https://www.cp.eng.chula.ac.th/~somchai/python101/images/ex03_06_02.png)

```python
# ตัวอย่าง 3.6.3: การจำแนกเกณฑ์ BMI
bmi = float(input())

if bmi <= 15:
    msg = "Very severely underweight"
elif bmi <= 16:
    msg = "Severely underweight"
elif bmi <= 18.5:
    msg = "Underweight"
elif bmi <= 25:
    msg = "Normal"
elif bmi <= 30:
    msg = "Overweight"
else:
    msg = "Obese"

print(msg)
```
