# 1.8 ลำดับการทำงานของตัวดำเนินการ (Operator Precedence)

## แนวคิด (Concept)

เมื่อมีตัวดำเนินการหลายตัวใน expression เดียว Python จะทำงานตามลำดับความสำคัญ (Precedence)

**ลำดับความสำคัญ (สูงสุดไปต่ำสุด):**
1. `()` วงเล็บ
2. `**` ยกกำลัง
3. `*`, `/`, `//`, `%` คูณ หาร
4. `+`, `-` บวก ลบ

**หลักการจำ:**
- วงเล็บก่อนเสมอ
- คูณหารก่อนบวกลบ
- ถ้าลำดับเท่ากัน ทำจากซ้ายไปขวา
- ยกกำลัง (`**`) ทำจากขวาไปซ้าย

## ตัวอย่างที่ 1: Operator Precedence พื้นฐาน

```python
# คูณก่อน แล้วค่อยบวก
result1 = 2 + 3 * 4
print(result1)  # 14 ไม่ใช่ 20 (3*4=12, แล้ว 2+12=14)

# วงเล็บก่อน แล้วค่อยคูณ
result2 = (2 + 3) * 4
print(result2)  # 20 (2+3=5, แล้ว 5*4=20)

# ลำดับเท่ากัน ทำซ้ายไปขวา
result3 = 10 - 3 + 2
print(result3)  # 9 (10-3=7, แล้ว 7+2=9)

# ยกกำลังทำจากขวาไปซ้าย
result4 = 2 ** 3 ** 2
print(result4)  # 512 (3²=9, แล้ว 2⁹=512)
```

## ตัวอย่างที่ 2: ตัวอย่างที่ซับซ้อน

```python
# ตัวอย่างที่ซับซ้อน
x = 5
y = 2

# ไม่มีวงเล็บ
result1 = x + y * 3 - 4 / 2
print("x + y * 3 - 4 / 2 =", result1)
# ลำดับ: y*3=6, 4/2=2, 5+6=11, 11-2=9.0

# ใช้วงเล็บเพื่อควบคุมลำดับ
result2 = (x + y) * (3 - 4) / 2
print("(x + y) * (3 - 4) / 2 =", result2)
# ลำดับ: (5+2)=7, (3-4)=-1, 7*(-1)=-7, -7/2=-3.5
```

## ตัวอย่างที่ 3: การคำนวณตามสูตรทางคณิตศาสตร์

```python
# สูตร: (a + b) / (c - d)
a, b, c, d = 10, 5, 8, 3
result = (a + b) / (c - d)
print("(a + b) / (c - d) =", result)  # (15) / (5) = 3.0

# สูตร: a² + b²
a, b = 3, 4
result = a ** 2 + b ** 2
print("a² + b² =", result)  # 9 + 16 = 25

# สูตร: √(a² + b²)
import math
result = math.sqrt(a ** 2 + b ** 2)
print("√(a² + b²) =", result)  # 5.0
```

## ตัวอย่างที่ 4: Modulo และ Division

```python
# Modulo และ division มีลำดับเท่ากับ multiplication
result1 = 100 // 3 % 5
print("100 // 3 % 5 =", result1)
# ลำดับ: 100//3=33, 33%5=3

result2 = 20 % 3 * 2
print("20 % 3 * 2 =", result2)
# ลำดับ: 20%3=2, 2*2=4

result3 = 20 * 3 % 5
print("20 * 3 % 5 =", result3)
# ลำดับ: 20*3=60, 60%5=0
```

## ข้อควรระวัง (Common Pitfalls)

```python
# ผิด: คิดว่าบวกก่อนคูณ
wrong = 2 + 3 * 4  # คิดว่าได้ 20 แต่จริงๆ ได้ 14
print("ไม่ใช้วงเล็บ:", wrong)

# ถูก: ใช้วงเล็บเพื่อให้ชัดเจน
correct = (2 + 3) * 4  # ได้ 20 ตามต้องการ
print("ใช้วงเล็บ:", correct)

# ผิด: ยกกำลังทำจากซ้ายไปขวา
# 2 ** 3 ** 2 ≠ (2 ** 3) ** 2
result1 = 2 ** 3 ** 2  # 2⁹ = 512
result2 = (2 ** 3) ** 2  # 8² = 64
print("2 ** 3 ** 2 =", result1)
print("(2 ** 3) ** 2 =", result2)
```

## เคล็ดลับ (Tips)

1. **เมื่อสงสัย ใช้วงเล็บเสมอ** - ทำให้โค้ดอ่านง่ายและไม่เกิดข้อผิดพลาด
2. **เขียนให้ชัดเจน** - แม้ว่าจะรู้ลำดับ precedence ก็ตาม การใช้วงเล็บช่วยให้คนอื่นเข้าใจโค้ดได้ง่ายขึ้น
3. **แยกคำนวณซับซ้อน** - ถ้า expression ยาวและซับซ้อน แยกออกเป็นหลายบรรทัด

```python
# แทนที่จะเขียนแบบนี้ (ยากต่อการอ่าน)
result = (a + b) * (c - d) / (e + f) ** 2

# เขียนแบบนี้ดีกว่า (อ่านง่าย เข้าใจง่าย)
numerator = (a + b) * (c - d)
denominator = (e + f) ** 2
result = numerator / denominator
```

## สรุป

### ลำดับความสำคัญ (Operator Precedence):
1. `()` - วงเล็บ (ลำดับสูงสุด)
2. `**` - ยกกำลัง (ทำจากขวาไปซ้าย)
3. `*`, `/`, `//`, `%` - คูณ หาร (ทำจากซ้ายไปขวา)
4. `+`, `-` - บวก ลบ (ทำจากซ้ายไปขวา)

### หลักการจำง่ายๆ:
- **วงเล็บก่อนเสมอ**
- **คูณหารก่อนบวกลบ**
- **เมื่อสงสัย ใส่วงเล็บ**
