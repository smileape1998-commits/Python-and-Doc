# 1.3 ตัวแปร (Variable)

## แนวคิด (Concept)

### ตัวแปรคืออะไร?
- **ตัวแปร** คือชื่อที่ใช้เก็บข้อมูล (ต้องมีชื่อกำกับ)
- ใช้เปลี่ยนค่าในที่เก็บได้
- เปรียบเสมือนกล่องที่มีชื่อติดอยู่ เราสามารถนำข้อมูลเข้าไปเก็บ และเรียกใช้ภายหลังได้

### การกำหนดค่า (Assignment)
- รูปแบบ: `variable_name = value`
- เครื่องหมาย `=` ใช้กำหนดค่าให้ตัวแปร (ไม่ใช่เครื่องหมายเท่ากับทางคณิตศาสตร์)

## ตัวอย่างที่ 1: การสร้างและใช้งานตัวแปร

````python
# ตัวอย่างการกำหนดค่า
name = "Python"      # นำค่า "Python" เก็บในตัวแปร name
lang = name          # นำข้อมูลใน name เก็บในตัวแปร lang
                     # ทำให้ lang กับ name เก็บค่าเหมือนกัน

print("name =", name)
print("lang =", lang)
````

## ตัวอย่างที่ 2: การเปลี่ยนค่าตัวแปร

````python
# การกำหนดและเปลี่ยนค่าตัวแปร
a = 1              # a = 1
b = 2              # a = 1; b = 2
c = a              # a = 1; b = 2; c = 1
d = c + b          # a = 1; b = 2; c = 1; d = 3
d = d + 5          # a = 1; b = 2; c = 1; d = 8

print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)
````

## กฎการตั้งชื่อตัวแปร

### ข้อควรปฏิบัติ:
1. **ประกอบด้วยอักษร ตัวเลข หรือเครื่องหมาย `_` เท่านั้น**
2. **ตัวอักษรใหญ่กับเล็กไม่เหมือนกัน** (Case-sensitive)
   - `first_name` กับ `First_Name` เป็นคนละตัวแปร
3. **ห้ามขึ้นต้นชื่อด้วยตัวเลข**
4. **ไม่ใช่ชื่อกับคำสงวนของภาษา** (Reserved Keywords)

### ตัวอย่างชื่อตัวแปร:
```python
first_name = "John"     # ถูกต้อง: ใช้ underscore คั่นคำ
last_name = "Wick"      # ถูกต้อง
```

## คำสงวนของ Python (Reserved Keywords)

คำเหล่านี้เป็นคำสงวนของ Python **ห้ามใช้เป็นชื่อตัวแปร**:

```
and        as         assert     break      class
continue   def        del        elif       else
except     exec       finally    for        from
global     if         import     in         is
lambda     nonlocal   not        or         pass
raise      return     try        while      with
yield      True       False      None
```

## ตัวอย่างที่ 3: การใช้งานตัวแปรจริง

````python
# การสร้างและกำหนดค่าตัวแปร
name = "สมชาย"
age = 20
height = 175.5

# ใช้งานตัวแปร
print("ชื่อ:", name)
print("อายุ:", age)
print("ส่วนสูง:", height, "cm")

# เปลี่ยนค่าตัวแปร
age = 21
print("อายุใหม่:", age)
````

## ข้อควรระวัง (Common Pitfalls)

````python
# ผิด: ขึ้นต้นด้วยตัวเลข
# 1name = "John"  # Error: SyntaxError

# ผิด: มีช่องว่าง
# student name = "Mary"  # Error: SyntaxError

# ผิด: ใช้คำสงวน
# print = 10  # Error: จะทำให้ฟังก์ชัน print() ใช้ไม่ได้

# ถูกต้อง
name1 = "John"
student_name = "Mary"
_value = 10

print(name1, student_name, _value)
````

## สรุป

- ตัวแปรใช้เก็บข้อมูลในหน่วยความจำ
- ใช้เครื่องหมาย `=` เพื่อกำหนดค่าให้ตัวแปร
- ชื่อตัวแปรต้องขึ้นต้นด้วยตัวอักษรหรือ `_`
- Python เป็น case-sensitive (แยกตัวพิมพ์เล็กใหญ่)
- ห้ามใช้คำสงวนของ Python เป็นชื่อตัวแปร
- ตั้งชื่อตัวแปรให้มีความหมายชัดเจน ใช้ `_` คั่นคำ (snake_case)
