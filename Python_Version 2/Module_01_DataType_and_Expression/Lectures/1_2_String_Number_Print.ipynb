# 1.2 String, Number และ print()

## แนวคิด (Concept)

### String (สตริง)
- ข้อความที่อยู่ในเครื่องหมาย `""` หรือ `''`
- ใช้แสดงข้อความ ตัวอักษร คำ หรือประโยค
- ตัวอย่าง: `"Hello"`, `'Hello Python'`, `"12345"`, `""` (String ว่าง)

### Number (ตัวเลข)
- **Integer (int)**: เลขจำนวนเต็ม เช่น `1234`, `0`, `-100`
- **Float (Floating Point)**: เลขทศนิยม เช่น `-1234.0`, `3.14`, `0.0`
- **Scientific Notation**: การเขียนตัวเลขแบบยกกำลัง 10
  - รูปแบบ: `aEb` หมายถึง a × 10ᵇ
  - ตัวอย่าง: `1.5E-15` = 1.5 × 10⁻¹⁵ (ตัวเลขเล็กมาก)
  - ตัวอย่าง: `1.963e34` = 1.963 × 10³⁴ (ตัวเลขใหญ่มาก)

### print() Function
- ใช้แสดงผลข้อมูลออกทางหน้าจอ
- สามารถแสดงได้หลายค่าโดยคั่นด้วยเครื่องหมาย comma

## ตัวอย่างที่ 1: String และ Number พื้นฐาน

```python
# String ตัวอย่าง
print("สวัสดี")
print('Python Programming')

# Number ตัวอย่าง
print(42)          # Integer
print(3.14159)     # Float
print(-100)        # Integer ติดลบ

# Scientific Notation (การเขียนตัวเลขแบบยกกำลัง 10)
print(1.963e34)    # 1.963 × 10³⁴ (ตัวเลขขนาดใหญ่มาก)
print(1.5e-15)     # 1.5 × 10⁻¹⁵ (ตัวเลขขนาดเล็กมาก)

# แสดงหลายค่าพร้อมกัน
print("อายุ:", 25, "ปี")
print("ผลคะแนน:", 95.5)
```

## ตัวอย่างที่ 2: การต่อ String (String Concatenation)

```python
# การต่อ String ด้วยเครื่องหมาย +
print("Hello" + " " + "!")
print("1" + "1")  # ได้ "11" ไม่ใช่ 2
```

## ตัวอย่างที่ 3: การบวก Number

```python
# การบวกตัวเลข
print(1 + 1)      # ได้ 2
print(1 + 1.0)    # ได้ 2.0 (int + float = float)
```

## ข้อควรระวัง (Common Pitfalls)

1. **เลขที่อยู่ในเครื่องหมายคำพูดคือ String ไม่ใช่ตัวเลข**
   - `"100"` คือ String ไม่สามารถนำมาคำนวณได้โดยตรง
   
2. **ไม่สามารถบวก String กับ Number โดยตรง**
   - `"1" + 1` จะเกิด TypeError
   - ต้องแปลงชนิดข้อมูลให้ตรงกันก่อน

## ตัวอย่างข้อผิดพลาดที่พบบ่อย

```python
# ผิด: "100" คือ String ไม่สามารถคำนวณได้
# print("100" + 50)  # จะเกิด Error: TypeError

# ผิด: ไม่สามารถบวก String กับ Number
# print("1" + 1)  # จะเกิด Error: TypeError

# ถูก: แปลง String เป็นตัวเลขก่อน
print(int("100") + 50)  # ได้ 150
```

## สรุป

- String คือข้อความที่อยู่ใน `""` หรือ `''`
- Number มี 2 ชนิดหลัก: int (จำนวนเต็ม) และ float (ทศนิยม)
- Scientific Notation ใช้เขียนตัวเลขขนาดใหญ่หรือเล็กมาก (เช่น 1.5e-15)
- print() ใช้แสดงผลข้อมูล สามารถแสดงหลายค่าพร้อมกันได้
- String + String = ต่อข้อความ, Number + Number = บวกเลข
- ห้ามบวก String กับ Number โดยตรง
