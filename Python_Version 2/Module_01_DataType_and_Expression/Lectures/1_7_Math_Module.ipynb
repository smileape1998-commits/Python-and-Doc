# 1.7 Math Module และ Built-in Functions

## แนวคิด (Concept)

### Math Module

**Math Module** เป็นโมดูลสำหรับฟังก์ชันทางคณิตศาสตร์ขั้นสูง

**การใช้งาน:**
1. Import โมดูล: `import math`
2. เรียกใช้ฟังก์ชัน: `math.function_name()`

**ฟังก์ชันที่ใช้บ่อย:**
- `math.sqrt(x)`: รากที่สอง
- `math.pow(x, y)`: ยกกำลัง
- `math.floor(x)`: ปัดเศษลง
- `math.ceil(x)`: ปัดเศษขึ้น
- `math.sin(x)`, `math.cos(x)`: ฟังก์ชันตรีโกณมิติ (รับค่าเป็น radian)
- `math.radians(x)`: แปลงองศาเป็น radian
- `math.log(x, base)`: ลอการิทึม
- `math.pi`: ค่าคงที่ pi (3.14159...)
- `math.e`: ค่าคงที่ e (2.71828...)

## ตัวอย่างที่ 1: การใช้ Math Module พื้นฐาน

```python
# นำเข้า math module
import math

# รากที่สอง
print("รากที่สองของ 16:", math.sqrt(16))  # 4.0

# ยกกำลัง
print("2 ยกกำลัง 3:", math.pow(2, 3))  # 8.0

# ปัดเศษลง
print("ปัดเศษลง 3.7:", math.floor(3.7))  # 3

# ปัดเศษขึ้น
print("ปัดเศษขึ้น 3.2:", math.ceil(3.2))  # 4

# ค่าคงที่
print("ค่า pi:", math.pi)  # 3.14159...
print("ค่า e:", math.e)    # 2.71828...
```

## ตัวอย่างที่ 2: การแปลงองศาเป็น Radian และคำนวณตรีโกณมิติ

```python
import math

# แปลงองศาเป็น radian
degree = float(input("ใส่องศา: "))

# วิธีที่ 1: คำนวณเอง
radian = degree * 3.14159 / 180

# วิธีที่ 2: ใช้ค่า pi จาก math
radian = degree * math.pi / 180

# วิธีที่ 3: ใช้ฟังก์ชัน radians (แนะนำ)
radian = math.radians(degree)

# คำนวณ sin และ cos
s = math.sin(radian)
c = math.cos(radian)

print("sin:", s)
print("cos:", c)
```

## ตัวอย่างที่ 3: การใช้ Logarithm

```python
import math

# ลอการิทึมฐาน 10
g = math.log(1E100, 10)  # log₁₀(10¹⁰⁰) = 100.0
print("log₁₀(10¹⁰⁰) =", g)

# ลอการิทึมธรรมชาติ (ฐาน e)
ln = math.log(math.e)  # ln(e) = 1.0
print("ln(e) =", ln)

# ลอการิทึมฐาน 2
log2 = math.log(8, 2)  # log₂(8) = 3.0
print("log₂(8) =", log2)
```

## Built-in Functions (ฟังก์ชันพื้นฐาน)

Python มีฟังก์ชันพื้นฐานที่สามารถใช้งานได้ทันที **ไม่ต้อง import**

### ตารางฟังก์ชันที่ใช้บ่อย:

| ฟังก์ชัน | ความหมาย | ตัวอย่าง |
|---------|---------|----------|
| `abs(x)` | ค่าสัมบูรณ์ | `abs(-2)` → `2` |
| `round(x, n)` | ปัดเศษทศนิยม n ตำแหน่ง | `round(2/3, 2)` → `0.67` |
| `max(list)` | หาค่ามากสุด | `max([4,1,5,3])` → `5` |
| `min(list)` | หาค่าน้อยสุด | `min([4,1,5,3])` → `1` |
| `sum(list)` | ผลรวมของข้อมูลใน list | `sum([4,1,5,3])` → `13` |
| `len(list)` | จำนวนข้อมูลใน list | `len([4,1,5,3])` → `4` |
| `str(x)` | แปลงเป็น string | `str(1234)` → `"1234"` |
| `int(x)` | แปลงเป็น integer | `int("123")` → `123` |
| `float(x)` | แปลงเป็น float | `float("-123.4")` → `-123.4` |
| `input()` | รับข้อมูลจากผู้ใช้ | `k = input()` |
| `print()` | แสดงผลออกหน้าจอ | `print(a, b, c)` |

## ตัวอย่างที่ 4: การใช้ Built-in Functions

```python
# ค่าสัมบูรณ์
print("abs(-10) =", abs(-10))  # 10
print("abs(5) =", abs(5))      # 5

# ปัดเศษ
print("round(3.14159, 2) =", round(3.14159, 2))  # 3.14
print("round(2.7) =", round(2.7))  # 3

# ทำงานกับ list
numbers = [4, 1, 5, 3, 9, 2]
print("max:", max(numbers))  # 9
print("min:", min(numbers))  # 1
print("sum:", sum(numbers))  # 24
print("len:", len(numbers))  # 6
```

## ตัวอย่างที่ 5: การใช้งานจริง - คำนวณพื้นที่วงกลม

```python
import math

# คำนวณพื้นที่วงกลม A = πr²
radius = float(input("ใส่รัศมีวงกลม: "))
area = math.pi * math.pow(radius, 2)
print("พื้นที่วงกลม:", area)

# หรือใช้ ** แทน pow()
area = math.pi * radius ** 2
print("พื้นที่วงกลม:", area)
```

## ตัวอย่างที่ 6: การคำนวณด้านตรงข้ามมุมฉาก (Pythagorean Theorem)

```python
import math

# สูตร: c² = a² + b²
# หรือ c = √(a² + b²)
a = 3
b = 4
c = math.sqrt(a**2 + b**2)
print("ด้านตรงข้ามมุมฉาก:", c)  # 5.0
```

## ข้อควรระวัง (Common Pitfalls)

```python
# ผิด: ลืม import module
# print(math.sqrt(25))  # NameError: name 'math' is not defined

# ถูก: import ก่อนใช้งาน
import math
print(math.sqrt(25))

# ผิด: ใช้ sqrt() โดยตรงโดยไม่มี math.
# print(sqrt(25))  # NameError

# ถูก: ต้องใช้ math.sqrt()
print(math.sqrt(25))
```

## สรุป

### Math Module:
- ต้อง `import math` ก่อนใช้งาน
- เรียกใช้ฟังก์ชันด้วย `math.function_name()`
- ฟังก์ชันที่ใช้บ่อย: sqrt(), pow(), sin(), cos(), floor(), ceil()
- ค่าคงที่: math.pi, math.e

### Built-in Functions:
- ใช้ได้ทันที ไม่ต้อง import
- ฟังก์ชันที่ใช้บ่อย: abs(), round(), max(), min(), sum(), len()
- ฟังก์ชันแปลงชนิด: int(), float(), str(), bool()
- ฟังก์ชัน I/O: input(), print()
