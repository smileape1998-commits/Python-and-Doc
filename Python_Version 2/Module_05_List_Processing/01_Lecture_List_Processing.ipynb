## 5.1 Basic List Operation (การดำเนินการพื้นฐานกับลิสต์)

### อธิบายแนวคิด (Concept Explanation)
ลิสต์ (List) คือกลุ่มของข้อมูลที่เรียงต่อกันเป็นลำดับ โดยสมาชิกแต่ละตัวจะมีดัชนี (Index) กำหนดตำแหน่ง เริ่มต้นจากตำแหน่งที่ 0 ลิสต์เป็นข้อมูลประเภท mutable หมายความว่าเราสามารถเปลี่ยนแปลง เพิ่ม หรือลบข้อมูลภายในลิสต์ได้หลังจากสร้างขึ้นมาแล้ว

การดำเนินการพื้นฐานกับลิสต์ ได้แก่:
1. การสร้างลิสต์ด้วยวงเล็บเหลี่ยม `[]` เช่น `x = []` หรือลิสต์ซ้อนลิสต์ `q = [x, y, 8, z]`
2. การหาจำนวนสมาชิกด้วยฟังก์ชัน `len()`
3. การเข้าถึงด้วยดัชนีและการตัดช่วง Slicing เช่น `z[1]`, `y[1:3]`
4. การทำซ้ำลิสต์ด้วยเครื่องหมาย `*` เช่น `a = [0] * 10`
5. การเชื่อมต่อลิสต์ด้วยเครื่องหมาย `+` เช่น `a = a + a`
6. การแก้ไขสมาชิกผ่านดัชนี `a[i] = i` และการตรวจสอบสมาชิกด้วยตัวดำเนินการ `in` 

### ข้อควรระวัง (Common Pitfalls)
- การเข้าถึงตำแหน่งที่ไม่มีอยู่จริงจะทำให้เกิดข้อผิดพลาด `IndexError: list index out of range` 
- การแก้ไขสมาชิกผ่านดัชนี สามารถทำได้เฉพาะตำแหน่งที่มีสมาชิกอยู่แล้วเท่านั้น ไม่สามารถแก้ไขดัชนีที่เกินจากขนาดของลิสต์เพื่อเพิ่มสมาชิกได้

## 5.2 List Method (เมธอดของลิสต์)

### อธิบายแนวคิด (Concept Explanation)
เมธอด (Method) คือฟังก์ชันการทำงานประจำตัวแปรลิสต์ ช่วยให้จัดการข้อมูลภายในลิสต์ได้สะดวกยิ่งขึ้น เมธอดพื้นฐานที่สำคัญได้แก่:

- `append(x)` : เพิ่มสมาชิก `x` ต่อท้ายลิสต์
- `remove(x)` : ลบสมาชิกตัวแรกที่มีค่าเท่ากับ `x` ออกจากลิสต์
- `insert(i, x)` : แทรกสมาชิก `x` ลงในตำแหน่งดัชนี `i`
- `pop(i)` : ลบและคืนค่าสมาชิกในตำแหน่งดัชนี `i` (หากไม่ระบุ `i` จะลบตัวสุดท้าย)
- `index(x)` : คืนค่าดัชนีตำแหน่งแรกที่พบ `x` ในลิสต์
- `sort()` : จัดเรียงสมาชิกในลิสต์จากน้อยไปมาก
- `x in d` : ตัวดำเนินการตรวจสอบว่ามี `x` อยู่ในลิสต์ `d` หรือไม่ (คืนค่า `True`/`False`)

## 5.3 Reading Data into a List (การรับข้อมูลเข้าสู่ลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การรับข้อมูลเข้าสู่ลิสต์สามารถทำได้หลายรูปแบบ ขึ้นอยู่กับลักษณะข้อมูลนำเข้า (Input Format):

1. **การรับข้อมูลทีละบรรทัดตามจำนวนที่ระบุ (`n` บรรทัด):** อ่านจำนวนข้อมูล `n` บรรทัดแรก แล้วใช้ลูป `range(n)` เพื่อรับข้อมูลทีละบรรทัด สามารถเลือกเก็บต่อท้ายลิสต์ด้วย `append()` หรือแทรกด้านหน้าลิสต์ด้วย `insert(0, e)` เพื่อเรียงย้อนกลับ
2. **การรับข้อมูลทีละบรรทัดจนกว่าจะพบค่าหยุด (Sentinel Value):** ใช้ลูป `while` รับข้อมูลทีละบรรทัดต่อเนื่อง จนกระทั่งผู้ใช้ป้อนคำสั่งหรือค่าหยุดที่กำหนด เช่น คำว่า `"end"` 
3. **การรับข้อมูลหลายสมาชิกในบรรทัดเดียว (Single-line Input):** ใช้ `input().split()` เพื่อแยกข้อมูลด้วยช่องว่าง แล้วใช้ลูปแปลงประเภทข้อมูล (เช่น `int` หรือ `float`)
4. **การรับข้อมูลที่มีรูปแบบสลับกัน (Interleaved Pattern):** ใช้ Slicing เช่น `x[::2]` เพื่อดึงสมาชิกดัชนีคู่ และ `x[1::2]` เพื่อดึงสมาชิกดัชนีคี่

### ข้อควรระวัง (Common Pitfalls)
- ข้อมูลที่ได้จาก `input()` หรือ `input().split()` จะมีประเภทข้อมูลเป็นสตริง (String) เสมอ หากต้องการคำนวณ ต้องแปลงเป็น `int` หรือ `float` ก่อนนำไปใช้
- ในการใช้ลูป `while` รับข้อมูลจนกว่าจะเจอค่าหยุด ต้องมั่นใจว่ามีการอัปเดตอ่านค่า `input()` ใหม่ภายในลูปทุกครั้งเพื่อป้องกันการเกิดลูปไม่รู้จบ (Infinite Loop)

2. **Index-based Iteration (`for i in range(len(d)):`):** วนลูปตามดัชนีตำแหน่ง เหมาะเมื่อต้องการใช้ทั้งดัชนีตำแหน่ง `i` และข้อมูล `d[i]` หรือเมื่อต้องการแก้ไขค่าสมาชิกภายในลิสต์เดิม

### ข้อควรระวัง (Common Pitfalls)
- **อย่าใช้วิธีวนลูปแบบ `for e in d: i = d.index(e)`:** การค้นหาตำแหน่ง `index(e)` ใหม่ทุกรอบจะทำงานช้ามาก ให้ใช้แบบดัชนี `for i in range(len(d)):` แทน
- **การแก้ไขสมาชิกผ่านตัวแปรลูป `e` จะไม่มีผลกับลิสต์เดิม:** เช่น `for e in d: if e < 0: e *= -1` ค่าใน `e` จะเปลี่ยนแต่ค่าในลิสต์ `d` จะคงเดิม หากต้องการแก้ไขค่าในลิสต์เดิม ต้องอ้างอิงตำแหน่งผ่านดัชนี `d[i] *= -1`

## 5.7 List Sorting (การเรียงลำดับข้อมูลในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การเรียงลำดับข้อมูลในภาษา Python มี 2 วิธีหลักที่มีลักษณะการทำงานแตกต่างกัน:

1. **`list.sort()` (In-place Sort):** จัดเรียงข้อมูลในตัวแปรลิสต์เดิมทันที ไม่สร้างลิสต์ใหม่ และไม่คืนค่ากลับมา (`None`)
2. **`sorted(iterable)` (Function Sort):** รับลิสต์เข้าไปเรียงลำดับ แล้วคืนค่าเป็นลิสต์ใหม่ โดยที่ลิสต์เดิมไม่เปลี่ยนแปลง

### การเรียงลำดับลิสต์ซ้อนลิสต์ (Nested List / 2D List Sorting)
- เมื่อสั่ง `.sort()` กับลิสต์ 2 มิติ Python จะเปรียบเทียบสมาชิกตัวแรก (`[0]`) ก่อน หากมีค่าเท่ากันจะเปรียบเทียบสมาชิกตัวถัดไป (`[1]`, `[2]`) ตามลำดับ เช่น `[[2541, 7, 12], [2538, 9, 8]]` จะเรียงตาม ปี, เดือน, วัน
- **เทคนิคกำหนดตัวเรียงลำดับ (Key Selection):** หากต้องการเรียงลำดับข้อมูลตามค่าใด (เช่น ราคา หรือ คะแนน) ให้นำค่านั้นมาวางไว้เป็นสมาชิกตัวแรกของลิสต์ย่อย `[sort_key, value]`

### ข้อควรระวัง (Common Pitfalls)
- **ประเภทข้อมูลตัวเลขเทียบกับสตริง:** หากป้อนตัวเลขแบบสตริง เช่น `"172.0"` กับ `"2.0"` การเรียงลำดับจะเรียงตามพจนานุกรม (ทำให้อักษร `'1'` มาก่อน `'2'`) หากต้องการเรียงตามค่าตัวเลขจริง ต้องแปลงเป็น `float` หรือ `int` ก่อนนำใส่ลิสต์

```python
# 5.7.1 ตัวอย่างการเรียงลำดับลิสต์ 1D และ 2D List (Nested List)

# 1. เรียงลำดับตัวเลข 1D List
x1 = [9, 4, 3, 2, 5, 0, 44, 53, 2, 1]
x1.sort()
print("x1 หลัง sort():", x1)

# 2. เรียงลำดับสตริงตามพจนานุกรม 1D List
x2 = ["Jisoo", "Jennie", "Rosé", "Lisa"]
x2.sort()
print("x2 หลัง sort():", x2)

# 3. เรียงลำดับ 2D List (เรียงตามสมาชิกตัวแรก [0], แล้ว [1], [2] ตามลำดับ)
x3 = [[2541, 7, 12], [2538, 9, 8], [2538, 9, 30], [2541, 4, 29]]
x3.sort()
print("x3 (เรียงตาม ปี, เดือน, วัน):", x3)

# 4. เรียงลำดับ 2D List โดยนำค่าหลัก (Key) ไว้ตำแหน่งแรก [length, name]
x4 = [[5, "Jisoo"], [5, "Jennie"], [4, "Rose"], [4, "Lisa"]]
x4.sort()
print("x4 (เรียงตามความยาวชื่อ แล้วตามชื่อ):", x4)
```
