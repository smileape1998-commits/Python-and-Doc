## 5.8 Split & Join (การแยกและรวมสตริงกับลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การแปลงข้อมูลระหว่าง สตริง (String) และ ลิสต์ของสตริง (List of Strings) เป็นทักษะสำคัญในการจัดการข้อมูลข้อความ:

1. **`str.split(sep)`:** แปลงสตริงให้กลายเป็นลิสต์ โดยใช้ตัวคั่น `sep`
   - `input().split()` : หากไม่ระบุตัวคั่น จะตัดด้วยช่องว่างทุกชนิด (Space, Tab, Newline) และจะข้ามช่องว่างซ้ำๆ ให้อัตโนมัติ
   - `s.split("/")` หรือ `s.split(",")` : ตัดด้วยอักขระตัวคั่นตามที่ระบุ
   - **Unpacking:** สามารถกระจายผลลัพธ์ใส่หลายตัวแปรได้ เช่น `d, m, y = s.split("/")`

2. **`sep.join(list)`:** รวมสมาชิกในลิสต์ของสตริงเข้าด้วยกันให้เป็นสตริงเดียว โดยคั่นด้วย `sep`
   - เช่น `", ".join(list)`, `" ".join(list)`, หรือ `"x".join(list)`

### ข้อควรระวัง (Common Pitfalls)
- **ชนิดข้อมูลในลิสต์ขณะใช้ `join()`:** เมธอด `join()` จะต้องใช้กับลิสต์ที่สมาชิกทุกตัวเป็น **สตริง (String)** เท่านั้น หากสมาชิกเป็นตัวเลข ต้องแปลงเป็นสตริงก่อนด้วย `str(k)` เช่น `results.append(str(k))` หรือ `" ".join([str(e) for e in list])`

```python
# 5.8.1 ตัวอย่างภาพรวมการใช้ split() และ join()

# 1. การใช้ split() รูปแบบต่างๆ
# อินพุต: "  A  B  C    DEF"
x1 = "  A  B  C    DEF".split()  # ตัดช่องว่างติดกันหลายตัวให้อัตโนมัติ
print("split() ช่องว่าง:", x1)

# อินพุต: "A, B,C,D,E"
x2 = "A, B,C,D,E".split(",")     # ตัดด้วยเครื่องหมายจุลภาค
print("split(',') :", x2)

# การใช้ Unpacking จาก split()
s_date = "10/3/2562"
d, m, y = s_date.split("/")
print(f"วัน: {d}, เดือน: {m}, ปี: {y}")

# 2. การใช้ join() รวมลิสต์ตัวเลขที่หารด้วย 5 หรือ 7 ลงตัวในช่วง 1-30
results = []
for k in range(1, 31):
    if k % 5 == 0 or k % 7 == 0:
        results.append(str(k))  # แปลงเป็นสตริงก่อน append

print("ผลลัพธ์จาก join():", ", ".join(results))
```

#### ตัวอย่าง 5.8.1: การประยุกต์ใช้งาน Split & Join ในแบบฝึกหัด

**ข้อ 1 (จัดรูปแบบชื่อธนาคาร):** อ่านชื่อย่อธนาคารคั่นด้วยช่องว่างหลายช่อง เช่น `SCB  BBL  KBANK` แล้วจัดรูปแบบให้คั่นด้วย comma และช่องว่างหนึ่งช่อง (`", ".join(t)`)

**ข้อ 2 (เพิ่มค่าตัวเลขในสตริง):** อ่านรายการตัวเลขคั่นด้วยช่องว่าง เช่น `10 20 30` เพิ่มค่าตัวละ 1 แล้วรวมผลลัพธ์เป็นสตริงคั่นด้วยช่องว่าง (`" ".join(y)` ได้ `"11 21 31"`)

**ข้อ 3 (แยกตัวประกอบบรรทัดเดียวกันคั่นด้วย x):** รับจำนวนเต็ม `n` แล้วหาตัวประกอบ นำมาเก็บลงลิสต์สตริง แล้วรวมแสดงในบรรทัดเดียวกันด้วย `"x".join(factors)` เช่น `8568` แสดง `2x2x2x3x3x7x17`


```python
# ข้อ 1: อ่านชื่อย่อธนาคารแล้วแสดงคั่นด้วย ", "
t = input().split()
print("ข้อ 1:", ", ".join(t))

# ข้อ 2: เพิ่มค่าจำนวนเต็มตัวละ 1 แล้วรวมเป็นสตริงด้วย " "
x_nums = input().split()
y_nums = []
for e in x_nums:
    y_nums.append(str(int(e) + 1))
output_str = " ".join(y_nums)
print("ข้อ 2 (output):", output_str)

# ข้อ 3: แสดงตัวประกอบของ n ในบรรทัดเดียวกันคั่นด้วย 'x'
# อินพุตตัวอย่าง: 8568
n = int(input())
k = 2
factors = []
while k <= n:
    if n % k == 0:
        factors.append(str(k))
        n //= k
    else:
        k += 1
print("ข้อ 3 (ตัวประกอบ):", "x".join(factors))
```
