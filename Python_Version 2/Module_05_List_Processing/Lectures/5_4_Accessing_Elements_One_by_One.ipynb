## 5.4 Accessing Elements One-by-One (การเข้าถึงสมาชิกทีละตัว)

### อธิบายแนวคิด (Concept Explanation)
การวนลูปเพื่อดึงสมาชิกทีละตัวมาประมวลผล สามารถทำได้ 2 รูปแบบหลัก:

1. **Member-based Iteration (`for e in d:`):** หยิบข้อมูลทีละตัวในลิสต์มาใช้งานโดยตรง อ่านง่ายที่สุด เหมาะสำหรับการอ่านค่าหรือสร้างลิสต์ใหม่
2. **Index-based Iteration (`for i in range(len(d)):`):** วนลูปตามดัชนีตำแหน่ง เหมาะเมื่อต้องการใช้ทั้งดัชนีตำแหน่ง `i` และข้อมูล `d[i]` หรือเมื่อต้องการแก้ไขค่าสมาชิกภายในลิสต์เดิม

### ข้อควรระวัง (Common Pitfalls)
- **อย่าใช้วิธีวนลูปแบบ `for e in d: i = d.index(e)`:** การค้นหาตำแหน่ง `index(e)` ใหม่ทุกรอบจะทำงานช้ามาก ให้ใช้แบบดัชนี `for i in range(len(d)):` แทน
- **การแก้ไขสมาชิกผ่านตัวแปรลูป `e` จะไม่มีผลกับลิสต์เดิม:** เช่น `for e in d: if e < 0: e *= -1` ค่าใน `e` จะเปลี่ยนแต่ค่าในลิสต์ `d` จะคงเดิม หากต้องการแก้ไขค่าในลิสต์เดิม ต้องอ้างอิงตำแหน่งผ่านดัชนี `d[i] *= -1`

```python
# 5.4.1 เปรียบเทียบการเปลี่ยนจำนวนติดลบให้เป็นบวกในลิสต์ d
d = [10, -5, 20, -15, 30]

# แบบที่ผิด: เปลี่ยนค่าในตัวแปรลูป e (ลิสต์ d ไม่เปลี่ยน)
for e in d:
    if e < 0:
        e *= -1
print("แบบที่ผิด (d ไม่เปลี่ยน):", d)

# แบบที่ถูกต้อง: เปลี่ยนค่าผ่านดัชนี d[i]
for i in range(len(d)):
    if d[i] < 0:
        d[i] *= -1
print("แบบที่ถูก (d เปลี่ยนค่าเป็นบวกทั้งหมด):", d)
```

#### ตัวอย่าง 5.4.1: การประมวลผลสร้างลิสต์ใหม่จากสมาชิกเดิม

**ข้อ 1:** สร้างลิสต์ `y` เก็บกำลังสองของแต่ละตัวใน `x`

**ข้อ 2:** สร้างลิสต์ `y` เก็บเฉพาะจำนวนบวกใน `x`

**ข้อ 3:** สร้างลิสต์ `y` เก็บรากที่สอง (`e**0.5`) เฉพาะจำนวนไม่ติดลบใน `x`


```python
# ตัวอย่าง 5.4.1 ข้อที่ 1: ยกกำลังสองของทุกตัวใน x
x = [4, -9, 16, 25, -36]
y1 = []
for e in x:
    y1.append(e**2)
print("y1 (กำลังสอง) =", y1)

# ตัวอย่าง 5.4.1 ข้อที่ 2: เฉพาะจำนวนบวกใน x
y2 = []
for e in x:
    if e > 0:
        y2.append(e)
print("y2 (เฉพาะจำนวนบวก) =", y2)

# ตัวอย่าง 5.4.1 ข้อที่ 3: รากที่สองเฉพาะจำนวนไม่ติดลบใน x
y3 = []
for e in x:
    if e >= 0:
        y3.append(e**0.5)
print("y3 (รากที่สองเฉพาะตัวไม่ติดลบ) =", y3)
```

#### ตัวอย่าง 5.4.2: การคัดกรองข้อมูลหุ้นด้วยลิสต์ขนานผ่านดัชนี

**คำอธิบาย:** รับชื่อหุ้นและราคาในบรรทัดเดียวกัน เช่น `KTB 18.9 BBL 176.5 SCB 132.0 KBANK 172.0 TMB 1.8` แยกเป็นลิสต์ชื่อหุ้น `s` และลิสต์ราคาหุ้น `p` (แปลงเป็น float) จากนั้นรับราคาสูงสุด `p0` แล้ววนลูปดัชนี `for i in range(len(p)):` เพื่อแสดงชื่อหุ้นที่มีราคาต่ำกว่า `p0`


```python
# ตัวอย่าง 5.4.2: คัดกรองชื่อหุ้นที่มีราคาต่ำกว่า p0
# อินพุตบรรทัดแรก: KTB 18.9 BBL 176.5 SCB 132.0 KBANK 172.0 TMB 1.8
# อินพุตบรรทัดสอง: 150.0
x = input().split()
s = x[::2]  # ลิสต์ชื่อหุ้น
p = []
for e in x[1::2]:
    p.append(float(e))  # ลิสต์ราคาหุ้น

p0 = float(input())  # ราคาสูงสุดที่กำหนด
print("หุ้นที่มีราคาต่ำกว่า", p0, ":")
for i in range(len(p)):
    if p[i] < p0:
        print(s[i])
```
