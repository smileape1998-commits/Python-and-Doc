## 5.6 List Searching (การค้นหาข้อมูลในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
การจัดโครงสร้างข้อมูลเพื่อรองรับการค้นหาข้ามชุดข้อมูล มี 3 รูปแบบหลัก:

1. **Single Flat List (ลิสต์เดี่ยวผสม):** เก็บข้อมูลคู่เรียงต่อกันในลิสต์เดียว เช่น `['KTB', 18.9, 'BBL', 176.5, ...]`
2. **Parallel Lists (ลิสต์ขนาน):** แยกข้อมูลออกเป็น 2 ลิสต์ที่มีดัชนีตำแหน่งสอดคล้องกัน เช่น `eng = ['BKK', 'CMI']` และ `th = ['กทม', 'ชม']`
3. **Nested List (ลิสต์ซ้อนลิสต์ / 2D List):** เก็บข้อมูลแต่ละคู่เป็นลิสต์ย่อยภายในลิสต์หลัก เช่น `[['BKK', 'กทม'], ['CMI', 'ชม']]`

### ข้อควรระวัง (Common Pitfalls)
- ในการค้นหาด้วย 2D List หรือลิสต์ขนาน ต้องใช้ `break` เมื่อพบคู่ข้อมูลที่ต้องการแล้ว เพื่อหยุดการวนลูปค้นหาโดยไม่จำเป็น

```python
# 5.6.1 เปรียบเทียบรูปแบบการเก็บและค้นหาข้อมูลราคาหุ้น 3 รูปแบบ
x = input().split()  # เช่น: KTB 18.9 BBL 176.5 SCB 132.0 KBANK 172.0
q = input()          # ชื่อหุ้นที่ต้องการค้นหา เช่น: SCB

# รูปแบบที่ 1: Single Flat List
s1 = []
for i in range(len(x)):
    if i % 2 == 0:
        s1.append(x[i])
    else:
        s1.append(float(x[i]))

if q in s1:
    print("Flat List พบราคา:", s1[s1.index(q) + 1])
else:
    print("Flat List: Not found")

# รูปแบบที่ 2: Parallel Lists (ลิสต์ขนาน)
names = x[::2]
prices = []
for e in x[1::2]:
    prices.append(float(e))

if q in names:
    print("Parallel Lists พบราคา:", prices[names.index(q)])
else:
    print("Parallel Lists: Not found")

# รูปแบบที่ 3: Nested List (2D List)
nested_data = []
for i in range(0, len(x), 2):
    nested_data.append([x[i], float(x[i + 1])])

found = False
for name, price in nested_data:
    if q == name:
        print("Nested List พบราคา:", price)
        found = True
        break
if not found:
    print("Nested List: Not found")
```

#### ตัวอย่าง 5.6.1: การค้นหาข้อมูลสองทิศทาง (Bi-directional Search)

**โจทย์:** รับข้อมูลชื่อย่อจังหวัด ไทย-อังกฤษ สลับกัน เช่น `BKK กทม CMI ชม KBI กบ` จากนั้นรับคำค้นหา `q` หากเป็นชื่อย่อไทย ให้แสดงชื่อย่ออังกฤษ หากเป็นชื่อย่ออังกฤษ ให้แสดงชื่อย่อไทย หากไม่ใช่ทั้งสอง ให้แสดง `Not found`

- **ข้อ 1 (Parallel Lists Search):** แยกข้อมูลด้วย Slicing เป็น 2 ลิสต์ `eng = x[::2]` และ `th = x[1::2]`
- **ข้อ 2 (Nested List Search):** สร้าง 2D List `abbr = [[eng, th], ...]` แล้ววนลูป unpacking `for e, t in abbr:` เพื่อค้นหาแบบสองทิศทาง


```python
# ตัวอย่าง 5.6.1 ข้อที่ 1: ค้นหาสองทิศทางด้วย Parallel Lists
x = input().split()
eng = x[::2]
th = x[1::2]
q = input()

if q in eng:
    i = eng.index(q)
    print(th[i])
elif q in th:
    i = th.index(q)
    print(eng[i])
else:
    print("Not found")

# ตัวอย่าง 5.6.1 ข้อที่ 2: ค้นหาสองทิศทางด้วย Nested List (2D List)
raw_input = input().split()
abbr = []
for i in range(0, len(raw_input), 2):
    abbr.append([raw_input[i], raw_input[i + 1]])

q_query = input()
found = False
for e, t in abbr:
    if q_query == t:
        print(e)
        found = True
        break
    elif q_query == e:
        print(t)
        found = True
        break

if not found:
    print("Not found")
```
