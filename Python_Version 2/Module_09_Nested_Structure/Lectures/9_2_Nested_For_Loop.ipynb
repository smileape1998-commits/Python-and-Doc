## 9.2 Nested for Loop (คำสั่ง for แบบซ้อน)

### อธิบายแนวคิด (Concept Explanation)
Nested for Loop คือ การใช้คำสั่ง `for` ซ้อนอยู่ภายในคำสั่ง `for` อีกชุดหนึ่ง โดยทั่วไปลูปนอกจะควบคุมการทำงานในแนวดิ่ง (เช่น แถว หรือ สมาชิกตัวที่ 1) และลูปในจะควบคุมการทำงานในแนวนอน (เช่น หลัก หรือ สมาชิกตัวที่ 2)

![Nested Loop Pattern](https://www.cp.eng.chula.ac.th/~somchai/python101/images/nested_01.png)
![Nested Match Index](https://www.cp.eng.chula.ac.th/~somchai/python101/images/nested_04_match.png)

**รูปแบบการกำหนดช่วงดัชนี (Index Patterns):**
1. **ทุกลำดับคู่พิกัด (All pairs):** `for i in range(3): for j in range(4):` ประมวลผลทุกพิกัด $(i, j)$
2. **คู่ข้อมูลตั้งแต่ i ขึ้นไป:** `for i in range(3): for j in range(i, 4):` ประมวลผลรูปสามเหลี่ยมด้านบนรวมทวิคูณ
3. **คู่ข้อมูลที่ไม่ซ้ำกัน (Unique pairs):** `for i in range(n): for j in range(i+1, n):` แจกแจงดัชนีของทุกคู่ข้อมูลที่ไม่ซ้ำกัน

### ข้อควรระวัง (Common Pitfalls)
1. **การสับสนลำดับดัชนี:** การอ้างอิง `range(i+1, n)` เพื่อหาคู่ที่ไม่ซ้ำ ต้องระวังไม่ให้ดัชนีลูปนอก `i` เกินขอบเขต
2. **จำนวนรอบการประมวลผล:** หากลูปนอกทำงาน $N$ รอบ และลูปในแบบจับคู่ทำงานประมาณ $N^2/2$ รอบ ควรระวังประสิทธิภาพเมื่อ $N$ มีขนาดใหญ่มาก

### แบบฝึกหัด 9-2 ข้อที่ 2: ตารางสูตรคูณแม่ 2 ถึง N ด้วยคำสั่ง for

**คำอธิบาย:** รับจำนวนเต็ม `N` พิมพ์ตารางสูตรคูณแม่ 2 ถึงแม่ `N` โดยคูณด้วย 1 ถึง 5 จัดรูปแบบให้ชิดขวาความกว้าง 4 ช่องด้วยคำสั่ง `for` เท่านั้น (ห้ามใช้ `while`)

```python
# แบบฝึกหัด 9-2 ข้อที่ 2: ตารางสูตรคูณแม่ 2 ถึง N ด้วยคำสั่ง for
def left_pad(n, k):
    return (" " * k + str(n))[-k:]

N = int(input())
for k in range(2, N + 1):
    line = ""
    for m in range(1, 6):
        line += left_pad(k * m, 4)
    print(line)
```
