## 9.1 Nested while Loop (คำสั่ง while แบบซ้อน)

### อธิบายแนวคิด (Concept Explanation)
Nested while Loop คือ โครงสร้างการวนซ้ำที่มีคำสั่ง `while` ซ้อนอยู่ภายในคำสั่ง `while` อีกชุดหนึ่ง โดยเรียกลูปที่อยู่ภายนอกว่า **ลูปนอก (Outer Loop)** และเรียกลูปที่อยู่ภายในว่า **ลูปใน (Inner Loop)**

![Nested while Loop Diagram](https://www.cp.eng.chula.ac.th/~somchai/python101/images/nested_01.png)
![Nested while Flowchart](https://www.cp.eng.chula.ac.th/~somchai/python101/images/nested_02_FC.png)

หลักการทำงานคือ ในทุกๆ 1 รอบการทำงานของลูปนอก ลูปในจะทำการประมวลผลซ้ำตั้งแต่เริ่มต้นจนจบกระบวนการ (จนกระทั่งเงื่อนไขของลูปในเป็นเท็จ)

### ข้อควรระวัง (Common Pitfalls)
1. **การลืมกำหนดค่าเริ่มต้นตัวแปรลูปใน:** ตัวแปรควบคุมลูปในต้องถูกตั้งค่าเริ่มต้นใหม่ทุกครั้งก่อนเข้าสู่ลูปใน
2. **การลืมปรับปรุงค่าตัวแปรลูป:** หากลืมเพิ่มหรือลดค่าตัวแปรควบคุมลูปในหรือลูปนอก จะทำให้เกิดการวนลูปไม่รู้จบ (Infinite Loop)

### แบบฝึกหัด 9-1 ข้อที่ 1: การพิมพ์ตารางสูตรคูณชิดขวาด้วย while loop

**คำอธิบาย:** รับจำนวนเต็ม `N` พิมพ์ตารางสูตรคูณแม่ 2 ถึงแม่ `N` โดยคูณด้วย 1 ถึง 5 และจัดรูปแบบให้ชิดขวาความกว้าง 4 ช่องด้วยฟังก์ชัน `left_pad`

```python
# แบบฝึกหัด 9-1 ข้อที่ 1: ตารางสูตรคูณแม่ 2 ถึง N
def left_pad(n, k):
    return (" " * k + str(n))[-k:]

N = int(input())
k = 2
while k <= N:
    m = 1
    line = ""
    while m <= 5:
        line += left_pad(k * m, 4)
        m += 1
    print(line)
    k += 1
```
