## 9.5 Nested List as Matrix (การใช้ลิสต์ซ้อนเป็นเมทริกซ์)

### อธิบายแนวคิด (Concept Explanation)
ในทางคอมพิวเตอร์ เราสามารถประยุกต์ใช้ Nested List เพื่อแทนโครงสร้างทางคณิตศาสตร์แบบ **เมทริกซ์ (Matrix)** ขนาด $M \times N$ โดยที่:
- $M$ คือ จำนวนแถว (Rows)
- $N$ คือ จำนวนหลัก (Columns)

การเข้าถึงสมาชิกตำแหน่งแถวที่ `r` หลักที่ `c` ทำได้โดยใช้ดัชนี `matrix[r][c]` และสามารถคำนวณผลรวมแถว ผลรวมหลัก หรือสร้างเมทริกซ์ชนิดพิเศษได้ด้วย Nested Loop

### ข้อควรระวัง (Common Pitfalls)
1. **ความสับสนระหว่างดัชนีแถวและหลัก:** การหาผลรวมหลัก (Column Sum) ต้องสลับเอาลูปหลัก `c` ไว้ภายนอก และเอาลูปแถว `r` ไว้ภายใน
2. **มิติของเมทริกซ์ไม่สอดคล้องกัน:** การอ้างอิง `M[0]` เพื่อหาจำนวนหลัก สมมติว่าทุกแถวมีความยาวเท่ากัน

### ตัวอย่าง 9.5.1: การบวกเมทริกซ์สองมิติ (Matrix Addition)

```python
# ตัวอย่าง 9.5.1: การบวกเมทริกซ์ A และ B ขนาด 2x3
A = [
    [1, 2, 3],
    [4, 5, 6]
]
B = [
    [7, 8, 9],
    [1, 2, 3]
]

rows = len(A)
cols = len(A[0])
C = []
for r in range(rows):
    row_list = []
    for c in range(cols):
        row_list.append(A[r][c] + B[r][c])
    C.append(row_list)

print("ผลบวกเมทริกซ์ C:", C)
```

### ตัวอย่าง 9.5.2: การทรานสโพสเมทริกซ์ (Matrix Transpose)

```python
# ตัวอย่าง 9.5.2: สลับแถวเป็นหลัก (Transpose) เมทริกซ์ A ขนาด 2x3 เป็น 3x2
A = [
    [1, 2, 3],
    [4, 5, 6]
]

rows = len(A)
cols = len(A[0])
A_transpose = []
for c in range(cols):
    new_row = []
    for r in range(rows):
        new_row.append(A[r][c])
    A_transpose.append(new_row)

print("เมทริกซ์สลับเปลี่ยน (Transpose):", A_transpose)
```

### แบบฝึกหัด 9-5 ข้อที่ 1: ฟังก์ชันหาผลรวมของทุกสมาชิกในเมทริกซ์ (sum_all)

**คำอธิบาย:** รับเมทริกซ์ `M` คืนผลรวมของตัวเลขทุกจำนวนในเมทริกซ์

```python
# แบบฝึกหัด 9-5 ข้อที่ 1: ฟังก์ชัน sum_all
def sum_all(M):
    s = 0
    for row in M:
        for e in row:
            s += e
    return s

M_test = [[1, 2, 3], [4, 5, 6]]
print("sum_all(M_test) =", sum_all(M_test))
```

### แบบฝึกหัด 9-5 ข้อที่ 2: ฟังก์ชันหาผลรวมในแต่ละแถวแนวแนวนอน (sum_row)

**คำอธิบาย:** รับเมทริกซ์ `M` คืนลิสต์ที่ช่องที่ `k` เก็บผลรวมของตัวเลขในแถวที่ `k`

```python
# แบบฝึกหัด 9-5 ข้อที่ 2: ฟังก์ชัน sum_row
def sum_row(M):
    s = []
    for r in range(len(M)):
        su = 0
        for c in range(len(M[r])):
            su += M[r][c]
        s.append(su)
    return s

print("sum_row(M_test) =", sum_row(M_test))
```

### แบบฝึกหัด 9-5 ข้อที่ 3: ฟังก์ชันหาผลรวมในแต่ละหลักแนวตั้ง (sum_col)

**คำอธิบาย:** รับเมทริกซ์ `M` คืนลิสต์ที่ช่องที่ `k` เก็บผลรวมของตัวเลขในหลักแนวตั้งที่ `k`

```python
# แบบฝึกหัด 9-5 ข้อที่ 3: ฟังก์ชัน sum_col
def sum_col(M):
    s = []
    for c in range(len(M[0])):
        su = 0
        for r in range(len(M)):
            su += M[r][c]
        s.append(su)
    return s

print("sum_col(M_test) =", sum_col(M_test))
```

### แบบฝึกหัด 9-5 ข้อที่ 4: ฟังก์ชันสร้างเมทริกซ์เอกลักษณ์ (identity)

**คำอธิบาย:** รับจำนวนเต็มบวก `n` คืนลิสต์ซ้อนลิสต์แทนเมทริกซ์เอกลักษณ์ขนาด `n x n` (ช่องแนวเส้นทแยงมุมมีค่าเป็น 1 ช่องอื่นเป็น 0)

```python
# แบบฝึกหัด 9-5 ข้อที่ 4: ฟังก์ชัน identity
def identity(n):
    I = []
    for r in range(n):
        I.append([0] * n)
        I[r][r] = 1
    return I

print("identity(3) =", identity(3))
```
