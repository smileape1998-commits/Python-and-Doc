## 9.4 Nested Lists (ลิสต์ซ้อนในลิสต์)

### อธิบายแนวคิด (Concept Explanation)
Nested List คือ ลิสต์ที่มีสมาชิกภายในเป็นลิสต์อีกชุดหนึ่ง (List of Lists) เป็นโครงสร้างข้อมูลที่ยืดหยุ่นสูง ใช้ในกรณีต่างๆ ดังนี้:

1. **เก็บข้อมูลระบุรายละเอียดที่มีข้อมูลย่อยเป็นลิสต์:** เช่น ข้อมูลนักแสดงที่มีลิสต์ชื่อผลงานซ้อนอยู่ภายใน `["Ranee", 1989, ["Plerng Boon", "Bubphe Sanniwat"]]`
2. **เก็บกลุ่มข้อมูลหลายๆ ชุด:** เช่น ข้อมูลรหัสนักศึกษาและเกรดเฉลี่ย `[["6131001021", 3.8], ["6130020221", 3.7]]`
3. **เก็บข้อมูลชั่วคราวเพื่อประมวลผล:** เช่น สร้างคู่อันดับ `[ความยาวคำ, คำ]` เพื่อนำไปเรียงลำดับสตริงตามความยาวคำ
4. **เก็บตารางหรือเมทริกซ์:** เช่น `[[1, 2, 3], [4, 5, 6]]`

### ข้อควรระวัง (Common Pitfalls)
**การสร้าง Nested List ด้วยตัวดำเนินการคูณ `x = [[0]] * 5` (Reference Copy Bug):**
การใช้ `[[0]] * 5` จะสร้างลิสต์ซ้อนที่มี 5 ช่อง แต่ทุกช่องจะ **อ้างอิงถึงหน่วยความจำตำแหน่งเดียวกัน** เมื่อสั่งแก้ไข `x[0][0] = 9` จะส่งผลให้ทุกช่องเปลี่ยนเป็น `[9]` ทั้งหมด

**วิธีการสร้างที่ถูกต้อง:** ต้องสร้างลิสต์ย่อยใหม่แยกคนละใบด้วยการวนลูป `append()` หรือ List Comprehension

### ตัวอย่าง 9.4.1: การสร้าง Nested List เก็บข้อมูลที่มีข้อมูลย่อยต่างประเภท

```python
# ตัวอย่าง 9.4.1: สร้างลิสต์เก็บข้อมูลนักแสดงและผลงานซีรีส์
name = "Ranee"
byear = 1989
series = ["Plerng Boon", "Bubphe Sanniwat", "Krong Kam"]
actress = [name, byear, series]

print("ข้อมูลนักแสดง:", actress)
print("ซีรีส์เรื่องแรก:", actress[2][0])
```

### ตัวอย่าง 9.4.2: การใช้ Nested List ช่วยเรียงลำดับสตริงตามความยาวคำ (Sort by Length)

```python
# ตัวอย่าง 9.4.2: เรียงลำดับคำในลิสต์ตามความยาวของคำ
def sort_by_length(words):
    x = []
    for w in words:
        x.append([len(w), w])
    x.sort()
    for k in range(len(x)):
        words[k] = x[k][1]

word_list = ["your", "kiss", "is", "on", "my", "list"]
sort_by_length(word_list)
print("หลังเรียงตามความยาว:", word_list)
```

### แบบฝึกหัด 9-4 ข้อที่ 1: ฟังก์ชันแยกตัวเลขลงถังตามหลักหน่วย (to_buckets)

**คำอธิบาย:** รับ `d` เป็นลิสต์ของจำนวนเต็ม คืนลิสต์ 10 ช่อง โดยช่องที่ `k` เก็บจำนวนเต็มที่มีหลักหน่วยเป็น `k` ตามลำดับเดิม

```python
# แบบฝึกหัด 9-4 ข้อที่ 1: ฟังก์ชัน to_buckets
def to_buckets(d):
    b = []
    for i in range(10):
        b.append([])
    for e in d:
        b[e % 10].append(e)
    return b

print("to_buckets([33, 21, 143, 33, 11]) =", to_buckets([33, 21, 143, 33, 11]))
```

### แบบฝึกหัด 9-4 ข้อที่ 2: ฟังก์ชันอ่านพิกัดจุดจากอินพุต (read_points)

**คำอธิบาย:** อ่านอินพุตบรรทัดเดียวในรูปแบบ `x0:y0, x1:y1, ...` แล้วเปลี่ยนเป็นลิสต์ของพิกัด `[[x0, y0], [x1, y1], ...]`

```python
# แบบฝึกหัด 9-4 ข้อที่ 2: ฟังก์ชัน read_points
def read_points():
    points = []
    x = input().split(", ")
    for s in x:
        x_val, y_val = s.split(":")
        p = [float(x_val), float(y_val)]
        points.append(p)
    return points
```

### แบบฝึกหัด 9-4 ข้อที่ 3: ฟังก์ชันหาคู่จุดที่อยู่ใกล้กันที่สุดทั้งหมด (all_closest_pairs)

**คำอธิบาย:** รับลิสต์พิกัดจุด คืนลิสต์แสดงดัชนีของคู่จุดที่มีระยะห่างน้อยที่สุดทั้งหมด โดยเรียงลำดับดัชนีจากน้อยไปมาก

```python
# แบบฝึกหัด 9-4 ข้อที่ 3: ฟังก์ชัน all_closest_pairs
def distance(p1, p2):
    dx = p1[0] - p2[0]
    dy = p1[1] - p2[1]
    return (dx**2 + dy**2)**0.5

def all_closest_pairs(points):
    min_distance = distance(points[0], points[1])
    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            d = distance(points[i], points[j])
            if d < min_distance:
                min_distance = d

    closest_pairs = []
    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            d = distance(points[i], points[j])
            if d == min_distance:
                closest_pairs.append([i, j])
    return closest_pairs

pts = [[30, 30], [30, 31], [31, 30], [30, 32]]
print("all_closest_pairs =", all_closest_pairs(pts))
```

### แบบฝึกหัด 9-4 ข้อที่ 4: แปลงลิสต์คู่การชื่นชอบเป็น Dictionary (to_dict)

**คำอธิบาย:** รับ `like_list` ในรูปแบบ `[[u1, u2], ...]` แล้วแปลงเป็น Dictionary ในรูปแบบ `{user: [ลิสต์คนที่ user ชอบ]}`

```python
# แบบฝึกหัด 9-4 ข้อที่ 4: ฟังก์ชัน to_dict
def to_dict(like_list):
    like_dict = {}
    for u1, u2 in like_list:
        if u1 not in like_dict:
            like_dict[u1] = [u2]
        else:
            like_dict[u1].append(u2)
    return like_dict

likes = [['A', 'B'], ['A', 'C'], ['C', 'B']]
print("to_dict =", to_dict(likes))
```
