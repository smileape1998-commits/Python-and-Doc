## 9.5 Nested List as Matrix (การใช้ลิสต์ซ้อนเป็นเมทริกซ์)

### อธิบายแนวคิด (Concept Explanation)
ประยุกต์ใช้ Nested List แทนเมทริกซ์ขนาด M x N โดยเข้าถึงด้วยดัชนี `matrix[r][c]`


```python
# ตัวอย่าง 9.5.1: การบวกเมทริกซ์ 2 มิติ
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
C = [[A[r][c] + B[r][c] for c in range(len(A[0]))] for r in range(len(A))]

# TODO: เขียนโค้ดประมวลผล (Live Code) ที่นี่

print("A + B =", C)
```

## 9.6 List Comprehension (การสร้างลิสต์แบบกระชับ)

### อธิบายแนวคิด (Concept Explanation)
ไวยากรณ์สร้างลิสต์บรรทัดเดียว `[expression for item in iterable if condition]`


```python
# ตัวอย่าง 9.6.1: กรองเฉพาะเลขคู่และยกกำลังสอง
result = [x**2 for x in range(1, 11) if x % 2 == 0]

# TODO: เขียนโค้ดประมวลผล (Live Code) ที่นี่

print(result)
```
