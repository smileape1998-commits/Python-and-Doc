## 7.2 Escape Char. & String Method (อักขระหลีกและเมธอดของสตริง)

### อักขระพิเศษ (Escape Characters)
อักขระหลีกใช้เครื่องหมาย Backslash (`\\`) เพื่อสร้างรหัสอักขระพิเศษในสตริง:
- `\\"` คือเครื่องหมายอัญประกาศคู่ `"` 
- `\\'` คือเครื่องหมายอัญประกาศเดี่ยว `'` 
- `\\\\` คือเครื่องหมาย Backslash `\\` 
- `\\n` คือรหัสสั่งให้ขึ้นบรรทัดใหม่ (Newline)
- `\\t` คือรหัสสั่งเว้นระยะแท็บ (Tab)

### ตัวอย่าง 7.2.1: การใช้งาน Escape Characters

```python
s = "Hello"
print(s)           # Hello

s = '"Hello"'
print(s)           # "Hello"

s = "'Hello'"
print(s)           # 'Hello'

s = "\"'\\Hello\\"
print(s)           # "'\\Hello\\"
```

### บริการของสตริง : String Methods
สมมติให้ `s = "  Hello World "` (ความยาวดัชนี 0 ถึง 13)

- `len(s)` ได้ `14`, `len("")` ได้ `0`
- `s.lower()` ได้ `"  hello world "` (แปลงอักขระเป็นพิมพ์เล็กทั้งหมด)
- `s.upper()` ได้ `"  HELLO WORLD "` (แปลงอักขระเป็นพิมพ์ใหญ่ทั้งหมด)
- `s.strip()` ได้ `"Hello World"` (ตัดช่องว่างทางซ้ายและขวาออกทั้งหมด)
- `s.find("o")` ได้ `6` (หาตำแหน่งแรกที่เจอคำที่ค้นหา)
- `s.find("ex")` ได้ `-1` (คืนค่า -1 เมื่อไม่พบคำที่ค้นหา)
- `s.find("o", 7)` ได้ `9` (เริ่มหาตั้งแต่ index ที่ 7 เป็นต้นไป)

### ตัวอย่าง 7.2.2: การใช้งาน String Methods

```python
s = "  Hello World "
print("len(s):", len(s))
print('len(""):', len(""))
print("s.lower():", f"'{s.lower()}'")
print("s.upper():", f"'{s.upper()}'")
print("s.strip():", f"'{s.strip()}'")
print('s.find("o"):', s.find("o"))
print('s.find("ex"):', s.find("ex"))
print('s.find("o", 7):', s.find("o", 7))
```
