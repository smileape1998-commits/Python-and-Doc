## 7.3 More String Processing Examples (ตัวอย่างการประมวลผลสตริงเพิ่มเติม)

### Method Chaining vs Function Composition

การเรียกใช้งานฟังก์ชันและเมธอดซ้อนกันสามารถเขียนได้ 2 รูปแบบ:

1. **Function Composition (การเรียกฟังก์ชันซ้อนฟังก์ชัน):**
   ```python
   x = math.radians(d)
   s = math.sin(x)
   y = abs(s)
   r = round(y, 2)
   
   # หรือเขียนซ้อนกันแบบรวบเดียว:
   r = round(abs(math.sin(math.radians(d))), 2)
   ```

2. **Method Chaining (การเรียกเมธอดต่อกันเป็นลูกโซ่):**
   ```python
   line1 = input()
   line2 = line1.strip()
   line3 = line2.upper()
   i = line3.find("OK")
   
   # หรือเขียนต่อกันเป็นลูกโซ่:
   i = input().strip().upper().find("OK")
   ```

````python
import math

# ตัวอย่าง 7.3.1: Function Composition
d = 45
r = round(abs(math.sin(math.radians(d))), 2)
print("r =", r)

# ตัวอย่าง 7.3.2: Method Chaining
text = "  hello world ok  "
i = text.strip().upper().find("OK")
print("index of OK =", i)
````

### ข้อควรระวัง (Caveats In String Processing)

- **ห้ามเปลี่ยนค่าภายในสตริง (String Immutability):**
  - `s[2] = "a"`  (`TypeError`)
  - `s[3:7] = "-^o^-"`  (`TypeError`)

- **String Methods ไม่เปลี่ยนค่าตัวสตริงเดิม:**
  - `s = "HellO"` 
  - `s.lower()` ได้ `'hello'` แต่ `s` ยังคงเก็บค่าเดิมคือ `'HellO'`

- **การเปลี่ยนค่าต้องนำผลลัพธ์ไปเก็บในตัวแปรใหม่หรือตัวแปรเดิม:**
  - `s = "HellO"` 
  - `s = s.lower()` แบบนี้ `s` จะเก็บค่าใหม่คือ `"hello"`

````python
# ตัวอย่าง 7.3.3: การทดสอบ String Immutability และการ Re-assign ตัวแปร
s = "HellO"
s.lower() # ได้ 'hello' แต่ไม่ได้เปลี่ยนค่าใน s
print("s หลัง s.lower():", s)

s = s.lower() # เอาผลลัพธ์มาเก็บคืนใน s
print("s หลัง s = s.lower():", s)
````

### แบบฝึกหัด 7-3 ข้อที่ 1

จงเขียนฟังก์ชัน `dedent(line)` ที่คืนสตริงใหม่ที่มีค่าภายในเหมือน `line` แต่ลดจำนวนช่องว่างเฉพาะที่อยู่ติดกันทางซ้ายสุดลงครึ่งหนึ่ง เช่น `dedent("    a =  0")` ได้ `"  a =  0"`

````python
def dedent(line):
    c = 0
    for e in line:
        if e != " ":
            break
        c += 1
    return line[c // 2:]

print(f"'{dedent('    a =  0')}'") # '  a =  0'
````

### แบบฝึกหัด 7-3 ข้อที่ 2

จงเขียนฟังก์ชัน `rot5_digits(s)` ที่คืนสตริงใหม่ที่มีค่าภายในเหมือน `s` แต่เปลี่ยนเฉพาะตัวเลขข้างในจากเลข 0,1,2,3,4,5,6,7,8 และ 9 เป็น 5,6,7,8,9,0,1,2,3 และ 4 ตามลำดับ เช่น `rot5_digits("My number is 02-218-6981")` ได้ `"My number is 57-763-1436"`

````python
def rot5_digits(s):
    rot5 = ""
    for e in s:
        if "0" <= e <= "9":
            rot5 += str((int(e) + 5) % 10)
        else:
            rot5 += e
    return rot5

print(rot5_digits("My number is 02-218-6981")) # My number is 57-763-1436
````

### แบบฝึกหัด 7-3 ข้อที่ 3

สตริง DNA เป็นสตริงที่มีตัวอักขระแค่ A, T, G และ C เท่านั้น จงเขียนฟังก์ชัน `count_bases(dna)` คืนลิสต์ที่เก็บจำนวนเบสที่พบใน `dna` โดยเก็บจำนวนของ A, T, G และ C ในช่องที่ 0, 1, 2 และ 3 ตามลำดับของลิสต์ที่คืน เช่น `count_bases("AAAATTTGGC")` ได้ `[4, 3, 2, 1]`

````python
def count_bases(dna):
    bases = "ATGC"
    freq = [0, 0, 0, 0]
    for e in dna:
        freq[bases.index(e)] += 1
    return freq

print(count_bases("AAAATTTGGC")) # [4, 3, 2, 1]
````
