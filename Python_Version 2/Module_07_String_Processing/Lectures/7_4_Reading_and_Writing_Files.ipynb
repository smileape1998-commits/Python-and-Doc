### การบันทึกข้อมูลลงแฟ้มข้อความ (Writing Files)

- `fout = open(filename, "w")` โหมด `"w"` บอกว่าเปิดแฟ้มเพื่อ**เขียน** (หากมีไฟล์เดิมจะถูกเขียนทับ)
- `fout.write("Text\n")` ใช้เขียนบรรทัดสตริงลงไฟล์ หากต้องการขึ้นบรรทัดใหม่**ต้องใส่ `\n` เอง**
- `fout.close()` บันทึกเสร็จแล้วต้องปิดแฟ้มเสมอ
