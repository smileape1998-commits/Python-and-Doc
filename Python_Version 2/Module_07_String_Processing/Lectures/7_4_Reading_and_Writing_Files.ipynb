## 7.4 Reading & Writing Files (การอ่านและการเขียนไฟล์)

### การอ่านข้อมูลจากไฟล์ด้วย `open()` และ `readline()` / `for line in file`

- การเปิดไฟล์อ่านใช้ `open(filename, "r")`
- อ่านทีละบรรทัดด้วย `file.readline()` หรือใช้ห่วงวน `for line in file:` เพื่อวนอ่านจนหมดไฟล์
- หลังใช้งานเสร็จต้องปิดไฟล์ด้วย `file.close()` เสมอ

### ตัวอย่าง 7.4.1: หาคะแนนเฉลี่ยนิสิต 3 คนแรกในแฟ้ม (`data.txt`)

```python
sum_points = 0; n = 3

infile = open("data.txt", "r")
for k in range(n):
    line = infile.readline() # อ่านบรรทัดถัดไป
    x = line.split()
    sum_points += float(x[1])
infile.close()

print("Average =", sum_points/n)
```

### ตัวอย่าง 7.4.2: อ่านแฟ้มคะแนนมาแสดงเรียงจากมากไปน้อย

```python
students = []

infile = open("data.txt", "r")
for line in infile: # อ่านรอบละบรรทัดจนหมดแฟ้ม
    sid, point = line.strip().split()
    point = float(point)
    students.append([point, sid])
infile.close()

students.sort() # เรียงจากน้อยไปมากตามคะแนน

for [point, sid] in students[::-1]:
    print(sid, point)
```

### การบันทึกข้อมูลลงแฟ้มข้อความ (Writing Files)

- `fout = open(filename, "w")` โหมด `"w"` บอกว่าเปิดแฟ้มเพื่อ**เขียน** (หากมีไฟล์เดิมจะถูกเขียนทับ)
- `fout.write("Text\n")` ใช้เขียนบรรทัดสตริงลงไฟล์ หากต้องการขึ้นบรรทัดใหม่**ต้องใส่ `\n` เอง**
- `fout.close()` บันทึกเสร็จแล้วต้องปิดแฟ้มเสมอ

### ตัวอย่าง 7.4.3: การเขียนข้อความลงแฟ้มเบื้องต้น

```python
fout = open("sample_output.txt", "w")
fout.write("First line")
fout.write("Text\n")
fout.close()
print("บันทึกเสร็จแล้วก็ปิดแฟ้ม")
```

### ตัวอย่าง 7.4.4: บันทึก Triangular Number ร้อยตัวลงแฟ้ม

```python
fout = open("tri_numbers.txt", "w")
for k in range(1, 101):
    n = k*(k+1)//2      # triangular number
    fout.write(str(n)+" ")
    if k%10 == 0:       # ครบ 10 ตัว ขึ้นบรรทัดใหม่
        fout.write("\n")
fout.close()
```

### แบบฝึกหัด 7-4 ข้อที่ 1

จงเขียนฟังก์ชัน `count_line(file_name)` เพื่อหาจำนวนบรรทัดของข้อมูลในแฟ้มชื่อ `file_name`

```python
def count_line(file_name):
    fn = open(file_name)
    c = 0
    for line in fn:
        c += 1
    fn.close()
    return c
```

### แบบฝึกหัด 7-4 ข้อที่ 2

จงเขียนฟังก์ชัน `print_merge(filename1, filename2)` อ่านข้อมูลของแต่ละบรรทัดจากทั้งสองแฟ้มมาสลับกันแสดงทางจอภาพ (เริ่มที่ `filename1`) ถ้าแฟ้มใดหมด ก็อ่านอีกแฟ้มมาแสดงให้หมด

```python
def print_merge(filename1, filename2):
    f1 = open(filename1)
    f2 = open(filename2)
    d1 = f1.readline()
    d2 = f2.readline()
    while len(d1) > 0 or len(d2) > 0:
        if len(d1) > 0:
            print(d1.strip())
            d1 = f1.readline()
        if len(d2) > 0:
            print(d2.strip())
            d2 = f2.readline()
    f1.close()
    f2.close()
```
