### การใช้อักขระและการเข้าถึงสตริงย่อย (Slicing Mechanics)
สมมติให้ `s = "ABCDEFG"` 

- `len(t)` ได้จำนวนตัวอักขระใน `t` โดย `len('')` ได้ `0`
- `s[0]` เหมือน `s[-len(s)]` ได้ `"A"` ส่วน `s[-1]` เหมือน `s[len(s)-1]` ได้ `"G"` 
- ดัชนี (index) ของสตริง `s` ต้องอยู่ในช่วง `0` ถึง `len(s)-1` จากซ้ายไปขวา และ `-1` ถึง `-len(s)` ถอยจากขวามาซ้าย ดังนั้นเราเขียน `s[k]` ได้โดยที่ `-len(s) <= k <= (len(s)-1)` เพราะฉะนั้น `"01234"[-6]` กับ `"01234"[5]` ผิด (`IndexError`)
- `s` เหมือน `s[:]` เหมือน `s[0:]` เหมือน `s[:len(s)]` เหมือน `s[::]` เหมือน `s[::1]`
- `s[::2]` หยิบตัวที่ index คู่ได้ `"ACEG"`, `s[1::2]` หยิบตัวที่ index คี่ได้ `"BDF"` 
- `s[::-1]` เหมือน `s[-1::-1]` เหมือน `s[-1:-(len(s)+1):-1]` ได้ `"GFEDCBA"` 

### กฎขอบเขตของการเฉือนสตริง `s[a:b]`
ถ้าเขียน `s[a:b]` เพื่อเลือกสตริงย่อยออกมา ค่า `a` กับ `b` เป็นอะไรก็ได้ ไม่เกิด error ข้อความจะถูกตัดให้อยู่ในขอบเขตจริงโดยอัตโนมัติ:
- `"01234"[2:50000]` ได้ `"234"`  |  `"01234"[4999:50000]` ได้ `""` 
- `"01234"[-500:-2]` ได้ `"012"`  |  `"01234"[-3:-500:-1]` ได้ `"210"` 
- `"01234"[-500:-300]` ได้ `""`

### ตัวอย่าง 7.1.2: การทดสอบขอบเขตของการ Slice สตริง s[a:b]

````python
s = "01234"
print('"01234"[2:50000] ->', s[2:50000])
print('"01234"[4999:50000] ->', f"'{s[4999:50000]}'")
print('"01234"[-500:-2] ->', s[-500:-2])
print('"01234"[-3:-500:-1] ->', s[-3:-500:-1])
print('"01234"[-500:-300] ->', f"'{s[-500:-300]}'")
````

## 7.2 Escape Char. & String Method (อักขระหลีกและเมธอดของสตริง)

### อักขระพิเศษ (Escape Characters)
อักขระหลีกใช้เครื่องหมาย Backslash (`\\`) เพื่อสร้างรหัสอักขระพิเศษในสตริง:
- `\\"` คือเครื่องหมายอัญประกาศคู่ `"` 
- `\\'` คือเครื่องหมายอัญประกาศเดี่ยว `'` 
- `\\\\` คือเครื่องหมาย Backslash `\\` 
- `\\n` คือรหัสสั่งให้ขึ้นบรรทัดใหม่ (Newline)
- `\\t` คือรหัสสั่งเว้นระยะแท็บ (Tab)

### ตัวอย่าง 7.2.1: การใช้งาน Escape Characters

````python
s = "Hello"
print(s)           # Hello

s = '"Hello"'
print(s)           # "Hello"

s = "'Hello'"
print(s)           # 'Hello'

s = "\"'\\Hello\\"
print(s)           # "'\\Hello\\"
````

### บริการของสตริง : String Methods
สมมติให้ `s = "  Hello World "` (ความยาวดัชนี 0 ถึง 13)

- `len(s)` ได้ `14`, `len("")` ได้ `0`
- `s.lower()` ได้ `"  hello world "` (แปลงอักขระเป็นพิมพ์เล็กทั้งหมด)
- `s.upper()` ได้ `"  HELLO WORLD "` (แปลงอักขระเป็นพิมพ์ใหญ่ทั้งหมด)
- `s.strip()` ได้ `"Hello World"` (ตัดช่องว่างทางซ้ายและขวาออกทั้งหมด)
- `s.find("o")` ได้ `6` (หาตำแหน่งแรกที่เจอคำที่ค้นหา)
- `s.find("ex")` ได้ `-1` (คืนค่า -1 เมื่อไม่พบคำที่ค้นหา)
- `s.find("o", 7)` ได้ `9` (เริ่มหาตั้งแต่ index ที่ 7 เป็นต้นไป)

## 7.3 More String Processing Examples (ตัวอย่างการประมวลผลสตริงเพิ่มเติม)

### Method Chaining vs Function Composition

การเรียกใช้งานฟังก์ชันและเมธอดซ้อนกันสามารถเขียนได้ 2 รูปแบบ:

1. **Function Composition (การเรียกฟังก์ชันซ้อนฟังก์ชัน):**
   ```python
   x = math.radians(d)
   s = math.sin(x)
   y = abs(s)
   r = round(y, 2)
   
   # หรือเขียนซ้อนกันแบบรวบเดียว:
   r = round(abs(math.sin(math.radians(d))), 2)
   ```

2. **Method Chaining (การเรียกเมธอดต่อกันเป็นลูกโซ่):**
   ```python
   line1 = input()
   line2 = line1.strip()
   line3 = line2.upper()
   i = line3.find("OK")
   
   # หรือเขียนต่อกันเป็นลูกโซ่:
   i = input().strip().upper().find("OK")
   ```

````python
import math

# ตัวอย่าง 7.3.1: Function Composition
d = 45
r = round(abs(math.sin(math.radians(d))), 2)
print("r =", r)

# ตัวอย่าง 7.3.2: Method Chaining
text = "  hello world ok  "
i = text.strip().upper().find("OK")
print("index of OK =", i)
````

### ข้อควรระวัง (Caveats In String Processing)

- **ห้ามเปลี่ยนค่าภายในสตริง (String Immutability):**
  - `s[2] = "a"` ❌ (`TypeError`)
  - `s[3:7] = "-^o^-"` ❌ (`TypeError`)

- **String Methods ไม่เปลี่ยนค่าตัวสตริงเดิม:**
  - `s = "HellO"` 
  - `s.lower()` ได้ `'hello'` แต่ `s` ยังคงเก็บค่าเดิมคือ `'HellO'`

- **การเปลี่ยนค่าต้องนำผลลัพธ์ไปเก็บในตัวแปรใหม่หรือตัวแปรเดิม:**
  - `s = "HellO"` 
  - `s = s.lower()` แบบนี้ `s` จะเก็บค่าใหม่คือ `"hello"`

````python
# ตัวอย่าง 7.3.3: การทดสอบ String Immutability และการ Re-assign ตัวแปร
s = "HellO"
s.lower() # ได้ 'hello' แต่ไม่ได้เปลี่ยนค่าใน s
print("s หลัง s.lower():", s)

s = s.lower() # เอาผลลัพธ์มาเก็บคืนใน s
print("s หลัง s = s.lower():", s)
````

- `fout.write("Text\n")` ใช้เขียนบรรทัดสตริงลงไฟล์ หากต้องการขึ้นบรรทัดใหม่**ต้องใส่ `\n` เอง**
- `fout.close()` บันทึกเสร็จแล้วต้องปิดแฟ้มเสมอ

### ตัวอย่าง 7.4.3: การเขียนข้อความลงแฟ้มเบื้องต้น

````python
fout = open("sample_output.txt", "w")
fout.write("First line")
fout.write("Text\n")
fout.close()
print("บันทึกเสร็จแล้วก็ปิดแฟ้ม")
````
