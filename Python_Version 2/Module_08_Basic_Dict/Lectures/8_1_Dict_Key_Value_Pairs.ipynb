## 8.1 Dict: Key-Value Pairs (คู่อันดับคีย์และค่า)

### การเปรียบเทียบ list กับ dict
- **list:** เก็บข้อมูลเป็นรายการเรียงจากซ้ายไปขวา ใช้จำนวนเต็ม (Index: 0, 1, 2, ...) เป็นตัวระบุตำแหน่งในการเข้าถึงข้อมูลใน list เช่น `x = ["MO", "TU", "WE", "TH", "FR", "SA", "SU"]` ดึง `x[2]` ได้ `"WE"` 
- **dict:** เก็บข้อมูลเป็นคู่ ๆ `key: value` ใช้ `key` เป็นตัวระบุ `value` ที่ต้องการใน dict (`key` เป็นได้ทั้ง `int`, `float`, `str`, ...)

### ตัวอย่าง 8.1.1: การเปรียบเทียบการใช้งาน List และ Dict

```python
# การใช้ list
x = ["MO", "TU", "WE", "TH", "FR", "SA", "SU"]
print("x[2] =", x[2])
print("x[4] =", x[4])
print("x[-1] =", x[-1])

# การใช้ dict
d = {"MO":"จ", "TU":"อ", "WE":"พ", "TH":"พฤ", "FR":"ศ", "SA":"ส", "SU":"อา"}
print('d["WE"] =', d["WE"])
print('d["TH"] =', d["TH"])
print('d["SA"] =', d["SA"])
```

### dict เหมือน mapping function (คีย์ห้ามซ้ำ ค่าซ้ำได้)
พจนานุกรมทำหน้าที่เหมือนฟังก์ชันจับคู่ (Mapping function):
- **keys ห้ามซ้ำกัน** (เปรียบเหมือน Domain)
- **values ซ้ำกันได้** (เปรียบเหมือน Range)

### ตัวอย่าง 8.1.2: dict จับคู่รหัสนิสิตกับเกรด

```python
grade = {
    "6130186221": "A",
    "6230221221": "A",
    "6231009821": "B",
    "6230543921": "C",
    "6230431521": "B",
    "6230276821": "A"
}

print('เกรดของ 6130186221 คือ:', grade["6130186221"])
print('เกรดของ 6230543921 คือ:', grade["6230543921"])
```

### การเข้าถึงข้อมูลด้วย key และข้อควรระวังเรื่อง KeyError
- การส่ง `key` ให้ `dict` จะได้ `value` ออกมา เช่น `v = grade["6231010621"]`
- **ถ้าใช้คีย์ที่ไม่มีอยู่ใน dict จะเกิดความผิดพลาด (`KeyError`)**
- มีแต่การส่ง `key` แล้วได้ `value` **ไม่มีแบบส่ง `value` แล้วได้ `key`**
