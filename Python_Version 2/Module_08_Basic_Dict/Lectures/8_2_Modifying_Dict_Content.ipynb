## 8.2 Modifying Dict Content (การเพิ่มและการเปลี่ยน value ใน dict)

### การเพิ่มและการเปลี่ยน value ใน dict
การกำหนดค่าให้คีย์ด้วยไวยากรณ์ `dict[key] = value` มีการทำงาน 2 กรณี:
1. **การเพิ่มคู่อันดับใหม่:** หาก `key` นั้นยังไม่มีอยู่ใน `dict` จะเป็นการสร้างคู่อันดับใหม่ทันที
2. **การเปลี่ยนค่าเดิม:** หาก `key` นั้นมีอยู่แล้วใน `dict` จะเป็นการเปลี่ยน `value` เดิมเป็นค่าใหม่

```python
def read_data():
    x = input().split(", ")
    r = {}
    r["ID"] = x[0]
    r["Name"] = x[1]
    r["Birthdate"] = [int(x[2]), int(x[3]), int(x[4])]
    return r
```
