## 1. 🤖 Function คืออะไร?

Function คือ **ลำดับคำสั่ง (Sequence of commands)** ที่ถูกเก็บไว้รวมกัน โดยมีหลักการสำคัญ 3 ข้อ:

| หลักการ | ความหมาย |
|---|---|
| **Abstraction** | รับ input → ทำงาน (ตามหน้าที่/argument) → ส่งกลับผลลัพธ์ |
| **Reusability** | เขียนครั้งเดียว ใช้ได้หลายโปรแกรม |
| **Easy to maintenance** | แก้ไขที่เดียว ทุกที่ที่ใช้งานก็ถูกต้องหมด |
## Do something to Data

* **Do something** ➔ Functions (ฟังก์ชัน/คำสั่ง)
* **Data** ➔ Values (ค่าข้อมูล)

---

### กระบวนการทำงาน (The Process)

**Value** ➔ `[ Function ⚙️ ]` ➔ **New Value**

*(นำค่าข้อมูลเดิม เข้าสู่กระบวนการของฟังก์ชัน เพื่อให้ได้ค่าข้อมูลใหม่ผลลัพธ์ออกมา)*

---

### รูปแบบการจัดการข้อมูล (Types of Functions)

| Standalone Functions | Methods of class | Operations (Magic Methods) |
| :--- | :--- | :--- |
| `print()` | `upper()` | `+`, `/`, `>`, `<` |
| `type()` | `replace()` | `==`, `in`, `or` |


ตัวอย่าง built-in เบื้องต้น:
```python
x = len("python")   # len() รับ input ส่ง output กลับมา
print(x)
```

## Python Standard Library

คลังคำสั่งมาตรฐานของ Python จะแบ่งออกเป็นโมดูลต่างๆ โดยมีทั้งแบบที่พร้อมใช้งานได้เลย และแบบที่ต้องเรียกเข้ามา (import) ก่อน

### 1. Built-in Module
*(คำสั่งพื้นฐานที่พร้อมใช้งานทันที ไม่ต้อง import)*
* **Functions (ฟังก์ชันเดี่ยว):** `len()`, `type()`, `print()`
* **Class str (คำสั่งเฉพาะของข้อความ):** `upper()`, `replace()`
* **Class int (คำสั่งเฉพาะของตัวเลข):** `to_byte()`

### 2. Math Module 
*(ต้องใช้คำสั่ง `import math` ก่อนถึงจะใช้งานได้)*
* **Functions (ฟังก์ชันคณิตศาสตร์):** `abs()`, `round()`, `floor()`

---

````python
# การใช้ Built-in Functions (ใช้ได้เลยไม่ต้อง import)
word = "Hello"
num = 10
print("\n--- Built-in Functions ---")
print(len(word))  # len() นับจำนวนตัวอักษร
print(type(word)) # type() เช็คชนิดข้อมูล

# Methods ของ String
print("\n--- Methods ของ String ---")
print(word.upper()) # upper() แปลงตัวอักษรเป็นตัวพิมพ์ใหญ่
print(num.bit_length()) # bit_length() แปลงตัวเลขเป็น binary และนับจำนวน bit

pi_value = 3.14159
# abs() และ round() จริงๆแล้วเป็น Built-in function ใน Python 
print("ค่าสัมบูรณ์ของ -10 คือ:", abs(-10))
print("ปัดเศษ pi_value คือ:", round(pi_value, 2))
````

## 3. 📦 Function-module

**Module** คือไฟล์ Python ที่รวบรวมฟังก์ชันและค่าคงที่ (constants) ไว้ด้วยกัน เช่น `math` module:
- มีฟังก์ชัน: `exp()`, `log()`, `sqrt()`, `sin()`, `cos()` ...
- มีค่าคงที่: `pi` = π

วิธีเรียกใช้ฟังก์ชันใน module:
```python
from module_name import function_name
```

````python
# เรียกใช้ฟังก์ชันใน module
from math import log
log(2)
````

````python
# Import ทุกฟังก์ชัน
from math import *
print(pi)
print(log(10))
print(sqrt(16))
````

````python
# เปลี่ยนชื่อฟังก์ชันใน module
from math import log as mylog
mylog(2)
````

## 4. ✏️ การสร้างฟังก์ชันเอง (`def`)

```python
def ชื่อฟังก์ชัน(parameter1, parameter2):
    # โค้ดที่ต้องการทำ
    return ผลลัพธ์
```

````python
# ฟังก์ชันไม่มี parameter
def greet():
    print("Hello")
    print("Welcome to Python")

greet()
print("----")
greet()
````

````python
# ฟังก์ชันที่มี parameter และ return
def multiply(x,y):
    ans = x * y
    return ans
    
result1 = multiply(4,5)
result2 = multiply(6,5)

print(f"4*5 = {result1}")
print(f"6*5 = {result2}")
````

## 5. 🔧 Default Argument Values (ค่าเริ่มต้น)

กำหนดค่า default ให้ parameter ได้ ถ้าผู้เรียกฟังก์ชันไม่ส่งค่ามา ก็จะใช้ค่า default นั้นแทน

````python
def show_info(name, salary=15000, lang="Python", age=25):
    print('Name: %s' %name)
    print('Salary: %d' %salary)
    print('Language: %s' %lang)
    print(f"Age = {age}")
````

````python
# calling function - ส่งแค่ชื่อ ค่าที่เหลือใช้ default
show_info('Ploy')
````

````python
# ส่งค่าทั้งหมด
show_info('Ploy2',salary=25000,lang='Java',age=20)
````

## 6. 🔑 Keyword Arguments

เรียกใช้ฟังก์ชันโดยใส่ **ชื่อของ parameter** ตอนส่งอาร์กิวเมนต์ ทำให้ไม่ต้องสนใจลำดับ

````python
def create_button(id, color='#ffffff', text='Button', size=16):
    print('Button ID: %d' % id)
    print('Attributes:')
    print('Color: %s' % color)
    print('Text: %s' % text)
    print('Size: %d px' % size)
    print()
````

````python
# เรียกด้วย positional argument
create_button(10)
# เรียกด้วย keyword argument บางส่วน
create_button(11, color='#4280f4', text='Sign up')

# ใช้ keyword argument ทั้งหมด (ไม่ต้องสนใจลำดับ)
create_button(id=12, color='#323f54', size=24)
create_button(color='#1cb22b', text='Log in', size=32, id=13)
````

## 7. λ Lambda Expressions (Anonymous Function)

**Lambda** คือ ฟังก์ชันแบบ **ไม่มีชื่อ (anonymous function)** เขียนในบรรทัดเดียว

```python
lambda parameter: expression
```

> 💡 มักใช้กับ `filter()` และ `map()` เพื่อย่อโค้ดให้สั้นลง

````python
# Lambda ที่รับ 1 parameter
f = lambda x: x+1
print(f(2))

# Lambda ที่รับ 2 parameters
g = lambda a,b: (a+b)/2
print(g(3,5))
print(g(10,11))
````

````python
# Lambda ที่ return lambda (Closure)
def make_incrementor(n):
    return lambda x: x+n

f = make_incrementor(13)
print(f(0))
print(f(1))
print(f(5))
````

````python
# Lambda กับ Built-in function: filter() และ map()
numbers = [2, 15, 5, 7, 10, 3, 28, 30]

# filter: กรองเฉพาะตัวที่หาร 5 ลงตัว
print(list(filter(lambda x: x % 5 == 0,numbers)))

# map: คูณทุกตัวด้วย 2
print(list(map(lambda x: x * 2,numbers)))
````
