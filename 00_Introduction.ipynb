## 1. Jupyter Notebook คืออะไร?

**Jupyter Notebook** คือเครื่องมือที่ช่วยให้เราเขียน Code + บันทึก + เห็นผลลัพธ์ได้ทันทีในหน้าเดียวกัน  

มี 2 ประเภทหลักของ **Cell** :
- 🟦 **Code Cell** → เขียน Python Code ได้เลย
- 📝 **Markdown Cell** → เขียนข้อความ / หัวข้อ / อธิบาย (เหมือน cell นี้)

### ⌨️ Shortcut ที่ใช้บ่อย
| Shortcut | ความหมาย |
|---|---|
| `Shift + Enter` | รัน cell แล้วไปต่อ |
| `Ctrl + Enter` | รัน cell อยู่กับที่ |
| `A` | เพิ่ม cell ด้านบน |
| `B` | เพิ่ม cell ด้านล่าง |
| `DD` | ลบ cell |
| `M` | เปลี่ยนเป็น Markdown cell |
| `Y` | เปลี่ยนเป็น Code cell |

#### Special Character

````python
# \" Double Quote
print("Hello \"Python\"")

# \' Single Quote
print("Hello \'Python\' ")

# \\ Backslash
print("Hello \\Python")
print("Path : C:\\Users\\Admin")

#\n New Line
print("Hello Python\n")
print("Hello \nPython")

#\t Tab
print("Hello \tPython")
````

## 5. Library คืออะไร?

**Library** คือชุดของ Code สำเร็จรูปที่คนอื่นเขียนไว้ให้แล้ว เราแค่ **นำมาใช้** ได้เลย  
เหมือนการซื้อของสำเร็จรูปมาทำอาหาร ไม่ต้องทำเองทุกอย่างตั้งแต่ต้น 🍳

### วิธี Import Library
```python
import library_name                    # import ทั้งหมด
import library_name as alias           # ตั้งชื่อย่อ
from library_name import function_name # import เฉพาะฟังก์ชัน
```

## 6. ตัวอย่าง Library ยอดนิยม

### 6.1 `math` — คณิตศาสตร์ (Built-in, ไม่ต้องติดตั้ง)

````python
import math  # นำเข้า library ชื่อ math

# ใช้งาน math
print('pi = '  , math.pi)
````

### 6.2 `random` — สุ่มตัวเลข (Built-in)

````python
import random  # นำเข้า library random

# สุ่มตัวเลข
print('Random number', random.random())

# สุ่มจาก List
fruits = ['apple', 'banana', 'cherry']
print('random', random.choice(fruits))
````

### 6.3 `datetime` — วันและเวลา (Built-in)

````python
from datetime import datetime  # import เฉพาะ class datetime จาก library datetime

# ดูวันเวลาปัจจุบัน
now = datetime.now()
print('Current date and time:', now)
````

### 6.4 `numpy` — คำนวณตัวเลขและอาร์เรย์ (ต้องติดตั้ง)

> ถ้ายังไม่ได้ติดตั้ง ให้รันคำสั่ง: `pip install numpy`

````python
import numpy as np  # ตั้งชื่อย่อว่า np (นิยมใช้กันทั่วไป)

# สร้าง Array
score = np.array([90, 85, 78])
print(score)
print('score mean', np.mean(score))
print('score std', np.std(score))
print('score max', np.max(score))
print('score min', np.min(score))
````

### 6.6 `matplotlib` — วาดกราฟ (ต้องติดตั้ง)

> ถ้ายังไม่ได้ติดตั้ง ให้รันคำสั่ง: `pip install matplotlib`

````python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# ====================================================
# ตั้งค่าฟอนต์ภาษาไทย (ต้องรันก่อนวาดกราฟเสมอ)
# ====================================================
# รายการฟอนต์ภาษาไทยที่มีใน Windows
thai_fonts    = ["TH Sarabun New", "Tahoma", "Cordia New", "Leelawadee UI"]
available     = [f.name for f in fm.fontManager.ttflist]
selected_font = next((f for f in thai_fonts if f in available), None)

if selected_font:
    plt.rcParams["font.family"] = selected_font
    print(f"✅ ใช้ฟอนต์: {selected_font}")
else:
    print("⚠️ ไม่พบฟอนต์ภาษาไทย — ลองติดตั้ง 'TH Sarabun New'")

# ====================================================
# ข้อมูลที่จะวาดกราฟ
months  = ["ม.ค.", "ก.พ.", "มี.ค.", "เม.ย.", "พ.ค."]
revenue = [15000, 18000, 12000, 20000, 17000]

# วาด Bar Chart
plt.figure(figsize=(8, 4))
plt.bar(months, revenue, color="steelblue")
plt.title("ยอดขายรายเดือน", fontsize=16)
plt.xlabel("เดือน", fontsize=13)
plt.ylabel("ยอดขาย (บาท)", fontsize=13)
plt.tight_layout()
plt.show()
````
