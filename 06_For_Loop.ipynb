## 1. 🔄 For Loop คืออะไร?

**For Loop** ใช้สำหรับ **วนซ้ำตามจำนวนรอบที่แน่นอน** หรือ **วนซ้ำดึงข้อมูลจากโครงสร้างข้อมูล (เช่น List, Tuple, String) ทีละตัว** จนกว่าจะครบ

เปรียบเทียบกับ While Loop:
- **While Loop:** วนซ้ำตราบใดที่เงื่อนไขยังเป็นจริง (True)
- **For Loop:** หยิบของในกล่องออกมาทีละชิ้นจนหมดกล่อง

---
## 2. 🍎 การใช้ For Loop กับ List และ Tuple

```python
# ตัวอย่างพื้นฐานกับ List
fruits = ["แอปเปิ้ล", "กล้วย", "ส้ม", "องุ่น"]

for fruit in fruits:
    print(f"ฉันชอบกิน {fruit}")
```

---
## 3. 🔠 การใช้ For Loop กับข้อความ (String)
นอกจาก List หรือ Tuple แล้ว เราสามารถใช้ For Loop วนตัวอักษรแต่ละตัวใน String ได้ด้วย

```python
word = "Python"

for letter in word:
    print(letter)
```

## 4. 🔢 ฟังก์ชัน `range()`
เมื่อเราต้องการวนลูปเป็นจำนวนรอบตัวเลขที่แน่นอน เราจะใช้ฟังก์ชัน `range()`

- `range(stop)` : สร้างลำดับตัวเลขตั้งแต่ 0 ถึง (stop - 1)
- `range(start, stop)` : สร้างลำดับตัวเลขตั้งแต่ start ถึง (stop - 1)
- `range(start, stop, step)` : กำหนดความห่างของการนับ (step) ได้

```python
# ตัวอย่าง: วน 5 รอบ (0 ถึง 4)
print("ตัวอย่าง range(5):")
for i in range(5):
    print(i, end=" ")
print()

# ตัวอย่าง: วนจากเลข 1 ถึง 5
print("\nตัวอย่าง range(1, 6):")
for i in range(1, 6):
    print(i, end=" ")
print()

# ตัวอย่าง: เพิ่มขึ้นทีละ 2 (เลขคี่)
print("\nตัวอย่าง range(1, 10, 2):")
for i in range(1, 10, 2):
    print(i, end=" ")
print()

# ตัวอย่าง: นับถอยหลัง (ลดทีละ 1)
print("\nตัวอย่าง range(5, 0, -1):")
for i in range(5, 0, -1):
    print(i, end=" ")
```

## 5. 🛑 การใช้ `break` และ ⏭️ `continue`

- `break` : หยุดลูปและออกจากลูปทันที (เหมือน While)
- `continue` : ข้ามการทำงานในรอบนั้น และเริ่มรอบต่อไปทันที

```python
print("🎯 การใช้ break: (จะหยุดเมื่อเจอเลข 3)")
for num in range(1, 6): # 1, 2, 3, 4, 5
    if num == 3:
        break
    print(num, end=" ")
print()

print("\n⏭️ การใช้ continue: (จะข้ามเลข 3)")
for num in range(1, 6): # 1, 2, 3, 4, 5
    if num == 3:
        continue
    print(num, end=" ")
```

## 6. 🔁 `for-else`
เช่นเดียวกับ While ลูป สำหรับ For ลูปนั้น `else` จะทำงานก็ต่อเมื่อ **ลูปทำงานจนจบครบทุกตัวโดยไม่โดน `break` ดักไว้**

```python
names = ["Alice", "Bob", "Charlie"]
search_name = "David"

for name in names:
    if name == search_name:
        print(f"🎉 เจอคุณ {search_name} ในระบบ!")
        break
else:
    print(f"❌ ไม่พบชื่อ {search_name} ในรายชื่อ")
```

## 7. 🪆 Nested Loop (ลูปซ้อนลูป)
เราสามารถเอา For Loop วางซ้อนกันได้ เพื่อใช้งานกับโครงสร้างที่มีหลายมิติ เช่น ตาราง, List ซ้อน List

```python
# ตัวอย่าง: สูตรคูณแม่ 2 ถึงแม่ 4
for i in range(2, 5): # 2 , 3 , 4          # ลูปนอก: กำหนดแม่สูตรคูณ
    print(f"--- แม่ {i} ---")
    for j in range(1, 13): # 1 , 2 , ... , 12   # ลูปใน: ตัวคูณ (1 ถึง 12)
        print(f"{i} x {j} = {i * j}")
    print() # ย่อหน้าขึ้นแม่ใหม่
```
