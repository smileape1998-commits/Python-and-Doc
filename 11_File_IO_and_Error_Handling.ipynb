## 1. 📖 การเปิดและอ่านไฟล์

การใช้โครงสร้าง `with open(..., mode) as file:` เป็นวิธีที่ดีที่สุด เพราะมันจะช่วย **ปิดไฟล์ให้อัตโนมัติ** เมื่อทำงานเสร็จ

| Mode | ความหมาย |
|---|---|
| `'r'` | Read — อ่านอย่างเดียว (พังถ้าไม่มีไฟล์) |
| `'w'` | Write — เขียนทับของเดิม (สร้างใหม่ถ้าไม่มีไฟล์) |
| `'a'` | Append — เขียนต่อท้ายไฟล์เดิม |

````python
# สร้างไฟล์จำลองเพื่อทดสอบ (รันเพื่อให้มีไฟล์ก่อน)
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("บรรทัดที่ 1: สวัสดี\nบรรทัดที่ 2: Python\nบรรทัดที่ 3: สบายดีไหม")

print("สร้างไฟล์ data.txt เสร็จแล้ว")
````

````python
# การอ่านทั้งหมดในรอบเดียว
print("--- Read All ---")
with open("data.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)
````

````python
# การอ่านทีละบรรทัด (แนะนำสำหรับไฟล์ใหญ่)
print("--- Read Line by Line ---")
with open("data.txt", "r", encoding="utf-8") as f:
    for line in f:
        print("->", line.strip())  # .strip() ลบ \n ที่ซ่อนอยู่ออก
````

## 2. 📝 การเขียนข้อมูลลงไฟล์
- Mode `w` (Write): เขียนทับของเดิม (ถ้าไม่มีไฟล์จะสร้างใหม่)
- Mode `a` (Append): เขียนต่อท้ายไฟล์เดิม

````python
# เขียนต่อท้าย (Append)
with open("data.txt", "a", encoding="utf-8") as f:
    f.write("\nบรรทัดที่ 4: เพิ่มข้อมูลจ้า")

# ลองอ่านดูอีกครั้ง
with open("data.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 3. 🛡️ การจัดการ Error เบื้องต้น (`try-except`)

เวลาโปรแกรมพัง (Error) มันจะหยุดทำงานทันที แต่ถ้าเรานำ `try-except` มาครอบไว้ โปรแกรมเราจะไม่พัง และยังทำงานต่อไปได้

```
try:
    # โค้ดที่อาจพัง
except ชื่อError:
    # จัดการเมื่อพัง
```

````python
# ตัวอย่าง: ZeroDivisionError
try:
    answer = 10 / 0
    print(answer)
except ZeroDivisionError:
    print("❌ ผิดพลาด: ห้ามหารด้วยศูนย์นะครับ!")

print("✅ โปรแกรมยังทำงานต่อได้ ไม่ค้าง!")
````

````python
# ตัวอย่าง: ValueError (แปลงข้อความที่ไม่ใช่ตัวเลขเป็น int)
try:
    age = int("twenty")  # พัง! ไม่ใช่ตัวเลข
except ValueError:
    print("⚠️ กรุณากรอกตัวเลขเท่านั้น!")

# ตัวอย่าง: รับค่าจาก input พร้อม try-except
try:
    num = int(input("ใส่ตัวเลข: "))
    print(f"คุณใส่: {num}")
except ValueError:
    print("⚠️ นั่นไม่ใช่ตัวเลขนะครับ!")
````

## 4. 🔗 การใช้งานร่วมกัน (อ่านไฟล์ + ป้องกันพัง)
ถ้าเราพยายามเปิดไฟล์ที่ไม่มีอยู่จริง โปรแกรมจะพัง (`FileNotFoundError`) เราแก้ได้ด้วย Try-Except

````python
try:
    with open("not_exist_file.txt", "r", encoding="utf-8") as f:
        text = f.read()
        print(text)
except FileNotFoundError:
    print("🚨 ไม่พบไฟล์ที่ระบุ กรุณาตรวจสอบชื่อไฟล์อีกครั้ง")
````

````python
# จับ Error หลายประเภทพร้อมกัน
filename = "data.txt"

try:
    with open(filename, "r", encoding="utf-8") as f:
        content = f.read()
    print("✅ อ่านไฟล์สำเร็จ")
    print(content)
except FileNotFoundError:
    print(f"❌ ไม่พบไฟล์ชื่อ '{filename}'")
except PermissionError:
    print(f"❌ ไม่มีสิทธิ์อ่านไฟล์ '{filename}'")
````
