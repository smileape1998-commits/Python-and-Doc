## 1. 🧠 แนวคิดการตัดสินใจ

ในชีวิตจริง เราตัดสินใจตลอดเวลา:
- **ถ้า** ฝนตก → เอาร่มไปด้วย **ไม่งั้น** → ไม่ต้องเอา
- **ถ้า** คะแนน >= 80 → เกรด A, **ถ้า** >= 70 → เกรด B ...

Python ใช้ `if`, `elif`, `else` ทำสิ่งเดียวกัน

```
โปรแกรมเริ่ม
     ↓
  ┌──────┐    True    ┌────────────────┐
  │เงื่อนไข│──────────→│ทำเมื่อ True   │
  └──────┘           └────────────────┘
     │ False                ↓
     ↓              ┌────────────────┐
  ┌──────────────┐  │  ต่อไป...     │
  │ทำเมื่อ False │→ └────────────────┘
  └──────────────┘
```

> ⚠️ Python ใช้ **Indentation (เว้น 4 ช่องว่าง)** แทนวงเล็บ `{}`

## 2. 📌 `if` พื้นฐาน

````python
# โครงสร้าง if พื้นฐาน
age = 20

if age >= 18:            # ← เงื่อนไข
    print("✅ ผู้ใหญ่แล้ว")  # ← รันเมื่อเงื่อนไขเป็น True (ต้องเว้น 4 ช่องว่าง)

print("บรรทัดนี้รันเสมอ ไม่ว่าเงื่อนไขจะเป็นอะไร")
````

````python
# ลองเปลี่ยนค่าแล้วดูผล
temperature = 38.5

if temperature > 37.5:
    print(f"🤒 อุณหภูมิ {temperature}°C — มีไข้! ควรพักผ่อน")
````

## 3. 🔀 `if-else`

````python
score = 55

if score >= 60:
    print(f"✅ คะแนน {score} — ผ่าน!")
else:                           # ← ถ้าเงื่อนไขด้านบนเป็น False
    print(f"❌ คะแนน {score} — ไม่ผ่าน")
````

````python
# ตรวจสอบเลขคู่/คี่
number = 7

if number % 2 == 0:
    print(f"{number} เป็นเลขคู่")
else:
    print(f"{number} เป็นเลขคี่")
````

## 4. 🔢 `if-elif-else` — เงื่อนไขหลายระดับ

ใช้เมื่อมีตัวเลือกมากกว่า 2 ทาง  
Python จะตรวจเงื่อนไขจากบนลงล่าง และหยุดทันทีที่เจอ `True`

````python
score = 75

if score >= 80:
    grade = "A"
elif score >= 70:    # ← elif = else if (ถ้าไม่เข้าใน if ด้านบน)
    grade = "B"
elif score >= 60:
    grade = "C"
elif score >= 50:
    grade = "D"
else:               # ← ถ้าไม่ตรงเงื่อนไขใดเลย
    grade = "F"

print(f"คะแนน {score} → เกรด: {grade}")
````

````python
# ตัวอย่าง: ค่าโดยสาร BTS ตามอายุ
age = int(input("อายุของคุณ: "))

if age < 3:
    price = 0
    msg   = "เด็กเล็ก ฟรี!"
elif age < 12:
    price = 20
    msg   = "เด็ก"
elif age < 60:
    price = 42
    msg   = "ผู้ใหญ่"
else:
    price = 20
    msg   = "ผู้สูงอายุ"

print(f"🚆 อายุ {age} ปี ประเภท: {msg} ราคา: {price} บาท")
````

## 5. 🔣 ตัวดำเนินการเปรียบเทียบ

| ตัวดำเนินการ | ความหมาย | ตัวอย่าง | ผล |
|------------|----------|---------|----|
| `==` | เท่ากัน | `5 == 5` | True |
| `!=` | ไม่เท่ากัน | `5 != 3` | True |
| `>` | มากกว่า | `10 > 5` | True |
| `<` | น้อยกว่า | `5 < 3` | False |
| `>=` | มากกว่าหรือเท่ากัน | `5 >= 5` | True |
| `<=` | น้อยกว่าหรือเท่ากัน | `4 <= 5` | True |
| `in` | อยู่ใน | `"a" in "cat"` | True |
| `not in` | ไม่อยู่ใน | `5 not in [1,2,3]` | True |

````python
fruits = ["แอปเปิ้ล", "มะม่วง", "กล้วย"]

# ตรวจสอบการมีอยู่ด้วย 'in'
x = "มะม่วง"
if x in fruits:
    print(f"✅ '{x}' มีอยู่ในรายการ")
else:
    print(f"❌ '{x}' ไม่มีในรายการ")

# ตรวจสอบ string
email = "user@mail.com"
if "@" in email and "." in email:
    print(f"✅ '{email}' เป็นอีเมลที่ถูกต้อง")
else:
    print(f"❌ '{email}' ไม่ใช่อีเมลที่ถูกต้อง")
````

## 6. 🔗 ตัวดำเนินการทางตรรกะ

| Operator | ความหมาย | ผล |
|----------|-----------|----|
| `and` | ทั้งสองต้อง True | True and False = **False** |
| `or` | อย่างน้อยหนึ่ง True | True or False = **True** |
| `not` | กลับค่า | not True = **False** |

````python
# AND — ทั้งสองต้องจริง
age    = 20
has_id = True

if age >= 18 and has_id:
    print("✅ เข้าได้")
else:
    print("❌ เข้าไม่ได้")


# OR — อย่างน้อยหนึ่งต้องจริง
is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("✅ วันนี้ไม่ต้องทำงาน")
else:
    print("❌ วันนี้ต้องทำงาน")


# NOT — กลับค่า
is_banned = False
if not is_banned:
    print("Normal user, can access")
````

````python
# ตัวอย่างรวม: เงื่อนไขซับซ้อน
score  = 75
attend = 85  # % การเข้าเรียน

if score >= 60 and attend >= 80:
    print("✅ ผ่าน (คะแนนผ่านและเข้าเรียนครบ)")
elif score >= 60 and attend < 80:
    print("⚠️ คะแนนผ่าน แต่เข้าเรียนไม่ครบ")
elif score < 60 and attend >= 80:
    print("⚠️ เข้าเรียนครบ แต่คะแนนไม่ผ่าน")
else:
    print("❌ ไม่ผ่าน")
````

## 7. 🎭 Nested If (if ซ้อน if)

````python
# สถานการณ์: ซื้อบัตรคอนเสิร์ต
is_member = True
balance   = 1000
price     = 800

if is_member:
    discount = 0.2    # สมาชิกลด 20%
    final    = price * (1 - discount)
    print(f"💳 ราคาสมาชิก: {final} บาท")
    if balance >= final:
        print(f"✅ ซื้อสำเร็จ! เงินเหลือ {balance - final} บาท")
    else:
        print(f"❌ เงินไม่พอ ขาดอีก {final - balance} บาท")
else:
    print(f"💰 ราคาปกติ: {price} บาท")
    
    if balance >= price:
        print(f"✅ ซื้อสำเร็จ!")
    else:
        print(f"❌ เงินไม่พอ")
````

## 9. ⚠️ ข้อผิดพลาดที่พบบ่อย

````python
# ❌ ผิดที่ 1: ใช้ = แทน == เปรียบเทียบ
x = 5
# if x = 5:   ← SyntaxError!
if x == 5:   # ✅ ถูก
    print("✅ x เท่ากับ 5")

# ❌ ผิดที่ 2: Indentation ผิด
score = 80
if score >= 60:
    print("ผ่าน")     # ✅ เว้น 4 ช่องว่าง
# print("ผ่าน")       # ❌ ไม่ได้เว้น = รันทุกกรณี

# ❌ ผิดที่ 3: ลืม : ท้ายเงื่อนไข
# if score >= 60   ← SyntaxError: ลืม :
if score >= 60:
    print("✅ มี : ท้ายเสมอ")
````

````python
# ❌ ผิดที่ 4: เงื่อนไขลำดับผิด
score = 85

# แบบผิด — เงื่อนไขกว้างเกินไปอยู่ก่อน
if score >= 60:
    print("C")    # ← จะได้ C เสมอ แม้คะแนน 85!
elif score >= 80:
    print("A")    # ← ไม่มีทางถึงบรรทัดนี้ได้!

print()

# แบบถูก — เริ่มจากกว้างน้อยสุดก่อน
if score >= 80:
    print("A")    # ✅ จะได้ A
elif score >= 60:
    print("C")
````
