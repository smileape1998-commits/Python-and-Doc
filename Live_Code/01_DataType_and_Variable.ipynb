## 1. 📦 ตัวแปร (Variable) คืออะไร?

ลองนึกภาพว่า **ตัวแปร** คือ **กล่อง** ที่มีป้ายชื่อ  
เราใส่ข้อมูลลงในกล่อง แล้วเรียกใช้ด้วยชื่อกล่องนั้น

```
        name
       ┌─────────┐
       │  "Ploy" │  ← ค่าที่เก็บอยู่
       └─────────┘
```

**วิธีสร้างตัวแปร:**
```python
ชื่อตัวแปร = ค่า
```

เครื่องหมาย `=` ใน Python **ไม่ใช่** "เท่ากับ" แต่หมายถึง **"กำหนดค่า"** (Assignment)

---
### ✅ กฎการตั้งชื่อตัวแปร

| ✅ ถูกต้อง | ❌ ผิด | เหตุผล |
|-----------|--------|--------|
| `name` | `1name` | ห้ามขึ้นต้นตัวเลข |
| `my_score` | `my score` | ห้ามมีช่องว่าง |
| `_value` | `if` | ห้ามใช้ Keyword ของ Python |
| `score2` | `score-2` | ห้ามใช้ `-` |

> 💡 **Best Practice**: ใช้ `snake_case` — คำเล็กทั้งหมด คั่นด้วย `_`  
> เช่น `first_name`, `total_score`, `is_student`

````python
# การสร้างตัวแปรและกำหนดค่า
name       = "Ploy"        # str
age        = 20            # int
height     = 165.5         # float
is_student = True          # bool

# แสดงผล
````

````python
# สามารถกำหนดหลายตัวแปรพร้อมกันได้
x, y, z = 10, 20, 30
print(x, y, z)

# กำหนดหลายตัวแปรให้ค่าเดียวกัน
a = b = c = 0
print(a, b, c)

# เปลี่ยนค่าตัวแปรได้ตลอดเวลา
score = 80
print(score)
score = 10
print(score)
````

## 2. 🔢 ชนิดข้อมูลหลัก

Python มีชนิดข้อมูลพื้นฐาน 4 แบบที่ต้องรู้:

```
┌──────────┬─────────────┬──────────────────────┐
│  ชนิด   │ Python Name │      ตัวอย่าง        │
├──────────┼─────────────┼──────────────────────┤
│ จำนวนเต็ม│    int      │  42, -10, 0          │
│ ทศนิยม   │    float    │  3.14, -0.5, 2.0     │
│ ข้อความ  │    str      │  "hello", 'สวัสดี'   │
│จริง/เท็จ │    bool     │  True, False         │
└──────────┴─────────────┴──────────────────────┘
```

### 2.1 int — จำนวนเต็ม

````python
# ตัวดำเนินการทางคณิตศาสตร์
print("บวก (+) :", 10 + 3)    # 13
print("ลบ  (-) :", 10 - 3)    # 7
print("คูณ (*) :", 10 * 3)    # 30
print("หาร (/) :", 10 / 3)    # 3.333... (ได้ float เสมอ)
print("หารเต็ม  :", 10 // 3)  # 3  (Floor Division — ตัดทศนิยม)
print("เศษหาร   :", 10 % 3)   # 1  (Modulo)
print("ยกกำลัง  :", 2 ** 10)  # 1024

# ตัวดำเนินการย่อ
n = 5
n += 3  # n = n + 3
print(n)  # 8
````

### 2.2 float — ทศนิยม

````python
# ⚠️ ระวัง: float มีความคลาดเคลื่อนอยู่บ้าง (เรื่องปกติในคอมพิวเตอร์)
pi = 3.1415
price = 19.99
print(pi)
print(price)
# ปัดเศษ
print("Round pi:", round(3.1415111, 2))  # ปัดเป็นทศนิยม 2 ตำแหน่ง
````

### 2.3 str — ข้อความ

````python
name = "Ploy"
last_name = "Smith"
# วิธีรวมข้อความ 3 แบบ
# แบบที่ 1: + (Concatenation)


# แบบที่ 2: f-string ← แนะนำ!


# แบบที่ 3: .format()
print("name: {}, last name: {}".format(name, last_name))


# เมธอดของ str ที่ใช้บ่อย
sentenvce = "Hello World!"
print("upper():", sentenvce.upper())  # HELLO WORLD!
print("lower():", sentenvce.lower())  # hello world!
print("replace():", sentenvce.replace("World", "Python"))  # Hello Python!
print("split():", sentenvce.split())  # ['Hello', 'World!']
print("len():", len(sentenvce))  # 12 (นับจำนวนตัวอักษร)
````

````python
# str เป็น Sequence — เข้าถึงแต่ละตัวอักษรด้วย index ได้
word = "Python"
# 0 1 2 3 4 5
# P y t h o n
# -6-5-4-3-2-1

print("Index 0:", word[0])  # P
print("Index 1:", word[1])  # y
print("Index -2:", word[-2])  # o
print("Index - 1", word[-1])

# ตรวจสอบว่ามีข้อความที่ต้องการอยู่ไหม
print(" 'Py' in word :","Py" in word)
````

### 2.4 bool — จริง / เท็จ

> 💡 `True` และ `False` **ต้องขึ้นต้นด้วยตัวใหญ่** เสมอ!

````python
is_raining = False
has_umbrella = True
# ผลจากการเปรียบเทียบจะเป็น bool
print("Is it raining?", is_raining)
print("Do I have an umbrella?", has_umbrella)

# bool มีค่าตัวเลขด้วย
# print("10 > 5 ", 10 > 5)  # True
# print("6 < 3  ", 6 < 3)  # False
````

## 3. 🔍 การตรวจสอบและแปลงชนิดข้อมูล

````python
# type() — ตรวจสอบชนิดข้อมูล
values = [43 , 3.14 , "Hello", True]
for x in values:
    print(x, "is of type", type(x).__name__) # __name__ คือชื่อของชนิดข้อมูล 'str'
````

````python
# การแปลงชนิดข้อมูล (Type Casting)
print("--- แปลงเป็น int ---")
print(int("42"))      # str → int : 42
print(int(3.99))      # float → int : 3  (ตัดทศนิยม ไม่ปัดเศษ!)
print(int(True))      # bool → int : 1

print("\n--- แปลงเป็น float ---")
print(float("3.14"))  # str → float
print(float(5))       # int → float : 5.0

print("\n--- แปลงเป็น str ---")
print(str(42))        # int → str
print(str(3.14))      # float → str
print("อายุ: " + str(20))  # ต้องแปลงก่อนรวมกับ str

print("\n--- แปลงเป็น bool ---")
print(bool(0))        # 0 → False
print(bool(1))        # 1 → True
print(bool(""))       # str ว่าง → False
print(bool("hello"))  # str ไม่ว่าง → True
````

## 4. ⌨️ การรับข้อมูลจากผู้ใช้ด้วย `input()`

> ⚠️ **สำคัญ**: `input()` **คืนค่าเป็น str เสมอ!**  
> ถ้าต้องการตัวเลข ต้องแปลงด้วย `int()` หรือ `float()` ก่อน

````python
# รับข้อความ
name = input("Enter your name: ")
print("Hello, " + name + "!")
````

````python
# รับตัวเลข — ต้องแปลงชนิดก่อน!
age_str = input("Enter your age: ")
age = int(age_str)  # แปลง str → int
print("You are", age, "years old.")
````

````python
# เขียนกระชับโดยรวม input() + int() ในบรรทัดเดียว
score = int(input("Enter your score: "))
````

## 5. ⚠️ ข้อผิดพลาดที่พบบ่อย (Common Mistakes)

````python
# ❌ ผิดที่ 1: ใช้ = แทน == เปรียบเทียบ
# x = 5
# if x = 5:   # SyntaxError!
#     print("เท่ากัน")

# ✅ ถูก:
x = 5
if x == 5:
    print("✅ ใช้ == เปรียบเทียบ")
````

````python
# ❌ ผิดที่ 2: รวม str กับ int โดยตรง
age = 20
try:
    print("อายุ: " + age)   # TypeError!
except TypeError as e:
    print("❌ Error:", e)

# ✅ ถูก:
print("อายุ: " + str(age))   # แปลงเป็น str ก่อน
print(f"อายุ: {age}")         # หรือใช้ f-string
````

````python
# ❌ ผิดที่ 3: ลืมแปลง input() เป็นตัวเลขก่อนคำนวณ
score_str = "85"   # จำลองว่ารับจาก input()
bonus     = 5

try:
    print(score_str + bonus)   # TypeError!
except TypeError as e:
    print("❌ Error:", e)

# ✅ ถูก:
print("✅ ผลลัพธ์:", int(score_str) + bonus)   # 90
````
