## 1. 📖 Dictionary คืออะไร?

**Dictionary** คือโครงสร้างข้อมูลที่จับคู่ **Key** (คีย์) กับ **Value** (ค่า)
เหมือนพจนานุกรมที่มีคำศัพท์ (Key) และคำแปล (Value)

- สร้างด้วยวงเล็บปีกกา `{ key: value, key: value }`
- คีย์ต้อง **ไม่ซ้ำกัน**
- เข้าถึงข้อมูลผ่าน key ไม่ใช่ index เหมือน List

```python
# การสร้าง Dictionary
my_dict = {
    "name": "Alice",
    "age": 30,}
print(my_dict)

# เข้าถึงข้อมูลด้วย Key
print("ชื่อ:", my_dict["name"])
print("อายุ:", my_dict["age"])
```

## 2. 🔑 การสร้างและเข้าถึงข้อมูลใน Dictionary

```python
student = {
    "name": "สมชาย",
    "age": 18,
    "grade": "A"
}

# เข้าถึงข้อมูล
print("ชื่อนักเรียน:", student["name"])
print("อายุ:", student["age"])

# ใช้ .get() ปลอดภัยกว่า ถ้าคีย์ไม่มีจะคืนค่า None แทนที่จะ Error
print("เกรด:", student.get("grade"))
print("คะแนน:", student.get("score"))  # None (ไม่มี key นี้)
```

## 3. ✏️ การเปลี่ยนแปลงและเพิ่มข้อมูลใน Dictionary

ทำได้ง่ายมาก แค่ใช้ `dict[key] = new_value`
- ถ้า key มีอยู่แล้ว → **อัปเดตค่าเดิม**
- ถ้า key ยังไม่มี → **เพิ่มคู่ key:value ใหม่**

```python
my_dict = {'name': 'Jack', 'age': 26}

# update value (อัปเดตค่าเดิม)
my_dict['age'] = 27
print(my_dict) # Output: {'age': 27, 'name': 'Jack'}

# add item (เพิ่มข้อมูลใหม่)
my_dict['address'] = 'Downtown'
print(my_dict) # Output: {'address': 'Downtown', 'age': 27, 'name': 'Jack'}
```

## 4. 🔄 ฟังก์ชันสำคัญของ Dictionary (`keys`, `values`, `items`)
มักใช้คู่กับ For Loop เพื่อดึงข้อมูลออกมาใช้งาน

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2022
}

# 1. วนซ้ำเพื่อเอาแค่ Key
print("--- Keys ---")
for k in car.keys():
    print(k)

# 2. วนซ้ำเพื่อเอาแค่ Value
print("\n--- Values ---")
for v in car.values():
    print(v)

# 3. วนซ้ำเอาทั้ง Key และ Value (ถูกใช้บ่อยสุด)
print("\n--- Items ---")
for k, v in car.items():
    print(f"{k} -> {v}")
```

## 5. 🧮 Set (เซ็ต) เป็นกลุ่มข้อมูล

**Set** คล้ายกับ List แต่มีคุณสมบัติพิเศษคือ:
- **ไม่เรียงลำดับ** (Unordered)
- **สมาชิกห้ามซ้ำกัน** (Unique items)
- เก็บด้วยวงเล็บปีกกา `{ }`

*มักใช้สำหรับลบข้อมูลที่ซ้ำกันทิ้ง*

```python
{1, 2}
```

```python
# แปลง List เป็น Set
a = set([1, 2, 3])
print(a, type(a))
```

```python
# แปลงจาก tuples เป็น set
aa = (1, 2, 3, 4, 5)
c = list(aa)
set(c)
```

## 6. 🛠️ การจัดการ Set

```python
# ตรวจสอบ membership (ค้นหาข้อมูลในเซ็ต)
a_set = {10, 11, 12, 13, 14, 16, 36, 17, 18, 19}
print(1 in a_set)   # False
print(10 in a_set)  # True
```

```python
# เพิ่มสมาชิกหลายค่าพร้อมกัน (.update)
b = {1, 2, 3}
b.update([4, 5])
print(b)  # {1, 2, 3, 4, 5}
```

```python
# ลบสมาชิก (.discard — ไม่ Error ถ้าไม่มีค่านั้น)
b = {1, 2, 3, 4, 5}
b.discard(4)
print(b)  # {1, 2, 3, 5}
```
