## 1. ✂️ การตัดข้อความ (Slicing)
ข้อความใน Python มองว่าเป็น List ของตัวอักษร เราสามารถตัดเอาบางส่วนมาได้ (คล้ายกับการทำ List Slicing)

รูปแบบ: `text[start:stop:step]`

```python
text = "Hello Python World"

print("ตัวอักษรแรก:", text[0])
print("ตัวอักษรสุดท้าย:", text[-1])
print("ตัด 5 ตัวแรก:", text[0:5])
print("ตัดคำว่า Python:", text[6:12])
print("ข้ามทีละ 2 ตัว:", text[::2])
print("สลับหลังมาหน้า (Reverse):", text[::-1])
```

---
## 2. 🛠️ String Methods ที่ใช้บ่อย
String ใน Python มีฟังก์ชันสำเร็จรูปให้ใช้งานเยอะมาก

```python
msg = "  python is AWESOME!  "

print("ต้นฉบับ:", f"'{msg}'")
print("ตัดช่องว่างหน้าหลัง:", f"'{msg.strip()}'")
print("ตัวพิมพ์เล็กหมด:", msg.lower())
print("ตัวพิมพ์ใหญ่หมด:", msg.upper())
print("ขึ้นต้นด้วยพิมพ์ใหญ่:", msg.strip().capitalize())
print("แทนที่ข้อความ:", msg.replace("python", "JavaScript"))
```

```python
# ตรวจสอบข้อความ
print("เป็นตัวเลขล้วนหรือไม่?", "12345".isdigit())   # True
print("เป็นตัวอักษรล้วนหรือไม่?", "Hello".isalpha())  # True

msg = "  python is AWESOME!  "
print("ขึ้นต้นด้วย python?", msg.strip().startswith("python"))  # True
print("ลงท้ายด้วย !?", msg.strip().endswith("!"))               # True
```

## 3. 🔪 การแยกและรวมข้อความ (`split` และ `join`)
สำคัญมากสำหรับจัดการกับข้อมูลที่เป็น CSV หรือ Log

```python
# Split: ตัด string ออกมาเป็น List
data_string = "apple,banana,orange,grape"
fruits_list = data_string.split(",")

print("ผลลัพธ์การ Split:", fruits_list)  # กลายเป็น List แล้ว
```

```python
# แบ่งคำจากประโยคที่มีเว้นวรรค
sentence = "I love learning python programming"
words = sentence.split(" ")
print("คำในประโยค:", words)
```

```python
# Join: เอา List มารวมกลับเป็น String
fruits_list = ["apple", "banana", "orange", "grape"]

joined_text = " | ".join(fruits_list)
print("ผลลัพธ์การ Join:", joined_text)
```

## 4. 🎨 String Formatting ฉบับเจาะลึก
นอกจากตัวแปรธรรมดา เรายังสามารถจัดระเบียบตัวเลขและทศนิยมใน f-string ได้

```python
pi = 3.14159265
price = 1500000

print(f"ค่า Pi เอาแค่ 2 ตำแหน่ง: {pi:.2f}")
print(f"ราคาใส่คอมม่า: {price:,.2f} บาท")
```
