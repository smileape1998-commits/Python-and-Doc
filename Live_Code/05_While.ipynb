## 1. 🧠 While Loop คืออะไร?

**While** = วนซ้ำ **ตราบใดที่** เงื่อนไขยังเป็น True

```
      ┌──────────────────────────────┐
      │                              │
 ─→ ┌──────────────┐   True    ┌────────────┐
    │ ตรวจเงื่อนไข │─────────→ │ ทำงาน...  │
    └──────────────┘           └────────────┘
            │ False
            ↓
    ออกจาก Loop
```

โดยทั่วไป While Loop ต้องมี 3 ส่วน:
1. **เริ่มต้น** — กำหนดตัวแปรควบคุม
2. **เงื่อนไข** — เงื่อนไขในการวน
3. **อัพเดท** — เปลี่ยนค่าเพื่อให้วนหยุดได้

> ⚠️ ถ้าลืม ส่วนที่ 3 จะเกิด **Infinite Loop** (วนไม่หยุด!) กด `I` สองครั้งใน Jupyter เพื่อหยุด

## 2. 🔄 While Loop พื้นฐาน

````python
# นับ 1 ถึง 5
count = 1               # ① เริ่มต้น

while count <= 5:
    print("นับ ",count)        # ② ทำงาน
    count += 1          # ③ เพิ่มค่า (อย่าลืม ไม่งั้นจะติดลูปไม่รู้จบ)
print("นับเสร็จแล้ว!")
````

````python
# นับถอยหลัง
countdown = 5

while countdown > 0:
    print("นับถอยหลัง ", countdown)
    countdown -= 1
    
print("นับถอยหลังเสร็จแล้ว!")
````

## 3. 🛑 `break` — หยุด Loop ทันที

ใช้เมื่อต้องการออกจาก Loop ก่อนที่เงื่อนไขจะเป็น False

````python
# ค้นหาค่าใน List
numbers = [3, 7, 2, 9, 5, 1, 8]
target  = 9
i = 0

while i < len(numbers):
    print(f"ดู Index {i} ที่มีค่า {numbers[i]}...")
    if numbers[i] == target:
        print(f"🎯 เจอ {target} ที่ index {i}!")
        break
    i += 1
````

---
## 4. ⏭️ `continue` — ข้ามรอบนั้น

````python
# ข้ามเลขคู่ แสดงเฉพาะเลขคี่
n = 0

while n < 10:
    n += 1
    if n % 2 == 0:
        continue   # ข้ามรอบนี้
    print(n, end=" ")

print()  # ขึ้นบรรทัดใหม่
````

## 5. 🔁 `while True` — Loop ที่หยุดด้วย break

Pattern ที่ใช้บ่อยมาก เมื่อไม่รู้ล่วงหน้าว่าจะวนกี่รอบ

````python
# โปรแกรมเมนู (วนจนกว่าผู้ใช้จะเลือกออก)
while True:
    print("\n" + "=" * 20)
    print("      เมนูหลัก")
    print("=" * 20)
    print("1. ดูข้อมูล")
    print("2. เพิ่มข้อมูล")
    print("3. ออก")
    
    choice = input("\nเลือก (1-3): ")
    
    if choice == "1":
        print("📋 กำลังดูข้อมูล...")
    elif choice == "2":
        print("➕ กำลังเพิ่มข้อมูล...")
    elif choice == "3":
        print("👋 ออกจากโปรแกรม!")
        break
    else:
        print("⚠️ กรุณาเลือก 1, 2, หรือ 3")
````

## 7. ↔️ While vs For — ใช้เมื่อไหร่?

| สถานการณ์ | ใช้ | ตัวอย่าง |
|-----------|-----|----------|
| รู้จำนวนรอบแน่นอน | `for` | วน 10 รอบ |
| วนผ่าน List/Tuple | `for` | `for x in list` |
| เงื่อนไขไม่รู้จำนวนรอบ | `while` | รับ Input จนถูก |
| วนจนกว่าจะหยุด | `while` | เมนู, Game Loop |

````python
# เปรียบเทียบ for vs while สำหรับงานเดียวกัน

# งาน: วน List
fruits = ["แอปเปิ้ล", "มะม่วง", "กล้วย"]

print("-- for loop (แนะนำ) --")
for f in fruits:
    print(f"  {f}")

print("-- while loop (เทียบเท่า) --")
i = 0
while i < len(fruits):
    print(f"  {fruits[i]}")
    i += 1

# สรุป: ถ้าวน List → ใช้ for สั้นกว่า อ่านง่ายกว่า
````

## 8. 🛠️ ตัวอย่างใช้งานจริง

````python
# เกมทายตัวเลข
import random

secret   = random.randint(1, 20)
attempts = 0
max_try  = 5

print(f"🎮 ทายตัวเลข 1-20 (มี {max_try} ครั้ง)")

while attempts < max_try:
    guess = int(input(f"\nครั้งที่ {attempts+1}: "))
    attempts += 1
    
    if guess == secret:
        print(f"🎉 ถูกต้อง! ใช้ {attempts} ครั้ง")
        bonus = max_try - attempts
        print(f"⭐ เหลือสิทธิ์: {bonus} ครั้ง")
        break
    elif guess < secret:
        left = max_try - attempts
        print(f"🔼 น้อยไป! (เหลือ {left} ครั้ง)")
    else:
        left = max_try - attempts
        print(f"🔽 มากไป! (เหลือ {left} ครั้ง)")
else:
    print(f"\n💔 หมดสิทธิ์! เฉลย: {secret}")
````

````python
# ตรวจสอบ Input จนกว่าจะถูกต้อง
while True:
    try:
        age = int(input("อายุของคุณ (1-120): "))
        if 1 <= age <= 120:
            print(f"✅ อายุ {age} ปี บันทึกแล้ว")
            break
        else:
            print("⚠️ กรุณาใส่อายุระหว่าง 1-120")
    except ValueError:
        print("⚠️ กรุณาใส่ตัวเลขเท่านั้น")
````

## 9. ⚠️ ข้อผิดพลาดที่พบบ่อย

````python
# ❌ ผิดที่ 1: ลืมอัพเดทตัวแปร → Infinite Loop!
# count = 0
# while count < 5:
#     print(count)    # วนไม่หยุด! count ไม่เพิ่มขึ้น

# ✅ ถูก
count = 0
while count < 5:
    print(count)
    count += 1   # ← ต้องมีบรรทัดนี้เสมอ!
````

````python
# ❌ ผิดที่ 2: เงื่อนไขไม่มีทางเป็น False
# while True:              # ← ต้องมี break!
#     print("วนไม่หยุด")

# ✅ ถูก
i = 0
while True:
    if i >= 3:
        break        # ← ต้องมี break
    print(i)
    i += 1
````
