## 1. 📦 Tuple คืออะไร?

**Tuple** คือการเก็บข้อมูลหลายค่า **คล้าย List** แต่ **แก้ไขไม่ได้** หลังสร้าง

```
┌─────────────────┬──────────────────────────────────────────┐
│                 │            List              Tuple       │
├─────────────────┼──────────────────────────────────────────┤
│ วงเล็บ          │            [ ]              ( )          │
│ แก้ไขได้        │            ✅ Yes            ❌ No        │
│ ความเร็ว        │            ปกติ             เร็วกว่า     │
│ ใช้เป็น Key Dict│            ❌ ไม่ได้         ✅ ได้        │
│ ใช้เมื่อ         │            ข้อมูลเปลี่ยนได้  ข้อมูลคงที่  │
└─────────────────┴──────────────────────────────────────────┘
```

## 2. 🏗️ การสร้าง Tuple

````python
# การสร้าง Tuple รูปแบบต่างๆ
colors   = ("แดง", "เขียว", "น้ำเงิน")        # ปกติ
point    = (10, 20)                       # พิกัด x, y
empty    = ()                             # ว่าง
no_paren = 1, 2, 3                        # ไม่ต้องมีวงเล็บก็ได้

# ⚠️ Tuple ที่มีสมาชิกเดียว ต้องมีเครื่องหมาย , ตามท้าย!
single_wrong = (42)    # นี่คือ int ธรรมดา!
single_right = (42,)   # นี่คือ Tuple!

print("colors        :", colors, type(colors))
print("point         :", point,  type(point))
print("no_paren      :", no_paren, type(no_paren))
print("single_wrong  :", single_wrong, type(single_wrong))   # int!
print("single_right  :", single_right, type(single_right))   # tuple!
````

````python
# Tuple ไม่สามารถแก้ไขได้
rgb = (255, 128, 0)

# ลองแก้ไข — จะ Error
try:
    rgb[0] = 999
except TypeError as e:
    print("❌ TypeError:", e)

# แต่ถ้า Tuple เก็บ List ไว้ข้างใน List นั้นแก้ได้!
mixed = (1, [2, 3], "สวัสดี")
mixed[1].append(4)  # แก้ไข List ข้างใน
print(mixed)  # (1, [2, 3, 4], "สวัสดี")
````

## 4. 🎁 Tuple Unpacking — ฟีเจอร์เด่นที่สุด!

````python
# Unpacking พื้นฐาน
point = (10, 20)
# x, y = point


# Unpacking สี RGB
# rgb = (255, 128, 0)


# สลับค่า — ไม่ต้องใช้ตัวแปรกลาง!
````

````python
# Star (*) Unpacking
numbers = (1, 2, 3, 4, 5)

first, *rest           = numbers
print("first:", first, "rest:", rest)

*beginning, last       = numbers
print("beginning:", beginning, "last:", last)

first2, *middle, last2 = numbers
print("first:", first2, "middle:", middle, "last:", last2)
````

````python
# Unpacking ใน for loop
students = [
    ("แพร",  20, 3.75),
    ("บอย",  22, 3.20),
    ("มิ้ว", 21, 2.90)
]

print(f"{'ชื่อ':8} {'อายุ':6} {'GPA':6}")
print("-" * 22)
for name, age, gpa in students:   # Unpack ตาม for loop
    print(f"{name:8} {age:<6} {gpa:<6}")
````

## 6. 💡 เมื่อไหร่ควรใช้ Tuple?

````python
# กรณี 1: ข้อมูลคงที่ ไม่ควรเปลี่ยน
SCREEN_RES  = (1920, 1080)    # ความละเอียดหน้าจอ
MY_BIRTHDAY = (2004, 8, 25)   # ปี, เดือน, วัน
THAI_COLORS = ("แดง", "ขาว", "น้ำเงิน")  # ธงชาติ

print("Screen:", SCREEN_RES)

# กรณี 2: Tuple เป็น Key ใน Dictionary ได้ (List ทำไม่ได้)
locations = {
    (13.75, 100.50): "กรุงเทพ",
    (18.79, 98.98) : "เชียงใหม่",
    (7.88,  98.40) : "ภูเก็ต"
}
print("พิกัด (13.75, 100.50) คือ:", locations[(13.75, 100.50)])

# กรณี 3: Function Return หลายค่า
def divide(a, b):
    quotient  = a // b    # ผลหาร
    remainder = a % b     # เศษ
    return quotient, remainder   # return tuple!

q, r = divide(17, 5)
print(f"17 ÷ 5 → ผลหาร: {q}, เศษ: {r}")
````

## 7. ⚠️ ข้อผิดพลาดที่พบบ่อย

````python
# ❌ ผิดที่ 1: ลืม comma สำหรับ tuple ตัวเดียว
wrong = (42)     # int ธรรมดา
right = (42,)    # tuple!
print(type(wrong), "|", type(right))

# ❌ ผิดที่ 2: Unpack ผิดจำนวน
pair = (10, 20)
try:
    a, b, c = pair   # มีแค่ 2 ค่า !
except ValueError as e:
    print("❌ ValueError:", e)

# ✅ ถูก
a, b = pair
print("✅ a, b =", a, b)
````
