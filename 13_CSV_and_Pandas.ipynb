## 1. 📄 CSV คืออะไร?

**CSV (Comma-Separated Values)** คือรูปแบบไฟล์ text ที่เก็บข้อมูลตาราง โดยใช้ `,` คั่นระหว่างคอลัมน์

```
name,age,score
Alice,30,90
Bob,25,85
Charlie,35,78
```

เป็นรูปแบบที่ใช้แลกเปลี่ยนข้อมูลกันมากที่สุด — Excel, Google Sheets, Database ต่างก็ export เป็น CSV ได้

---
## 2. 📂 อ่าน/เขียน CSV ด้วย `csv` module

`csv` เป็น built-in module ไม่ต้องติดตั้งเพิ่ม

````python
import csv

# สร้างไฟล์ CSV ตัวอย่างก่อน
sample_data = [
    ["name",    "age", "score", "city"],
    ["Alice",    30,    90,     "Bangkok"],
    ["Bob",      25,    85,     "Chiang Mai"],
    ["Charlie",  35,    78,     "Phuket"],
    ["Diana",    28,    92,     "Bangkok"],
    ["Eve",      22,    65,     "Chiang Mai"],
]

with open("students.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(sample_data)   # เขียนทีละหลายแถว

print("✅ สร้างไฟล์ students.csv เรียบร้อย")
````

````python
# อ่าน CSV ด้วย csv.reader
print("--- อ่านไฟล์ students.csv ---")
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)   # แต่ละ row เป็น List
````

````python
# อ่าน CSV ด้วย csv.DictReader — ได้เป็น Dict (สะดวกกว่า)
print("--- อ่านแบบ DictReader ---")
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)   # แถวแรกเป็น header อัตโนมัติ
    for row in reader:
        print(f"{row['name']:10} อายุ {row['age']} คะแนน {row['score']}")
````

````python
# เขียน CSV ด้วย csv.DictWriter
new_students = [
    {"name": "Frank",  "age": 27, "score": 88, "city": "Bangkok"},
    {"name": "Grace",  "age": 31, "score": 95, "city": "Phuket"},
]

with open("new_students.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["name", "age", "score", "city"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()          # เขียน header
    writer.writerows(new_students)

print("✅ สร้างไฟล์ new_students.csv เรียบร้อย")
````

## 6. 🔍 ดูและสำรวจข้อมูล

````python
import pandas as pd

data = {
    "name":  ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "age":   [30, 25, 35, 28, 22],
    "score": [90, 85, 78, 92, 65],
    "city":  ["Bangkok", "Chiang Mai", "Phuket", "Bangkok", "Chiang Mai"]
}
df = pd.DataFrame(data)

print("--- head() : 3 แถวแรก ---")
print(df.head(3))

print("\n--- tail() : 2 แถวสุดท้าย ---")
print(df.tail(2))

print("\n--- shape : (แถว, คอลัมน์) ---")
print(df.shape)

print("\n--- columns ---")
print(df.columns.tolist())
````

````python
# info() — ดูชนิดข้อมูลและค่าว่าง
df.info()
````

````python
# describe() — สถิติพื้นฐานคอลัมน์ตัวเลข
df.describe()
````

---
## 7. 🎯 เลือกและกรองข้อมูล

````python
# เลือกคอลัมน์
print("--- เลือก 1 คอลัมน์ (ได้ Series) ---")
print(df["name"])

print("\n--- เลือกหลายคอลัมน์ (ได้ DataFrame) ---")
print(df[["name", "score"]])
````

````python
# เลือกแถว
print("--- iloc : เลือกด้วย index ตัวเลข ---")
print(df.iloc[0])        # แถวที่ 0
print()
print(df.iloc[1:3])      # แถว 1-2

print("\n--- loc : เลือกด้วยชื่อ ---")
print(df.loc[0, "score"])    # แถว 0 คอลัมน์ score
````

````python
# กรองข้อมูล (Filtering)
print("--- คะแนน >= 85 ---")
print(df[df["score"] >= 85])

print("\n--- อยู่ Bangkok และคะแนน > 80 ---")
print(df[(df["city"] == "Bangkok") & (df["score"] > 80)])
````

---
## 8. 📈 การคำนวณสถิติ

````python
# สถิติพื้นฐาน
print(f"คะแนนเฉลี่ย : {df['score'].mean():.2f}")
print(f"คะแนนสูงสุด : {df['score'].max()}")
print(f"คะแนนต่ำสุด : {df['score'].min()}")
print(f"ผลรวมคะแนน  : {df['score'].sum()}")
print(f"จำนวนนักเรียน: {len(df)}")

# นับจำนวนตามกลุ่ม
print("\n--- จำนวนนักเรียนตามเมือง ---")
print(df["city"].value_counts())
````

````python
# groupby — สรุปข้อมูลตามกลุ่ม
print("--- คะแนนเฉลี่ยตามเมือง ---")
print(df.groupby("city")["score"].mean())

print("\n--- สถิติคะแนนตามเมือง ---")
print(df.groupby("city")["score"].agg(["mean", "max", "min", "count"]))
````

## 9. ✏️ แก้ไขข้อมูล

````python
# เพิ่มคอลัมน์ใหม่
df["grade"] = df["score"].apply(lambda s: "A" if s >= 90 else ("B" if s >= 80 else "C"))
print(df)

# เรียงข้อมูล
print("\n--- เรียงตามคะแนน มากไปน้อย ---")
print(df.sort_values("score", ascending=False))
````

````python
# จัดการค่าว่าง (NaN)
import numpy as np

# จำลอง DataFrame ที่มีค่าว่าง
df_nan = pd.DataFrame({
    "name":  ["Alice", "Bob",  "Charlie", "Diana"],
    "score": [90,       None,   78,         None]
})

print("ก่อนแก้ค่าว่าง:")
print(df_nan)

print("\nมีค่าว่างไหม?")
print(df_nan.isnull().sum())       # นับจำนวน NaN แต่ละคอลัมน์

# แทนค่าว่างด้วยค่าเฉลี่ย
mean_score = df_nan["score"].mean()
df_nan["score"] = df_nan["score"].fillna(mean_score)
print(f"\nหลังเติมค่าว่างด้วยค่าเฉลี่ย ({mean_score}):")
print(df_nan)
````
