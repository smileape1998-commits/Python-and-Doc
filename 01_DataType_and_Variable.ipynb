## 1. ตัวแปร (Variable) คืออะไร?

เปรียบเทียบ **ตัวแปร** เสมือน **กล่อง** ที่มีการติดป้ายชื่อระบุเอาไว้ เพื่อใช้สำหรับจัดเก็บข้อมูล และสามารถเรียกใช้งานหรืออ้างอิงถึงข้อมูลนั้นได้ผ่านชื่อที่ตั้งไว้

```text
        name
       ┌─────────┐
       │  "Ploy" │  ← ค่าที่จัดเก็บอยู่
       └─────────┘
```

**รูปแบบการสร้างและกำหนดค่าตัวแปร:**
```python
ชื่อตัวแปร = ค่าข้อมูล
```

*หมายเหตุ:* เครื่องหมาย `=` ในภาษา Python ไม่ได้มีความหมายว่า "เท่ากับ" ทางคณิตศาสตร์ แต่หมายถึง **การกำหนดค่า** (Assignment) ให้นำข้อมูลด้านขวาไปเก็บไว้ในตัวแปรด้านซ้าย

---
### กฎการตั้งชื่อตัวแปร

| ตัวอย่างที่ถูกต้อง | ตัวอย่างที่ผิด | เหตุผล |
|--------------------|----------------|--------|
| `name` | `1name` | ห้ามขึ้นต้นด้วยตัวเลข |
| `my_score` | `my score` | ห้ามมีช่องว่าง |
| `_value` | `if` | ห้ามใช้คำสงวน (Keyword) ของภาษา Python |
| `score2` | `score-2` | ห้ามใช้เครื่องหมายขีดกลาง (`-`) หรือสัญลักษณ์พิเศษอื่นนอกจาก `_` |

> **ข้อแนะนำ (Best Practice):**
> ควรตั้งชื่อตัวแปรในรูปแบบ `snake_case` ซึ่งเป็นการใช้ตัวอักษรพิมพ์เล็กทั้งหมดและคั่นแต่ละคำด้วยเครื่องหมายขีดล่าง (`_`)
> ตัวอย่างเช่น `first_name`, `total_score`, `is_student`

````python
print("My name is Python")
print("Python is a programming language")
print("Python want to work quickly")
````

````python
# การสร้างตัวแปรและกำหนดค่า
name = "Python"
age = 30
# แสดงผล
print("My name is " + name)
print("My name is " + name + " and I am " + str(age) + " years old.")

# f-string
print(f"My name is {name} and I am {age} years old")
````

````python
# สามารถกำหนดหลายตัวแปรพร้อมกันได้
x, y, z = 10, 20, 30
print(x, y, z)

# กำหนดหลายตัวแปรให้ค่าเดียวกัน
a = b = c = 0
print(a, b, c)

# เปลี่ยนค่าตัวแปรได้ตลอดเวลา
score = 80
print(score)
score = 10
print(score)
````

## 2. ชนิดข้อมูลหลัก (Data Types)

ภาษา Python มีชนิดข้อมูลพื้นฐาน 3 ประเภทที่สำคัญ ดังนี้:

### 1. กลุ่มไม่มีค่า (No Value)
* **None Type:** คือค่า `None` ใช้แทนข้อมูลที่ไม่มีค่าหรือว่างเปล่า

---


````python
gender = None # NoneType
````


### 2. กลุ่มข้อมูลเดี่ยว (Single Values / Primitive Data Types)
* **Integer (`int`):** ตัวเลขจำนวนเต็ม เช่น `10`, `-5`
* **Float:** ตัวเลขที่มีทศนิยม เช่น `3.14`, `2.5`
* **String (`str`):** ข้อความที่อยู่ในเครื่องหมายคำพูดเดี่ยว `' '` หรือคู่ `" "` เช่น `'Hello'`
* **Boolean (`bool`):** ค่าตรรกะที่มีได้เพียง `True` หรือ `False` (ใช้ในการตัดสินใจ/เงื่อนไข)
---


````python
# Data Types
x = 10 # int    
y = 3.14 # float
z = "Hello" # str
h = "123" # str
is_student = True # bool
is_passed = False # bool
a = "" # str - blank
b = " " # str - empty space
````

````python
# ตัวดำเนินการทางคณิตศาสตร์
print("บวก (+) :", 10 + 3)    # 13
print("ลบ  (-) :", 10 - 3)    # 7
print("คูณ (*) :", 10 * 3)    # 30
print("หาร (/) :", 10 / 3)    # 3.333... (ได้ float เสมอ)
print("หารเต็ม  :", 10 // 3)  # 3  (Floor Division — ตัดทศนิยม)
print("เศษหาร   :", 10 % 3)   # 1  (Modulo)
print("ยกกำลัง  :", 2 ** 10)  # 1024

# ตัวดำเนินการย่อ
n = 5
n += 3  # n = n + 3
print(n)  # 8
````

### float — ทศนิยม

````python
# ⚠️ ระวัง: float มีความคลาดเคลื่อนอยู่บ้าง (เรื่องปกติในคอมพิวเตอร์)
pi = 3.1415
price = 19.99
print(pi)
print(price)
# ปัดเศษ
print("Round pi:", round(3.1415111, 2))  # ปัดเป็นทศนิยม 2 ตำแหน่ง
````

### str — ข้อความ

````python
# str เป็น Sequence — เข้าถึงแต่ละตัวอักษรด้วย index ได้
word = "Python"
# 0 1 2 3 4 5
# P y t h o n
# -6-5-4-3-2-1

print("Index 0:", word[0])  # P
print("Index 1:", word[1])  # y
print("Index -2:", word[-2])  # o
print("Index - 1", word[-1])

# ตรวจสอบว่ามีข้อความที่ต้องการอยู่ไหม
print(" 'Py' in word :","Py" in word)
````

### bool — จริง / เท็จ

> 💡 `True` และ `False` **ต้องขึ้นต้นด้วยตัวใหญ่** เสมอ!

````python
is_raining = False
has_umbrella = True
# ผลจากการเปรียบเทียบจะเป็น bool
print("Is it raining?", is_raining)
print("Do I have an umbrella?", has_umbrella)

# bool มีค่าตัวเลขด้วย
print("10 > 5 ", 10 > 5)  # True
print("6 < 3  ", 6 < 3)  # False
````


### 3. กลุ่มข้อมูลหลายค่า (Multiple Values / Data Structures)
ใช้สำหรับเก็บข้อมูลหลายๆ ค่าไว้ในตัวแปรเดียว ได้แก่:
* **List:** เก็บข้อมูลเป็นลำดับ แก้ไขได้ (ใช้ `[ ]`)
* **Tuple:** คล้าย List แต่แก้ไขไม่ได้ (Immutable) เพื่อความปลอดภัยของข้อมูล (ใช้ `( )`)
* **Set:** เก็บข้อมูลที่ไม่ซ้ำกัน (ไม่มีลำดับที่แน่นอน) (ใช้ `{ }`)
* **Dictionary (`dict`):** เก็บข้อมูลแบบคู่ Key: Value เหมาะสำหรับการจัดระเบียบข้อมูลที่มีป้ายกำกับ (ใช้ `{ }`)

## 3. 🔍 การตรวจสอบและแปลงชนิดข้อมูล

````python
# type() — ตรวจสอบชนิดข้อมูล
values = [43 , 3.14 , "Hello", True]
for x in values:
    print(x, "is of type", type(x).__name__) # __name__ คือชื่อของชนิดข้อมูล 'str'
````

````python
# การแปลงชนิดข้อมูล (Type Casting)
print("--- แปลงเป็น int ---")
print(int("42"))      # str → int : 42
print(int(3.99))      # float → int : 3  (ตัดทศนิยม ไม่ปัดเศษ!)
print(int(True))      # bool → int : 1

print("\n--- แปลงเป็น float ---")
print(float("3.14"))  # str → float
print(float(5))       # int → float : 5.0

print("\n--- แปลงเป็น str ---")
print(str(42))        # int → str
print(str(3.14))      # float → str
print("อายุ: " + str(20))  # ต้องแปลงก่อนรวมกับ str

print("\n--- แปลงเป็น bool ---")
print(bool(0))        # 0 → False
print(bool(1))        # 1 → True
print(bool(""))       # str ว่าง → False
print(bool("hello"))  # str ไม่ว่าง → True
````

## 4. ข้อผิดพลาดที่พบบ่อย (Common Mistakes)

````python
# ❌ ผิดที่ 1: ใช้ = แทน == เปรียบเทียบ
# x = 5
# if x = 5:   # SyntaxError!
#     print("เท่ากัน")

# ✅ ถูก:
x = 5
if x == 5:
    print("✅ ใช้ == เปรียบเทียบ")
````

````python
# ❌ ผิดที่ 2: รวม str กับ int โดยตรง
age = 20
try:
    print("อายุ: " + age)   # TypeError!
except TypeError as e:
    print("❌ Error:", e)

# ✅ ถูก:
print("อายุ: " + str(age))   # แปลงเป็น str ก่อน
print(f"อายุ: {age}")         # หรือใช้ f-string
````

````python
# ❌ ผิดที่ 3: ลืมแปลง input() เป็นตัวเลขก่อนคำนวณ
score_str = "85"   # จำลองว่ารับจาก input()
bonus     = 5

try:
    print(score_str + bonus)   # TypeError!
except TypeError as e:
    print("❌ Error:", e)

# ✅ ถูก:
print("✅ ผลลัพธ์:", int(score_str) + bonus)   # 90
````
